```python
def read_sheet(path, cols):
    return pd.read_excel(path, sheet_name=0, header=None, skiprows=1) \
             .iloc[:, :len(cols)].set_axis(cols, axis=1)

def parse_std(s):
    s = str(s).replace(',', '.').strip()
    m = re.search(r'([\d.]+)\s*±\s*([\d.]+)', s)
    if m: return float(m.group(1)), float(m.group(2))
    try: return float(s), float(s) * 0.07
    except: return np.nan, np.nan

def load_and_prepare(data_dir='data'):
    c6 = ['Round','ML','Rank','No.','ΦHEA','ΦBA','ΦCBEA','ΦATAC','ΦPEA','ΦAAm','Fa_max','Fa_ave','d','Q']
    c5 = ['No.','ΦHEA','ΦBA','ΦCBEA','ΦATAC','ΦPEA','ΦAAm','Fa_max','Fa_ave','d','Q','G','Tanδ','k']
    cc = ['ΦHEA','ΦBA','ΦCBEA','ΦATAC','ΦPEA','ΦAAm']

    df = pd.concat([
        read_sheet(f'{data_dir}/41586_2025_9269_MOESM5_ESM.xlsx', c5),
        read_sheet(f'{data_dir}/41586_2025_9269_MOESM6_ESM.xlsx', c6)
    ], ignore_index=True)[cc + ['Fa_max','Fa_ave']]

    for c in cc + ['Fa_max']:
        df[c] = pd.to_numeric(df[c].astype(str).str.replace(',', '.'), errors='coerce')

    df = df.dropna(subset=cc)
    df = df[~df[cc].round(4).duplicated()].reset_index(drop=True)

    df[cc] = df[cc].div(df[cc].sum(axis=1), axis=0)

    parsed = df['Fa_ave'].apply(parse_std)
    df['Fa_mean'] = parsed.apply(lambda x: x[0])
    df['Fa_std']  = parsed.apply(lambda x: x[1])

    mask = df['Fa_mean'].isna() & df['Fa_max'].notna()
    df.loc[mask, 'Fa_mean'] = df.loc[mask, 'Fa_max']
    df.loc[mask, 'Fa_std']  = df.loc[mask, 'Fa_mean'] * 0.05

    df = df[df['Fa_mean'] > 0].copy()
    return df.reset_index(drop=True), cc

mon_cols = ['ΦHEA','ΦBA','ΦCBEA','ΦATAC','ΦPEA','ΦAAm']
df, comp_cols = load_and_prepare()

#import data 
adhesion = df['Fa_mean'].values  
fractions_base_line = df.iloc[:, 0:6].to_numpy()
```

```python
monomers = {
    'HEA': 'C=CC(=O)OCCO',                            # 2-Hydroxyethyl acrylate
    'BA': 'C=CC(=O)OCCCC',                            # Butyl acrylate
    'CBEA': 'C=CC(=O)OCCC(=O)O',                      # Carboxyethyl acrylate
    'ATAC': 'C=CC(=O)OCC[N+](C)(C)C',                 # [2-(Acryloyloxy)ethyl]trimethylammonium
    'PEA': 'C=CC(=O)OCCOc1ccccc1',                    # 2-Phenoxyethyl acrylate
    'AAm': 'C=CC(=O)N'                                # Acrylamide
}
mon_names = list(monomers.keys())
mon_smiles = list(monomers.values())
mon_mols = [Chem.MolFromSmiles(s) for s in mon_smiles]
```

```python
def custom_2d_autocorr_for_metrics(mol, lags=8):
    if mol is None or mol.GetNumAtoms() < 2:
        return np.zeros(3 * lags, dtype=np.float32)
    mol_calc = Chem.Mol(mol)
    rdPartialCharges.ComputeGasteigerCharges(mol_calc)
    dm = rdmolops.GetDistanceMatrix(mol_calc)
    contribs = rdMolDescriptors._CalcCrippenContribs(mol_calc)
    logp_arr = np.array([c[0] for c in contribs], dtype=np.float32)
    charge_arr = np.array([float(a.GetProp('_GasteigerCharge')) if a.HasProp('_GasteigerCharge') else 0.0 for a in mol_calc.GetAtoms()], dtype=np.float32)
    arom_arr = np.array([1.0 if atom.GetIsAromatic() or atom.GetHybridization() == Chem.HybridizationType.SP2 else 0.0 for atom in mol_calc.GetAtoms()], dtype=np.float32)
    prop_arrays = [logp_arr, charge_arr, arom_arr]
    autocorr_vec = []
    for prop_arr in prop_arrays:
        for lag in range(1, lags + 1):
            mask = (dm == lag)
            upper_mask = np.triu(mask, k=1)
            i_idx, j_idx = np.where(upper_mask)
            if len(i_idx) == 0:
                autocorr_vec.append(0.0)
            else:
                autocorr_vec.append(np.sum(prop_arr[i_idx] * prop_arr[j_idx]) / len(i_idx))
    return np.array(autocorr_vec, dtype=np.float32)

sizes = [2048, 16384, 1024, 512, 256, 128]
fps_data = []
for s in sizes:
    gen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=s, atomInvariantsGenerator=rdFingerprintGenerator.GetMorganFeatureAtomInvGen())
    bin_arr = np.array([list(gen.GetFingerprint(mol)) for mol in mon_mols], dtype=np.int8)
    cnt_arr = np.array([gen.GetCountFingerprintAsNumPy(mol) for mol in mon_mols], dtype=np.uint16)
    fps_data.append((bin_arr, cnt_arr))
(ffp_arrays, cffp_arrays), (ffp_array, cffp_array), (ffp_arrays1, cffp_arrays1), (ffp_arrays2, cffp_arrays2), (ffp_arrays3, cffp_arrays3), (ffp_arrays4, cffp_arrays4) = fps_data

custom_feats = [custom_2d_autocorr_for_metrics(mol, lags=8) for mol in mon_mols]
rdkit_feats = [rdMolDescriptors.CalcAUTOCORR2D(mol) for mol in mon_mols]
merged = [np.concatenate([c, r]) for c, r in zip(custom_feats, rdkit_feats)]
autocorr_2d_arrays = np.array(merged, dtype=np.float32)


val_df = pd.DataFrame({
    'autocorr': list(autocorr_2d_arrays),
    'cffp_256': list(cffp_arrays3),
}, index=list(monomers.keys()))
```

```python
comp = df[mon_cols].values.astype(np.float64)
row_sums = comp.sum(axis=1, keepdims=True)
comp = np.where(row_sums > 1.5, comp / 100.0, comp)
row_sums = comp.sum(axis=1, keepdims=True)
row_sums[row_sums == 0] = 1.0
comp /= row_sums
valid_mask = comp.sum(axis=1) > 0
valid_comps = comp[valid_mask]
n_poly_valid = int(valid_mask.sum())

np.random.seed(42)
N_mon, batch_size = 1000, 256

def flush_batch(datasets, buffers, count, size):
    for name in datasets:
        datasets[name][count:count+size] = buffers[name][:size]

with h5py.File('./data/DL/small_poly_datasets.h5', 'w') as f:
    hdf_datasets, buffers = {}, {}
    for name, (src, dtype) in features_source.items():
        shape_3d = (n_poly_valid, N_mon, src.shape[1])
        chunk_shape = (min(batch_size, n_poly_valid), N_mon, src.shape[1])
        hdf_datasets[name] = f.create_dataset(
            name, shape=shape_3d, dtype=dtype,
            compression="gzip", compression_opts=4,
            chunks=chunk_shape, shuffle=True
        )
        buffers[name] = np.empty((batch_size, N_mon, src.shape[1]), dtype=dtype)

    buf_idx = saved_count = 0
    for fracs in valid_comps:
        exact = fracs * N_mon
        counts = np.floor(exact).astype(np.int32)
        diff = N_mon - counts.sum()
        if diff > 0:
            rem_indices = np.argsort(exact % 1)[::-1]
            counts[rem_indices[:diff]] += 1
        global_chain = np.repeat(mon_indices, counts)
        np.random.shuffle(global_chain)
        for name, (src, _) in features_source.items():
            buffers[name][buf_idx] = src[global_chain]
        buf_idx += 1
        if buf_idx == batch_size:
            flush_batch(hdf_datasets, buffers, saved_count, batch_size)
            saved_count += batch_size
            buf_idx = 0
            sys.stdout.write(f"typed:{saved_count}/{n_poly_valid}...\r")
            sys.stdout.flush()
    if buf_idx > 0:
        flush_batch(hdf_datasets, buffers, saved_count, buf_idx)
        saved_count += buf_idx
        sys.stdout.write(f"typed: {saved_count}/{n_poly_valid}...\r")
        sys.stdout.flush()

print(f"\n N:{n_poly_valid} in 'small_poly_datasets.h5'!")
```

```python
with h5py.File('./data/DL/Deep_poly_datasets_Tr.h5', 'w') as f:
    dset_ffp = f.create_dataset('cffp_256', shape=(0, SEQ_LEN, ffp_len), maxshape=(None, SEQ_LEN, ffp_len), dtype=np.float32, compression='gzip', compression_opts=4)
    dset_auto = f.create_dataset('autocorr', shape=(0, SEQ_LEN, auto_len), maxshape=(None, SEQ_LEN, auto_len), dtype=np.float32, compression='gzip', compression_opts=4)

    for start in range(0, TOTAL_SAMPLES, BATCH_SIZE):
        end = min(start + BATCH_SIZE, TOTAL_SAMPLES)
        batch_size = end - start
        ffp_chunk = np.empty((batch_size, SEQ_LEN, ffp_len), dtype=np.float32)
        auto_chunk = np.empty((batch_size, SEQ_LEN, auto_len), dtype=np.float32)
        for i in range(batch_size):
            k = random.randint(1, 7)
            idxs = random.sample(range(num_monomers), k)
            r = np.random.dirichlet(np.ones(k))
            picks = np.random.choice(idxs, size=SEQ_LEN, p=r)
            ffp_chunk[i] = [ffp_list[j] for j in picks]
            auto_chunk[i] = [auto_list[j] for j in picks]
        dset_ffp.resize(dset_ffp.shape[0] + batch_size, axis=0)
        dset_auto.resize(dset_auto.shape[0] + batch_size, axis=0)
        dset_ffp[-batch_size:] = ffp_chunk
        dset_auto[-batch_size:] = auto_chunk
        print(f"Gen: {end} from {TOTAL_SAMPLES}")
print("Deep_poly_datasets_Tr.h5")
```

```python
scaler = StandardScaler()
X_autocorr = scaler.fit_transform(autocorr_2d_arrays)
X_ffp = cffp_arrays3.astype(np.float32)
X_combined = np.concatenate([X_ffp, X_autocorr], axis=1)

families = valid_df['family'].values
numbers = valid_df['number'].values

unique_families = np.unique(families)
palette = sns.color_palette('tab10', len(unique_families))
fam2col = dict(zip(unique_families, palette))
colors = [fam2col[f] for f in families]

def plot_pca_tsne_grid(X_list, titles, numbers, colors, families, fam2col):
    fig, axes = plt.subplots(2, 3, figsize=(24,14))
    for col, (X, ttl) in enumerate(zip(X_list, titles)):
        for row, (transform, name) in enumerate(zip(
            [PCA(n_components=2, random_state=42), TSNE(n_components=2, random_state=42, perplexity=10)],
            ['PCA', 't-SNE'])):
            X_2d = transform.fit_transform(X)
            ax = axes[row, col]
            ax.scatter(X_2d[:,0], X_2d[:,1], c=colors, edgecolors='k', s=80, alpha=0.7)
            for i, num in enumerate(numbers):
                ax.annotate(str(num), (X_2d[i,0], X_2d[i,1]), textcoords="offset points", xytext=(4,4), fontsize=7, alpha=0.8)
            ax.set_title(f'{name}: {ttl}')
            if row == 0:
                ax.set_xlabel(f'PC1 ({transform.explained_variance_ratio_[0]:.1%})')
                ax.set_ylabel(f'PC2 ({transform.explained_variance_ratio_[1]:.1%})')
            else:
                ax.set_xlabel('t-SNE 1')
                ax.set_ylabel('t-SNE 2')
    handles = [plt.Line2D([0],[0], marker='o', color='w', markerfacecolor=fam2col[f], markersize=8, label=f) for f in unique_families]
    fig.legend(handles=handles, title='Family', loc='upper right', bbox_to_anchor=(1.12,0.95))
    plt.tight_layout()
    plt.show()

X_list = [X_autocorr, X_ffp, X_combined]
titles = ['2D Autocorrelation', 'Functional FPs', 'Combined (FFP + Autocorr)']
plot_pca_tsne_grid(X_list, titles, numbers, colors, families, fam2col)

dist_combined = pdist(X_combined, metric='euclidean')
Z = linkage(dist_combined, method='ward')
plt.figure(figsize=(14,7))
dendrogram(Z, labels=[str(n) for n in numbers], leaf_rotation=90, leaf_font_size=10,
           color_threshold=0, above_threshold_color='gray', link_color_func=lambda k: 'gray')
ax = plt.gca()
for lbl in ax.get_xmajorticklabels():
    idx = np.where(numbers == int(lbl.get_text()))[0][0]
    lbl.set_color(fam2col[families[idx]])
plt.title('Hierarchical Clustering Dendrogram (Combined Features, Ward linkage)')
plt.ylabel('Euclidean Distance')
plt.tight_layout()
plt.show()
```

# DeepLearning Part

#### Model 

```python
class RoPE(nn.Module):
    def __init__(self, dim, max_len=2048, base=10000.0):
        super().__init__()
        inv = 1.0 / (base ** (torch.arange(0, dim, 2).float() / dim))
        pos = torch.arange(max_len).float().unsqueeze(1)
        ang = pos * inv
        self.register_buffer('cos', ang.cos().unsqueeze(0))
        self.register_buffer('sin', ang.sin().unsqueeze(0))

    def forward(self, x):
        n = x.size(1)
        x1, x2 = x[..., 0::2], x[..., 1::2]
        c, s = self.cos[:, :n], self.sin[:, :n]
        return torch.cat([x1*c - x2*s, x1*s + x2*c], dim=-1)
```

```python
class TransformerVAE_CLS(nn.Module):
    def __init__(self, ffp_dim=256, auto_dim=216, seq_len=1000, d_model=128, nhead=4, num_layers=3, latent_dim=64, dropout=0.1):
        super().__init__()
        self._apply_rope = RoPE(dim=d_model, max_len=1024)
        self.seq_len = seq_len
        self.d_model = d_model
        self.ffp_projection = nn.Linear(ffp_dim, d_model//2)
        self.auto_projection = nn.Linear(auto_dim, d_model//2)
        self.feature_fusion = nn.Linear(d_model, d_model)
        self.ffp_norm = nn.LayerNorm(d_model//2)
        self.auto_norm = nn.LayerNorm(d_model//2)
        self.feature_norm = nn.LayerNorm(d_model)
        self.cls_token = nn.Parameter(torch.randn(1, 1, d_model))
        encoder_layer = nn.TransformerEncoderLayer(d_model=d_model, nhead=nhead, dim_feedforward=d_model*4, dropout=dropout, batch_first=True, norm_first=True, activation=F.mish)
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.fc_mu = nn.Linear(d_model, latent_dim)
        self.fc_logvar = nn.Linear(d_model, latent_dim)
        self.fc_decoder_input = nn.Linear(latent_dim, d_model)
        self.decoder_queries = nn.Parameter(torch.randn(1, seq_len, d_model))
        decoder_layer = nn.TransformerDecoderLayer(d_model=d_model, nhead=nhead, dim_feedforward=d_model*4, dropout=dropout, batch_first=True, norm_first=True, activation=F.mish)
        self.transformer_decoder = nn.TransformerDecoder(decoder_layer, num_layers=num_layers)
        self.output_ffp = nn.Sequential(nn.Linear(d_model, ffp_dim), nn.Softplus())
        self.output_auto = nn.Linear(d_model, auto_dim) 

    def encode(self, ffp, auto):
        b = ffp.size(0)
        h_ffp = F.mish(self.ffp_norm(self.ffp_projection(ffp)))
        h_auto = F.mish(self.auto_norm(self.auto_projection(auto)))
        x = self.feature_fusion(torch.cat([h_ffp, h_auto], dim=-1))
        x = self.feature_norm(x)
        x = self._apply_rope(x)
        x = torch.cat([self.cls_token.expand(b, -1, -1), x], dim=1)
        x = self.transformer_encoder(x)
        cls_out = x[:, 0, :]
        return self.fc_mu(cls_out), self.fc_logvar(cls_out)

    def reparameterize(self, mu, logvar):
        if not self.training:
            return mu
        std = torch.exp(0.5*logvar)
        eps = torch.randn_like(std)
        return mu + eps*std

    def decode(self, z):
        b = z.size(0)
        memory = self.fc_decoder_input(z).unsqueeze(1)
        tgt = self.decoder_queries.expand(b, -1, -1)
        tgt = self._apply_rope(tgt)
        out = self.transformer_decoder(tgt=tgt, memory=memory)
        return self.output_ffp(out), self.output_auto(out)

    def forward(self, ffp, auto):
        mu, logvar = self.encode(ffp, auto)
        z = self.reparameterize(mu, logvar)
        recon_ffp, recon_auto = self.decode(z)
        return recon_ffp, recon_auto, mu, logvar
```

```python
class TransformerVAE_SET(nn.Module):
    def __init__(self, ffp_dim=256, auto_dim=216, seq_len=1000, d_model=128, nhead=4, num_layers=3, latent_dim=64, dropout=0.1):
        super().__init__()
        self.seq_len = seq_len
        self.d_model = d_model
        self.ffp_projection = nn.Linear(ffp_dim, d_model//2)
        self.auto_projection = nn.Linear(auto_dim, d_model//2)
        self.feature_fusion = nn.Linear(d_model, d_model)
        self.ffp_norm = nn.LayerNorm(d_model//2)
        self.auto_norm = nn.LayerNorm(d_model//2)
        self.feature_norm = nn.LayerNorm(d_model)
        self.cls_token = nn.Parameter(torch.randn(1, 1, d_model))
        encoder_layer = nn.TransformerEncoderLayer(d_model=d_model, nhead=nhead, dim_feedforward=d_model*4, dropout=dropout, batch_first=True, norm_first=True, activation=F.mish)
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.fc_mu = nn.Linear(d_model, latent_dim)
        self.fc_logvar = nn.Linear(d_model, latent_dim)
        self.fc_decoder_input = nn.Linear(latent_dim, d_model)
        self.decoder_queries = nn.Parameter(torch.randn(1, seq_len, d_model))
        decoder_layer = nn.TransformerDecoderLayer(d_model=d_model, nhead=nhead, dim_feedforward=d_model*4, dropout=dropout, batch_first=True, norm_first=True, activation=F.mish)
        self.transformer_decoder = nn.TransformerDecoder(decoder_layer, num_layers=num_layers)
        self.output_ffp = nn.Sequential(nn.Linear(d_model, ffp_dim), nn.Softplus())
        self.output_auto = nn.Linear(d_model, auto_dim)

    def encode(self, ffp, auto):
        b = ffp.size(0)
        h_ffp = F.mish(self.ffp_norm(self.ffp_projection(ffp)))
        h_auto = F.mish(self.auto_norm(self.auto_projection(auto)))
        x = self.feature_fusion(torch.cat([h_ffp, h_auto], dim=-1))
        x = self.feature_norm(x)
        x = torch.cat([self.cls_token.expand(b, -1, -1), x], dim=1)
        x = self.transformer_encoder(x)
        cls_out = x[:, 0, :]
        return self.fc_mu(cls_out), self.fc_logvar(cls_out)

    def reparameterize(self, mu, logvar):
        if not self.training:
            return mu
        std = torch.exp(0.5*logvar)
        eps = torch.randn_like(std)
        return mu + eps*std

    def decode(self, z):
        b = z.size(0)
        memory = self.fc_decoder_input(z).unsqueeze(1)
        tgt = self.decoder_queries.expand(b, -1, -1)
        out = self.transformer_decoder(tgt=tgt, memory=memory)
        return self.output_ffp(out), self.output_auto(out)

    def forward(self, ffp, auto):
        mu, logvar = self.encode(ffp, auto)
        z = self.reparameterize(mu, logvar)
        recon_ffp, recon_auto = self.decode(z)
        return recon_ffp, recon_auto, mu, logvar
```

```python
class VAE_DeepSets(nn.Module):
    def __init__(self, ffp_dim=256, auto_dim=216, seq_len=1000,
                 d_model=128, nhead=4, num_layers=3, latent_dim=64, dropout=0.1):
        super().__init__()
        self.seq_len = seq_len
        self.latent_dim = latent_dim
        self.ffp_dim = ffp_dim
        self.auto_dim = auto_dim
        in_features = ffp_dim + auto_dim
        self.element_net = nn.Sequential(nn.Linear(in_features, d_model),nn.LayerNorm(d_model), nn.Mish(), nn.Linear(d_model, d_model), nn.LayerNorm(d_model))
        self.global_net = nn.Sequential(nn.Linear(d_model, d_model),nn.LayerNorm(d_model), nn.Mish(), nn.Linear(d_model, latent_dim * 2))
        self.slots = nn.Parameter(torch.randn(1, seq_len, d_model) * 0.2)
        self.shared_decoder = nn.Sequential(nn.Linear(latent_dim + d_model, d_model),nn.LayerNorm(d_model),nn.Mish(),nn.Linear(d_model, ffp_dim + auto_dim))

    def encode(self, ffp, auto):
        x = torch.cat([ffp, auto], dim=-1)          
        x = self.element_net(x)                     
        x_pool = x.mean(dim=1)                       
        out = self.global_net(x_pool)               
        mu, logvar = out.chunk(2, dim=-1)
        return mu, logvar

    def reparameterize(self, mu, logvar):
        if not self.training:
            return mu
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def decode(self, z):
        b = z.size(0)
        z_exp = z.unsqueeze(1).expand(-1, self.seq_len, -1)   
        slots = self.slots.expand(b, -1, -1)                
        combined = torch.cat([z_exp, slots], dim=-1)        
        out = self.shared_decoder(combined)                  
        recon_ffp = F.softplus(out[..., :self.ffp_dim])
        recon_auto = out[..., self.ffp_dim:]
        return recon_ffp, recon_auto

    def forward(self, ffp, auto):
        mu, logvar = self.encode(ffp, auto)
        z = self.reparameterize(mu, logvar)
        recon_ffp, recon_auto = self.decode(z)
        return recon_ffp, recon_auto, mu, logvar
```

#### loss

```python
def chamfer_distance(source, target):
    dist = torch.cdist(source, target, p=2, compute_mode="use_mm_for_euclid_dist_if_necessary").pow(2)
    src_to_tgt_min, _ = torch.min(dist, dim=-1)
    tgt_to_src_min, _ = torch.min(dist, dim=-2)
    chamfer_loss = torch.mean(src_to_tgt_min) + torch.mean(tgt_to_src_min)
    
    return chamfer_loss


def loss_function_set(recon_ffp, ffp_true, recon_auto, auto_true, mu, logvar, beta=1.0):
    ffp_loss = chamfer_distance(recon_ffp/50.0, ffp_true/50.0)
    auto_loss = chamfer_distance(recon_auto/110.0, auto_true/110.0)
    kl = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp()) / mu.numel()
    total = ffp_loss*30 + auto_loss*20 + beta * kl
    return total, ffp_loss, auto_loss, kl

def loss_function_cls(recon_ffp, ffp_true, recon_auto, auto_true, mu, logvar, beta=1.0):
    ffp_loss = F.poisson_nll_loss(recon_ffp, ffp_true.float(), log_input=False, full=True, reduction='mean')
    auto_loss = F.mse_loss(recon_auto, auto_true.float())
    kl_loss = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp()) / mu.numel()
    total = 10*ffp_loss + auto_loss + beta * kl_loss
    return total, ffp_loss, auto_loss, kl_loss
```

## Train

```python
class PolymerDataset(Dataset):
    def __init__(self, h5_path):
        self.h5_path = h5_path
        with h5py.File(self.h5_path, 'r') as f:
            self.length = f['cffp_256'].shape[0]
        self.file = None
        self.ffp = None
        self.auto = None

    def __len__(self):
        return self.length

    def _init_h5(self):
        if self.file is None:
            self.file = h5py.File(self.h5_path, 'r')
            self.ffp = self.file['cffp_256']
            self.auto = self.file['autocorr']

    def __getitem__(self, idx):
        self._init_h5()
        ffp_sample = self.ffp[idx]
        auto_sample = self.auto[idx]
        return (torch.from_numpy(ffp_sample).float(),
                torch.from_numpy(auto_sample).float())

    def close(self):
        if self.file is not None:
            self.file.close()
            self.file = None
```

```python
#sinkhorn_eval = SamplesLoss("sinkhorn", p=2, blur=0.05, scaling=0.95)
```

```python
def train(model, train_loader, val_loader, epochs, lr, beta, device, model_name, loss_fn,
          warmup_epochs=15, start_beta=0.0, patience=10):
    model.to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5)
    best_val_loss = float('inf')
    no_improve = 0

    writer = SummaryWriter(log_dir=f"runs/{model_name}_VAE")
    
    history = {'train_loss': [], 'val_loss': [], 'val_recon': [], 'val_geom': [],
               'val_ffp': [], 'val_auto': [], 'val_kl': [], 'beta': []}

    for epoch in range(1, epochs+1):
        epoch_start = time.time() 
        current_beta = start_beta + (beta - start_beta) * (epoch - 1) / max(1, warmup_epochs - 1) if epoch <= warmup_epochs else beta
        history['beta'].append(current_beta)
        train_start = time.time()                   
        model.train()
        train_loss_sum = torch.tensor(0.0, device=device)
        train_ffp_sum  = torch.tensor(0.0, device=device)
        train_auto_sum = torch.tensor(0.0, device=device)
        train_kl_sum   = torch.tensor(0.0, device=device)
        for ffp, auto in train_loader:
            ffp, auto = ffp.to(device), auto.to(device)
            optimizer.zero_grad()
            recon_ffp, recon_auto, mu, logvar = model(ffp, auto)
            loss, f_loss, a_loss, kl_loss = loss_fn(recon_ffp, ffp, recon_auto, auto, mu, logvar, current_beta)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            
            bs = ffp.size(0)
            train_loss_sum += loss.detach() * bs
            train_ffp_sum  += f_loss.detach() * bs
            train_auto_sum += a_loss.detach() * bs
            train_kl_sum   += kl_loss.detach() * bs

        n_train = len(train_loader.dataset)
        avg_train = train_loss_sum.item() / n_train
        avg_train_ffp   = train_ffp_sum.item() / n_train
        avg_train_auto  = train_auto_sum.item() / n_train
        avg_train_kl    = train_kl_sum.item() / n_train
        avg_train_recon = avg_train_ffp + avg_train_auto
        
        train_time = time.time() - train_start
        val_start = time.time()                     
        model.eval()
        val_loss_sum = torch.tensor(0.0, device=device)
        val_ffp_sum  = torch.tensor(0.0, device=device)
        val_auto_sum = torch.tensor(0.0, device=device)
        val_kl_sum   = torch.tensor(0.0, device=device)
        val_geom_cpu = 0.0

        with torch.no_grad():
            for ffp, auto in val_loader:
                ffp, auto = ffp.to(device), auto.to(device)
                recon_ffp, recon_auto, mu, logvar = model(ffp, auto)
                loss, f_loss, a_loss, kl_loss = loss_fn(recon_ffp, ffp, recon_auto, auto, mu, logvar, current_beta)
                bs = ffp.size(0)
                val_loss_sum += loss.detach() * bs
                val_ffp_sum  += f_loss.detach() * bs
                val_auto_sum += a_loss.detach() * bs
                val_kl_sum   += kl_loss.detach() * bs

                geom_ffp_val = chamfer_distance(recon_ffp.float(), ffp.float()).item()
                geom_auto_val = chamfer_distance(recon_auto.float(), auto.float()).item()
                val_geom_cpu += (geom_ffp_val + geom_auto_val) * bs

        n_val = len(val_loader.dataset)
        avg_val  = val_loss_sum.item() / n_val
        avg_ffp  = val_ffp_sum.item() / n_val
        avg_auto = val_auto_sum.item() / n_val
        avg_kl   = val_kl_sum.item() / n_val
        avg_recon = avg_ffp + avg_auto
        avg_geom  = val_geom_cpu / n_val
        val_time = time.time() - val_start
        for k, v in zip(['train_loss','val_loss','val_recon','val_geom','val_ffp','val_auto','val_kl'],
                        [avg_train, avg_val, avg_recon, avg_geom, avg_ffp, avg_auto, avg_kl]):
            history[k].append(v)
        writer.add_scalars("Loss/Total", {"Train": avg_train, "Validation": avg_val}, epoch)
        writer.add_scalars("Loss/Reconstruction_Total", {"Train": avg_train_recon, "Validation": avg_recon}, epoch)
        writer.add_scalars("Loss/FFP_Component", {"Train": avg_train_ffp, "Validation": avg_ffp}, epoch)
        writer.add_scalars("Loss/Auto_Component", {"Train": avg_train_auto, "Validation": avg_auto}, epoch)
        writer.add_scalars("Loss/KL_Divergence", {"Train": avg_train_kl, "Validation": avg_kl}, epoch)
        writer.add_scalar("Metric/Validation_Geometry", avg_geom, epoch)
        writer.add_scalar("Hyperparameters/Beta", current_beta, epoch)
        writer.add_scalar("Hyperparameters/Learning_Rate", optimizer.param_groups[0]['lr'], epoch)
        if epoch == 1 or epoch % 5 == 0:
            for name, param in model.named_parameters():
                tb_name = name.replace('.', '/')
                writer.add_histogram(f"Weights/{tb_name}", param.data, epoch)
                if param.grad is not None:
                    writer.add_histogram(f"Gradients/{tb_name}", param.grad.data, epoch)
        
        epoch_time = time.time() - epoch_start
        print(f"{model_name} | Epoch {epoch:3d} | β: {current_beta:.3f} | "
              f"Tr: {avg_train:.4f} | Val: {avg_val:.4f} | Recon: {avg_recon:.4f} | Geom: {avg_geom:.4f} | "
              f"ffp: {avg_ffp:.4f} auto: {avg_auto:.4f} KL: {avg_kl:.4f}"
              f"Time: {epoch_time:.1f}s (train {train_time:.1f}s, val {val_time:.1f}s)")

        if epoch > warmup_epochs:
            scheduler.step(avg_val)
            if avg_val < best_val_loss:
                best_val_loss = avg_val
                no_improve = 0
                torch.save(model.state_dict(), f"./data/DL/models/{model_name}_VAE.pt")
            else:
                no_improve += 1
            if no_improve >= patience:
                print(f"Early stopping at epoch {epoch}")
                break
                
    writer.close()
    return history
```

```python
model_config = {
    'ffp_dim': 256,
    'auto_dim': 216,
    'seq_len': 1000,
    'd_model': 128,
    'nhead': 4,
    'num_layers': 3,
    'latent_dim': 64,
    'dropout': 0.12,
}

train_config = {
    'batch_size': 32,
    'epochs': 100,
    'lr': 1e-4,
    'beta': 1.0,
    'warmup_epochs': 20,
    'start_beta': 0.0,
    'patience': 15,
}

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

train_ds = PolymerDataset('./data/DL/Deep_poly_datasets_Tr.h5')
val_ds   = PolymerDataset('./data/DL/Deep_poly_datasets_Val.h5')

train_loader = DataLoader(train_ds, batch_size=train_config['batch_size'], shuffle=True, num_workers=4, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=train_config['batch_size'], shuffle=False, num_workers=4, pin_memory=True)

model_factories = {
    'CLS': lambda: TransformerVAE_CLS(**model_config),
    'SET': lambda: TransformerVAE_SET(**model_config),
    'DS':  lambda: VAE_DeepSets(**model_config),
}

loss_fns = {
    'CLS': loss_function_cls,
    'SET': loss_function_set,
    'DS':  loss_function_set,
}

histories = {}

for name, create_model in model_factories.items():
    print(f"\n{'='*50}\nTraining {name}\n{'='*50}")
    
    model = create_model().to(device)
    
    histories[name] = train(
        model, train_loader, val_loader,
        epochs=train_config['epochs'],
        lr=train_config['lr'],
        beta=train_config['beta'],
        device=device,
        model_name=name,
        loss_fn=loss_fns[name],
        warmup_epochs=train_config['warmup_epochs'],
        start_beta=train_config['start_beta'],
        patience=train_config['patience']
    )
    
    del model
    if device.type == 'cuda':
        torch.cuda.empty_cache()
    gc.collect()

train_ds.close()
val_ds.close()

print(f"\n{'='*50}\nFinal Results:\n{'='*50}")
for name, hist in histories.items():
    last_geom = hist['val_geom'][-1]
    print(f"{name}: final val_geom = {last_geom:.6f}")
```

## loading

```python
def load_pretrained_vae(model_type, checkpoint_path, device):
    model_config = {
        'ffp_dim': 256,
        'auto_dim': 216,
        'seq_len': 1000,
        'd_model': 128,
        'nhead': 4,
        'num_layers': 3,
        'latent_dim': 64,
        'dropout': 0.12,
    }

    models = {
        "CLS": TransformerVAE_CLS,
        "SET": TransformerVAE_SET,
        "DS": VAE_DeepSets,
    }

    vae = models[model_type](**model_config)
    state_dict = torch.load(checkpoint_path, map_location=device)
    vae.load_state_dict(state_dict)
    vae.to(device)
    vae.eval()

    return vae
```

```python
class H5LatentDataset(Dataset):
    def __init__(self, h5_path):
        self.file = h5py.File(h5_path, 'r')
        self.ffp = self.file['cffp_256']
        self.auto = self.file['autocorr']

    def __len__(self):
        return self.ffp.shape[0]

    def __getitem__(self, idx):
        return (
            torch.from_numpy(self.ffp[idx]).float(),
            torch.from_numpy(self.auto[idx]).float()
        )

    def close(self):
        self.file.close()

def save_latent_vectors_npy(
    model_type,        # "CLS", "SET", "DS"
    checkpoint_path,   # .pt 
    h5_input_path,     # HDF5
    output_npy_path,   # latent_vectors.npy
    device,
    batch_size=32
):
    vae = load_pretrained_vae(model_type, checkpoint_path, device)
    vae.eval()

    dataset = H5LatentDataset(h5_input_path)
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=False, num_workers=2)

    all_mu = []
    with torch.no_grad():
        for ffp, auto in loader:
            ffp, auto = ffp.to(device), auto.to(device)
            mu, _ = vae.encode(ffp, auto)
            all_mu.append(mu.cpu().numpy())

    all_mu = np.concatenate(all_mu, axis=0)
    np.save(output_npy_path, all_mu)
    print(f" {len(all_mu)} --- {output_npy_path}")
    dataset.close()
    return all_mu
```

```python
def generate_latent_h5(
    model_type: str,
    checkpoint_path: str,
    device: torch.device,
    ffp_list: List[np.ndarray],
    auto_list: List[np.ndarray],
    output_h5_path: str,
    total_samples: int,
    seq_len: int,
    batch_size: int = 5000,
    max_monomers: int = 7,
) -> None:
    np.random.seed(SEED)
    
    vae = load_pretrained_vae(model_type, checkpoint_path, device)
    vae.eval()
    num_monomers = len(ffp_list)
    ffp_arr = np.array(ffp_list, dtype=np.float32)
    auto_arr = np.array(auto_list, dtype=np.float32)
    
    with torch.no_grad():
        dummy_ffp = torch.from_numpy(ffp_arr[:1]).repeat(1, seq_len, 1).to(device)
        dummy_auto = torch.from_numpy(auto_arr[:1]).repeat(1, seq_len, 1).to(device)
        mu_dummy, _ = vae.encode(dummy_ffp, dummy_auto)
        latent_dim = mu_dummy.shape[1]
    
    with h5py.File(output_h5_path, 'w') as f:
        frac_dset = f.create_dataset('fractions', shape=(0, num_monomers), maxshape=(None, num_monomers), dtype=np.float32, compression='gzip', compression_opts=4)
        mu_dset = f.create_dataset('mu', shape=(0, latent_dim), maxshape=(None, latent_dim), dtype=np.float32, compression='gzip', compression_opts=4)
        pbar = tqdm(total=total_samples, desc="Generating")
        for start in range(0, total_samples, batch_size):
            end = min(start + batch_size, total_samples)
            cur_batch = end - start
            frac_batch = np.empty((cur_batch, num_monomers), dtype=np.float32)
            mu_batch = np.empty((cur_batch, latent_dim), dtype=np.float32)
            ffp_arr_local = ffp_arr
            auto_arr_local = auto_arr
            n_monomers = num_monomers
            seq_len_local = seq_len
            
            for i in range(cur_batch):
                k = np.random.randint(1, min(max_monomers, n_monomers) + 1)
                idxs = np.random.choice(n_monomers, size=k, replace=False)
                r = np.random.dirichlet(np.ones(k))
                
                frac_i = np.zeros(n_monomers, dtype=np.float32)
                frac_i[idxs] = r
                frac_batch[i] = frac_i
                
                picks = np.random.choice(idxs, size=seq_len_local, p=r)
                ffp_seq = ffp_arr_local[picks]
                auto_seq = auto_arr_local[picks]
                
                ffp_t = torch.from_numpy(ffp_seq).unsqueeze(0).to(device)
                auto_t = torch.from_numpy(auto_seq).unsqueeze(0).to(device)
                with torch.no_grad():
                    mu_i, _ = vae.encode(ffp_t, auto_t)
                mu_batch[i] = mu_i.squeeze(0).cpu().numpy()
            
            frac_dset.resize(frac_dset.shape[0] + cur_batch, axis=0)
            mu_dset.resize(mu_dset.shape[0] + cur_batch, axis=0)
            frac_dset[-cur_batch:] = frac_batch
            mu_dset[-cur_batch:] = mu_batch
            
            pbar.update(cur_batch)
        pbar.close()
    
    print(f"Save to {output_h5_path}")
```

```python
ffp_list = valid_df['cffp_256'].tolist()
auto_list = valid_df['autocorr'].tolist()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

models = ['CLS','SET', 'DS']
checkpoints = {m: f'./data/DL/models/{m}_VAE.pt' for m in models}
outputs = {m: f'./data/DL/latent_db_{m}.h5' for m in models}

for model in models:
    print(f"\nGenerating latent database for {model}...")
    generate_latent_h5(
        model_type=model,
        checkpoint_path=checkpoints[model],
        device=device,
        ffp_list=ffp_list,
        auto_list=auto_list,
        output_h5_path=outputs[model],
        total_samples=5000000,
        seq_len=1000,
        batch_size=16,
        max_monomers=7
    )
    print(f"Done: {outputs[model]}")
```

```python
from __future__ import annotations

def generate_latent_h5_multi(
    models: dict[str, str], device: torch.device, valid_df,
    output_h5_path: str, total_samples: int, seq_len: int, *,
    h5_batch_size: int = 8192, vae_batch_size: int = 256,
    max_monomers: int = 7, seed: int = SEED, compression: str | None = "lzf",
) -> None:
    ffp  = np.asarray(valid_df["cffp_256"].tolist(), np.float32)
    auto = np.asarray(valid_df["autocorr"].tolist(), np.float32)
    names, smiles = valid_df["name"].tolist(), valid_df["smiles"].tolist()
    numbers = valid_df["number"].to_numpy(np.int32)
    M, fdim, adim = ffp.shape[0], ffp.shape[1], auto.shape[1]
    assert auto.shape[0] == M == len(names) == len(numbers) == len(smiles)
    kmax = min(max_monomers, M)
    k_ar = np.arange(kmax, dtype=np.int32)[None, :]
    rng  = np.random.default_rng(seed)
    Path(output_h5_path).parent.mkdir(parents=True, exist_ok=True)
    if device.type == "cuda":
        torch.backends.cuda.matmul.allow_tf32 = True
        torch.backends.cudnn.allow_tf32 = True

    vaes = {mt: torch.compile(load_pretrained_vae(mt, ckpt, device).eval(), mode="max-autotune") for mt, ckpt in models.items()}    
    with torch.inference_mode():
        d_ffp  = torch.from_numpy(ffp[:1]).repeat(1, seq_len, 1).to(device)
        d_auto = torch.from_numpy(auto[:1]).repeat(1, seq_len, 1).to(device)
        dims   = {mt: int(vae.encode(d_ffp, d_auto)[0].shape[1]) for mt, vae in vaes.items()}
    del d_ffp, d_auto
    if device.type == "cuda": torch.cuda.empty_cache()

    B = vae_batch_size
    ffp_tbl, auto_tbl = torch.from_numpy(ffp).to(device), torch.from_numpy(auto).to(device)   # крошечные, 1× H2D
    pk_pin = torch.empty((B, seq_len), dtype=torch.int64, pin_memory=True); pk_np = pk_pin.numpy()
    pk_gpu = torch.empty((B, seq_len), dtype=torch.int64, device=device)
    mu_pin = {mt: torch.empty((B, dims[mt]), pin_memory=True) for mt in models}
    mu_np  = {mt: mu_pin[mt].numpy() for mt in models}
    keys_buf = np.empty((B, M), np.float32);  gam_buf = np.empty((B, kmax), np.float32)
    frs_buf  = np.empty((B, kmax), np.float32); cum_buf = np.empty((B, kmax), np.float32)
    u_buf    = np.empty((B, seq_len), np.float32); fo_buf = np.zeros((B, M), np.float32)

    with h5py.File(output_h5_path, "w") as f:
        frac_d = f.create_dataset("fractions", (total_samples, M), np.float32,
                                   chunks=(min(h5_batch_size, total_samples), M), compression=compression)
        mu_d = {mt: f.create_dataset(f"mu_{mt}", (total_samples, dims[mt]), np.float32,
                                      chunks=(min(h5_batch_size, total_samples), dims[mt]), compression=compression)
                for mt in models}
        sdt = h5py.string_dtype("utf-8")
        f.attrs.update(monomer_names=np.array(names, sdt), monomer_numbers=numbers,
                       monomer_smiles=np.array(smiles, sdt),
                       num_monomers=M, seq_len=seq_len, seed=seed,
                       total_samples=total_samples, max_monomers=max_monomers,
                       models=np.array(list(models), sdt),
                       layout="row i <-> sample i; column j <-> valid_df.iloc[j]",
                       **{f"latent_dim_{mt}": dims[mt] for mt in models})

        pbar, t0 = tqdm(total=total_samples, desc="Generating", unit="smp"), time.time()
        try:
            for o0 in range(0, total_samples, h5_batch_size):
                o1, oc = min(o0 + h5_batch_size, total_samples), min(h5_batch_size, total_samples - o0)
                fo = np.empty((oc, M), np.float32)
                mo = {mt: np.empty((oc, dims[mt]), np.float32) for mt in models}
                for m0 in range(0, oc, B):
                    m1 = min(m0 + B, oc); mc = m1 - m0
                    ks = rng.integers(1, kmax + 1, size=mc)
                    keys_buf[:mc] = rng.random((mc, M))
                    sel = np.argpartition(keys_buf[:mc], kmax - 1, axis=1)[:, :kmax]
                    gam_buf[:mc] = rng.standard_gamma(1.0, (mc, kmax)) * (k_ar < ks[:, None])
                    np.divide(gam_buf[:mc], gam_buf[:mc].sum(1, keepdims=True), out=frs_buf[:mc])
                    fo_buf[:mc].fill(0.0)
                    np.put_along_axis(fo_buf[:mc], sel, frs_buf[:mc], axis=1)
                    fo[m0:m1] = fo_buf[:mc]
                    np.cumsum(frs_buf[:mc], axis=1, out=cum_buf[:mc])
                    cum_buf[:mc, -1] = 1.0
                    u_buf[:mc] = rng.random((mc, seq_len))
                    pk_small = (u_buf[:mc, :, None] > cum_buf[:mc, None, :]).sum(-1)
                    pk_np[:mc] = np.take_along_axis(sel, pk_small, axis=1)
                    pk_gpu[:mc].copy_(pk_pin[:mc], non_blocking=True)
                    ft = ffp_tbl[pk_gpu[:mc]]
                    at = auto_tbl[pk_gpu[:mc]]
                    with torch.inference_mode():
                        for mt, vae in vaes.items():
                            mu_pin[mt][:mc].copy_(vae.encode(ft, at)[0], non_blocking=True)
                    torch.cuda.synchronize()
                    for mt in models: mo[mt][m0:m1] = mu_np[mt][:mc]

                frac_d[o0:o1] = fo
                for mt in models: mu_d[mt][o0:o1] = mo[mt]
                pbar.update(oc)
                if device.type == "cuda":
                    pbar.set_postfix(smp_s=f"{pbar.n / max(time.time() - t0, 1e-6):.0f}",
                                     gpu=f"{torch.cuda.max_memory_allocated() / 1e9:.1f}GB")
            f.flush()
        finally:
            pbar.close()
    print(f"Saved {total_samples} samples -> {output_h5_path}")
```

```python
from __future__ import annotations
import itertools, math, time; from pathlib import Path; import h5py, numpy as np, torch; from scipy.stats import qmc; from tqdm import tqdm
def build_architecture_pool(M: int, kmax: int, seed: int):
    total = sum(math.comb(M, k) for k in range(1, kmax + 1))
    idx = np.empty((total, kmax), dtype=np.int32); sizes = np.empty(total, dtype=np.int8)
    i = 0
    for k in range(1, kmax + 1):
        for combo in itertools.combinations(range(M), k):
            idx[i, :k] = combo
            if k < kmax: idx[i, k:] = combo[0]
            sizes[i] = k; i += 1
    rng = np.random.default_rng(seed); perm = rng.permutation(total)
    return idx[perm], sizes[perm], total
def build_sample_order(total_universe: int, total_samples: int, seed: int):
    rng = np.random.default_rng(seed + 1); r = total_samples // total_universe; rem = total_samples - r * total_universe
    arch_chunks, repeat_chunks = [], []
    for e in range(r):
        arch_chunks.append(rng.permutation(total_universe)); repeat_chunks.append(np.full(total_universe, e, dtype=np.int32))
    if rem:
        arch_chunks.append(rng.permutation(total_universe)[:rem]); repeat_chunks.append(np.full(rem, r, dtype=np.int32))
    arch_id = np.concatenate(arch_chunks) if arch_chunks else np.empty(0, np.int64)
    repeat_id = np.concatenate(repeat_chunks) if repeat_chunks else np.empty(0, np.int32)
    order = rng.permutation(total_samples); n_repeats = r + (1 if rem else 0)
    return arch_id[order], repeat_id[order], max(n_repeats, 1)
def build_sobol_bank(kmax: int, n_repeats: int, seed: int):
    bank = {}
    for k in range(1, kmax + 1):
        d = k - 1
        if d == 0: bank[k] = np.zeros((n_repeats, 0), dtype=np.float64); continue
        sampler = qmc.Sobol(d=d, scramble=True, seed=seed + 1000 + k)
        m = max(1, math.ceil(math.log2(max(n_repeats, 1)))); bank[k] = sampler.random_base2(m)[:n_repeats]
    return bank
def rotated_dirichlet(u: np.ndarray, shift: np.ndarray) -> np.ndarray:
    n, d = u.shape
    if d == 0: return np.ones((n, 1), dtype=np.float32)
    rotated = (u + shift) % 1.0; s = np.sort(rotated, axis=1)
    edges = np.concatenate([np.zeros((n, 1)), s, np.ones((n, 1))], axis=1)
    return np.diff(edges, axis=1).astype(np.float32)
def generate_latent_h5_multi(models: dict[str, str], device: torch.device, valid_df, output_h5_path: str, total_samples: int, seq_len: int, *, h5_batch_size: int = 8192, vae_batch_size: int = 256, max_monomers: int = 5, seed: int = 0, compression: str | None = "lzf") -> None:
    ffp = np.asarray(valid_df["cffp_256"].tolist(), np.float32); auto = np.asarray(valid_df["autocorr"].tolist(), np.float32)
    names, smiles = valid_df["name"].tolist(), valid_df["smiles"].tolist(); numbers = valid_df["number"].to_numpy(np.int32)
    M, fdim, adim = ffp.shape[0], ffp.shape[1], auto.shape[1]
    assert auto.shape[0] == M == len(names) == len(numbers) == len(smiles)
    kmax = min(max_monomers, M); Path(output_h5_path).parent.mkdir(parents=True, exist_ok=True)
    if device.type == "cuda": torch.backends.cuda.matmul.allow_tf32 = True; torch.backends.cudnn.allow_tf32 = True
    total_universe_estimate = sum(math.comb(M, k) for k in range(1, kmax + 1))
    if total_universe_estimate > 50_000_000:
        print(f"[coverage] WARNING: universe is estimated at {total_universe_estimate:,} architectures -- building it with a python itertools.combinations loop at this scale can take minutes and tens of GB of memory (idx: ~{total_universe_estimate*kmax*4/1e9:.1f} GB). This function is designed for M~64, kmax~5, not for arbitrarily large M/kmax.")
    arch_idx, arch_k, total_universe = build_architecture_pool(M, kmax, seed)
    if total_samples < total_universe:
        print(f"[coverage] WARNING: total_samples={total_samples:,} < universe={total_universe:,} -- full coverage will NOT be achieved, actual coverage is ~{100*total_samples/total_universe:.1f}% of architectures (r=0, random partial pass only).")
    sample_arch, sample_repeat, n_repeats = build_sample_order(total_universe, total_samples, seed)
    sobol_bank = build_sobol_bank(kmax, n_repeats, seed)
    shift_rng = np.random.default_rng(seed + 2); arch_shift = shift_rng.random((total_universe, max(kmax - 1, 1))).astype(np.float32)
    print(f"[coverage] universe={total_universe:,} architectures, {n_repeats} repeats per architecture (the last one may be partial)")
    vaes = {mt: torch.compile(load_pretrained_vae(mt, ckpt, device).eval(), mode="max-autotune") for mt, ckpt in models.items()}
    with torch.inference_mode():
        d_ffp = torch.from_numpy(ffp[:1]).repeat(1, seq_len, 1).to(device); d_auto = torch.from_numpy(auto[:1]).repeat(1, seq_len, 1).to(device)
        dims = {mt: int(vae.encode(d_ffp, d_auto)[0].shape[1]) for mt, vae in vaes.items()}
    del d_ffp, d_auto
    if device.type == "cuda": torch.cuda.empty_cache()
    B = vae_batch_size; ffp_tbl, auto_tbl = torch.from_numpy(ffp).to(device), torch.from_numpy(auto).to(device)
    pk_pin = torch.empty((B, seq_len), dtype=torch.int64, pin_memory=True); pk_np = pk_pin.numpy()
    pk_gpu = torch.empty((B, seq_len), dtype=torch.int64, device=device)
    mu_pin = {mt: torch.empty((B, dims[mt]), pin_memory=True) for mt in models}; mu_np = {mt: mu_pin[mt].numpy() for mt in models}
    frs_buf = np.empty((B, kmax), np.float32); cum_buf = np.empty((B, kmax), np.float32)
    u_buf = np.empty((B, seq_len), np.float32); fo_buf = np.empty((B, M), np.float32)
    rng = np.random.default_rng(seed + 3)
    with h5py.File(output_h5_path, "w") as f:
        frac_d = f.create_dataset("fractions", (total_samples, M), np.float32, chunks=(min(h5_batch_size, total_samples), M), compression=compression)
        mu_d = {mt: f.create_dataset(f"mu_{mt}", (total_samples, dims[mt]), np.float32, chunks=(min(h5_batch_size, total_samples), dims[mt]), compression=compression) for mt in models}
        sdt = h5py.string_dtype("utf-8")
        f.attrs.update(monomer_names=np.array(names, sdt), monomer_numbers=numbers, monomer_smiles=np.array(smiles, sdt), num_monomers=M, seq_len=seq_len, seed=seed, total_samples=total_samples, max_monomers=max_monomers, total_architectures=total_universe, n_repeats=n_repeats, models=np.array(list(models), sdt), layout="row i <-> sample i; column j <-> valid_df.iloc[j]", sampling="full-coverage architectures + Cranley-Patterson Sobol fractions", **{f"latent_dim_{mt}": dims[mt] for mt in models})
        pbar, t0 = tqdm(total=total_samples, desc="Generating", unit="smp"), time.time()
        try:
            for o0 in range(0, total_samples, h5_batch_size):
                o1, oc = min(o0 + h5_batch_size, total_samples), min(h5_batch_size, total_samples - o0)
                fo = np.empty((oc, M), np.float32); mo = {mt: np.empty((oc, dims[mt]), np.float32) for mt in models}
                for m0 in range(0, oc, B):
                    m1 = min(m0 + B, oc); mc = m1 - m0; g0 = o0 + m0
                    a_ids = sample_arch[g0:g0 + mc]; r_ids = sample_repeat[g0:g0 + mc]
                    sel = arch_idx[a_ids]; ks = arch_k[a_ids]
                    frs_buf[:mc].fill(0.0)
                    for k in range(1, kmax + 1):
                        rows = np.flatnonzero(ks == k)
                        if rows.size == 0: continue
                        u = sobol_bank[k][r_ids[rows] % sobol_bank[k].shape[0]]
                        shift = arch_shift[a_ids[rows], :max(k - 1, 0)]
                        frs_buf[rows, :k] = rotated_dirichlet(u, shift)
                    fo_buf[:mc].fill(0.0)
                    active_mask = np.arange(kmax)[None, :] < ks[:, None]
                    row_idx, col_idx = np.where(active_mask)
                    fo_buf[row_idx, sel[row_idx, col_idx]] = frs_buf[row_idx, col_idx]
                    fo[m0:m1] = fo_buf[:mc]
                    np.cumsum(frs_buf[:mc], axis=1, out=cum_buf[:mc]); cum_buf[:mc, -1] = 1.0
                    u_buf[:mc] = rng.random((mc, seq_len))
                    pk_small = (u_buf[:mc, :, None] > cum_buf[:mc, None, :]).sum(-1)
                    pk_np[:mc] = np.take_along_axis(sel, pk_small, axis=1)
                    pk_gpu[:mc].copy_(pk_pin[:mc], non_blocking=True)
                    ft = ffp_tbl[pk_gpu[:mc]]; at = auto_tbl[pk_gpu[:mc]]
                    with torch.inference_mode():
                        for mt, vae in vaes.items(): mu_pin[mt][:mc].copy_(vae.encode(ft, at)[0], non_blocking=True)
                    torch.cuda.synchronize()
                    for mt in models: mo[mt][m0:m1] = mu_np[mt][:mc]
                frac_d[o0:o1] = fo
                for mt in models: mu_d[mt][o0:o1] = mo[mt]
                pbar.update(oc)
                if device.type == "cuda": pbar.set_postfix(smp_s=f"{pbar.n / max(time.time() - t0, 1e-6):.0f}", gpu=f"{torch.cuda.max_memory_allocated() / 1e9:.1f}GB")
            f.flush()
        finally: pbar.close()
    print(f"Saved {total_samples} samples -> {output_h5_path}")
```

```python
def plot_training_history(histories):
    metrics = [
        ('train_loss', 'Train Total Loss'), ('val_loss', 'Val Total Loss'),
        ('val_recon', 'Val Recon Loss (ffp+auto)'), ('val_ffp', 'Val FFP Loss'),
        ('val_auto', 'Val Auto Loss'), ('val_kl', 'Val KL Divergence'),
        ('val_geom', 'Val Geometry (Chamfer)')
    ]
    fig, axes = plt.subplots(len(metrics), 2, figsize=(14, 3.2*len(metrics)), constrained_layout=True)
    for (key, title), (ax1, ax2) in zip(metrics, axes):
        for name, h in histories.items():
            v = np.where(np.isfinite(h[key]), np.array(h[key]), np.nan)
            e = np.arange(1, len(v)+1)
            ax1.plot(e, v, label=name, lw=1.5)
            ax2.plot(e, np.abs(v), label=name, lw=1.5)
        for ax, scale in [(ax1, 'linear'), (ax2, 'log')]:
            ax.set(title=f'{title} ({scale})', xlabel='Epoch', yscale=scale)
            ax.grid(True, alpha=0.3); ax.legend(fontsize=8)
    fig2, ax = plt.subplots(figsize=(10, 3.5), constrained_layout=True)
    for name, h in histories.items():
        ax.plot(range(1, len(h['beta'])+1), h['beta'], label=name, lw=2)
    ax.set(title='Beta Warm-up Schedule', xlabel='Epoch', ylabel='Beta')
    ax.grid(True, alpha=0.3); ax.legend()
    plt.show()
```

```python
class RawTestDataset(Dataset):
    def __init__(self, h5_path):
        self.file = h5py.File(h5_path, 'r')
        self.ffp = self.file['cffp_256']
        self.auto = self.file['autocorr']
    def __len__(self): return self.ffp.shape[0]
    def __getitem__(self, idx):
        return torch.from_numpy(self.ffp[idx]).float(), torch.from_numpy(self.auto[idx]).float()
    def close(self): self.file.close()

test_ds = RawTestDataset('./data/DL/Deep_poly_datasets_Val.h5')
test_loader = DataLoader(test_ds, batch_size=64, shuffle=False, num_workers=0)

model_cls = load_pretrained_vae('CLS', './data/DL/models/CLS_VAE.pt', device)
model_set = load_pretrained_vae('SET', './data/DL/models/SET_VAE.pt', device)
model_ds = load_pretrained_vae('DS', './data/DL/models/DS_VAE.pt', device)

def get_latent_mu(model, loader):
    latents = []
    with torch.no_grad():
        for ffp, auto in loader:
            ffp, auto = ffp.to(device), auto.to(device)
            mu, _ = model.encode(ffp, auto)
            latents.append(mu.cpu().numpy())
    return np.concatenate(latents, axis=0)

z_cls = get_latent_mu(model_cls, test_loader)
z_set = get_latent_mu(model_set, test_loader)
z_ds = get_latent_mu(model_ds, test_loader)


def plot_pca(z_dict, title="PCA of Latent Space"):
    pca = PCA(n_components=2)
    fig, axes = plt.subplots(1, len(z_dict), figsize=(5*len(z_dict), 5))
    if len(z_dict) == 1: axes = [axes]
    for ax, (name, z) in zip(axes, z_dict.items()):
        if z is None: continue
        z_2d = pca.fit_transform(z)
        ax.scatter(z_2d[:,0], z_2d[:,1], s=1, alpha=0.5)
        ax.set_title(f'{name} (PCA)')
    plt.suptitle(title)
    plt.tight_layout()
    plt.show()

plot_pca({'CLS': z_cls, 'SET': z_set, 'DS': z_ds})

def plot_tsne(z_dict, title="t‑SNE of Latent Space", n_samples=3000):
    tsne = TSNE(n_components=2, perplexity=30, random_state=42)
    fig, axes = plt.subplots(1, len(z_dict), figsize=(5*len(z_dict), 5))
    if len(z_dict) == 1: axes = [axes]
    for ax, (name, z) in zip(axes, z_dict.items()):
        if z is None: continue
        idx = np.random.choice(len(z), min(n_samples, len(z)), replace=False)
        z_sub = z[idx]
        z_2d = tsne.fit_transform(z_sub)
        ax.scatter(z_2d[:,0], z_2d[:,1], s=5, alpha=0.5)
        ax.set_title(f'{name} (t‑SNE)')
    plt.suptitle(title)
    plt.tight_layout()
    plt.show()

plot_tsne({'CLS': z_cls, 'SET': z_set, 'DS': z_ds})
```

## Full all MLP pipeline

```python
def read_sheet(path, cols):
    return pd.read_excel(path, sheet_name=0, header=None, skiprows=1) \
             .iloc[:, :len(cols)].set_axis(cols, axis=1)

def parse_std(s):
    s = str(s).replace(',', '.').strip()
    m = re.search(r'([\d.]+)\s*±\s*([\d.]+)', s)
    if m: return float(m.group(1)), float(m.group(2))
    try: return float(s), float(s) * 0.07
    except: return np.nan, np.nan

def load_and_prepare(data_dir='data'):
    c6 = ['Round','ML','Rank','No.','ΦHEA','ΦBA','ΦCBEA','ΦATAC','ΦPEA','ΦAAm','Fa_max','Fa_ave','d','Q']
    c5 = ['No.','ΦHEA','ΦBA','ΦCBEA','ΦATAC','ΦPEA','ΦAAm','Fa_max','Fa_ave','d','Q','G','Tanδ','k']
    cc = ['ΦHEA','ΦBA','ΦCBEA','ΦATAC','ΦPEA','ΦAAm']

    df = pd.concat([
        read_sheet(f'{data_dir}/41586_2025_9269_MOESM5_ESM.xlsx', c5),
        read_sheet(f'{data_dir}/41586_2025_9269_MOESM6_ESM.xlsx', c6)
    ], ignore_index=True)[cc + ['Fa_max','Fa_ave','Q','G','Tanδ','k']]

    for c in cc + ['Fa_max']:
        df[c] = pd.to_numeric(df[c].astype(str).str.replace(',', '.'), errors='coerce')

    df = df.dropna(subset=cc)
    df = df[~df[cc].round(4).duplicated()].reset_index(drop=True)
    df[cc] = df[cc].div(df[cc].sum(axis=1), axis=0)

    p = df['Fa_ave'].apply(parse_std)
    df['Fa_mean'] = p.apply(lambda x: x[0]); df['Fa_std'] = p.apply(lambda x: x[1])
    mask = df['Fa_mean'].isna() & df['Fa_max'].notna()
    df.loc[mask,'Fa_mean'] = df.loc[mask,'Fa_max']
    df.loc[mask,'Fa_std']  = df.loc[mask,'Fa_mean']*0.05
    df = df[df['Fa_mean'] > 0].copy()

    q = df['Q'].apply(parse_std)
    df['Q_mean'] = q.apply(lambda x: x[0])

    for c in ['G','Tanδ','k']:
        df[c] = pd.to_numeric(df[c].astype(str).str.replace(',', '.').str.extract(r'([\d.]+)')[0], errors='coerce')

    return df.reset_index(drop=True), cc

def split_data(df, val_size=0.15, test_size=0.10, seed=42, n_bins=10):
    bins = pd.qcut(df['Fa_mean'], q=n_bins, labels=False, duplicates='drop')
    tr, te = train_test_split(df, test_size=test_size, stratify=bins, random_state=seed)
    tr, va = train_test_split(tr, test_size=val_size / (1 - test_size),
                              stratify=bins.loc[tr.index], random_state=seed)
    return tr.reset_index(drop=True), va.reset_index(drop=True), te.reset_index(drop=True)

def get_latent_for_composition(fracs, vae, mon_indices, features, device):
    exact = fracs * 1000
    counts = np.floor(exact).astype(np.int32)
    counts[np.argsort(exact % 1)[::-1][:1000 - counts.sum()]] += 1
    chain = np.repeat(mon_indices, counts)
    np.random.shuffle(chain)
    with torch.no_grad():
        auto = torch.from_numpy(features['autocorr'][0][chain]).float().unsqueeze(0).to(device)
        cffp = torch.from_numpy(features['cffp_256'][0][chain]).float().unsqueeze(0).to(device)
        mu, _ = vae.encode(cffp, auto)
    return mu.squeeze(0).cpu().numpy()

def augment_and_create_tensors(df, latent_col, seed=42, n_aug=30):
    rng = np.random.default_rng(seed)
    z = np.repeat(np.stack(df[latent_col].values), n_aug, axis=0)
    m = np.repeat(df['Fa_mean'].values, n_aug)
    s = np.repeat(np.maximum(df['Fa_std'].values, 0.01), n_aug)
    y = np.clip(rng.normal(m, s), 0, None).astype(np.float32)
    return z.astype(np.float32), y

def generate_pairs(df, latent_col='latent', target_col='Fa_mean'):
    pairs = []
    n = len(df)
    for i in range(n):
        for j in range(i+1, n):
            fa_i = df.iloc[i][target_col]
            fa_j = df.iloc[j][target_col]
            pairs.append({
                'z_A': df.iloc[i][latent_col],
                'z_B': df.iloc[j][latent_col],
                'label': 1.0 if fa_i > fa_j else 0.0
            })
    return pd.DataFrame(pairs)
```

### Models

```python
class MLP(nn.Module):
    def __init__(self, latent_dim=64, hidden_dim=256, dropout=0.15):
        super().__init__()
        self.input_layer = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.Mish(),
            nn.Dropout(dropout)
        )
        self.res_block1 = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.Mish(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim)
        )
        self.act1 = nn.Mish()
        self.head = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim // 4), # 256 -> 64
            nn.LayerNorm(hidden_dim // 4),
            nn.Mish(),
            nn.Dropout(dropout / 2),
            nn.Linear(hidden_dim // 4, 1)
        )
        self._init_weights()

    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Linear):
                nn.init.kaiming_normal_(m.weight, nonlinearity='linear')
                if m.bias is not None:
                    nn.init.zeros_(m.bias)
        with torch.no_grad():
            self.head[-1].bias.data.fill_(100.0)

    def forward(self, z):
        if z.ndim == 3: 
            z = z.mean(dim=1)
        h = self.input_layer(z)
        h = self.act1(h + self.res_block1(h))
        return self.head(h).squeeze(-1)
```

```python
class MLPNLL(nn.Module):
    def __init__(self, input_dim=64, hidden_dim=256, dropout=0.25):
        super().__init__()
        self.input_block = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.Mish(),
            nn.Dropout(dropout)
        )
        self.res_block = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.Mish(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim)
        )
        self.res_act = nn.Mish()
        self.mu_branch = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.LayerNorm(hidden_dim // 2),
            nn.Mish(),
            nn.Dropout(dropout / 2),
            nn.Linear(hidden_dim // 2, 1)
        )
        
        self.var_branch = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.LayerNorm(hidden_dim // 2),
            nn.Mish(),
            nn.Dropout(dropout / 2),
            nn.Linear(hidden_dim // 2, 1)
        )
        
        self._init_weights()
        
    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Linear):
                nn.init.kaiming_normal_(m.weight, nonlinearity='linear')
                if m.bias is not None:
                    nn.init.zeros_(m.bias)
        with torch.no_grad():
            self.mu_branch[-1].bias.data.fill_(100.0)
            self.var_branch[-1].bias.data.fill_(np.log(100.0))
            
    def forward(self, z):
        h = self.input_block(z)
        h = self.res_act(h + self.res_block(h))        
        mu = self.mu_branch(h).squeeze(-1)
        log_var = self.var_branch(h).squeeze(-1)
        log_var = torch.clamp(log_var, min=-3.0, max=8.0)
        return mu, log_var
```

```python
TARGETS = ['Fa_mean','Q_mean','G','Tanδ','k']
class MTMLP(nn.Module):
    def __init__(s, D=64, hidden=256, dropout=0.15):
        super().__init__()
        s.tower = nn.Sequential(
            nn.Linear(D,hidden), nn.LayerNorm(hidden), nn.Mish(), nn.Dropout(dropout),
            nn.Linear(hidden,hidden), nn.LayerNorm(hidden), nn.Mish(), nn.Dropout(dropout),
            nn.Linear(hidden,hidden//2), nn.LayerNorm(hidden//2), nn.Mish())
        s.heads = nn.ModuleList([nn.Linear(hidden//2, 1) for _ in TARGETS])
        s.log_sigma = nn.Parameter(torch.zeros(len(TARGETS)))
    def forward(s, z):
        h = s.tower(z)
        return torch.cat([head(h) for head in s.heads], -1)
```

```python
class PSMLP(nn.Module):
    def __init__(self, D=64, hidden=128, dropout=0.2):
        super().__init__()
        self.tower = nn.Sequential(
            nn.Linear(D, hidden), nn.LayerNorm(hidden), nn.Mish(), nn.Dropout(dropout),
            nn.Linear(hidden, D), nn.LayerNorm(D),      nn.Mish(), nn.Dropout(dropout))
        self.head = nn.Sequential(
            nn.Linear(2 * D, hidden), nn.Mish(), nn.Dropout(dropout),
            nn.Linear(hidden, 32), nn.Mish(), nn.Linear(32, 1))

    def _phi(s, hA, hB):
        d = hA - hB
        z = torch.cat([d, torch.abs(d)], dim=-1)
        return s.head(z).squeeze(dim=-1).view(-1)

    def forward(s, xA, xB):
        hA, hB = s.tower(xA), s.tower(xB)
        return s._phi(hA, hB)    
```

```python
class APSMLP(nn.Module):
    def __init__(self, D=64, hidden=128, dropout=0.2):
        super().__init__()
        self.tower = nn.Sequential(
            nn.Linear(D, hidden), nn.LayerNorm(hidden), nn.Mish(), nn.Dropout(dropout),
            nn.Linear(hidden, D), nn.LayerNorm(D),      nn.Mish(), nn.Dropout(dropout))
        self.head = nn.Sequential(
            nn.Linear(2 * D, hidden), nn.Mish(), nn.Dropout(dropout),
            nn.Linear(hidden, 32), nn.Mish(), nn.Linear(32, 1))

    def _phi(s, hA, hB):
        d = hA - hB
        z = torch.cat([d, torch.abs(d)], dim=-1)
        return s.head(z).squeeze(dim=-1).view(-1)

    def forward(s, xA, xB):
        hA, hB = s.tower(xA), s.tower(xB)
        return 0.5 * (s._phi(hA, hB) - s._phi(hB, hA))
```

```python
class ResBlock(nn.Module):
    def __init__(s, h, dp):
        super().__init__()
        s.f = nn.Sequential(
            nn.Linear(h, h), nn.LayerNorm(h), nn.Mish(), nn.Dropout(dp),
            nn.Linear(h, h), nn.LayerNorm(h))
    def forward(s, x): return x + s.f(x)

class DSMLP(nn.Module):
    def __init__(s, D=64, hidden=256, n_blocks=3, dropout=0.1):
        super().__init__()
        s.stem = nn.Sequential(
            nn.Linear(D, hidden), nn.LayerNorm(hidden), nn.Mish(), nn.Dropout(dropout))
        s.blocks = nn.ModuleList([ResBlock(hidden, dropout) for _ in range(n_blocks)])
        s.head = nn.Sequential(
            nn.Linear(hidden, hidden // 2), nn.LayerNorm(hidden // 2), nn.Mish(),
            nn.Dropout(dropout),
            nn.Linear(hidden // 2, 1))
        s._init()

    def _init(s):
        for m in s.modules():
            if isinstance(m, nn.Linear):
                nn.init.kaiming_normal_(m.weight, nonlinearity='linear')
                if m.bias is not None: nn.init.zeros_(m.bias)

    def score(s, x):
        h = s.stem(x)
        for b in s.blocks: h = b(h)
        return s.head(h).squeeze(dim=-1).view(-1)

    def forward(s, a, b):
        return s.score(a) - s.score(b)
```

```python
def _reg_loaders(z_tr, y_tr, z_va, y_va, bs=64):
    tr = TensorDataset(torch.from_numpy(z_tr), torch.from_numpy(y_tr))
    va = TensorDataset(torch.from_numpy(z_va), torch.from_numpy(y_va))
    return (DataLoader(tr, batch_size=bs,  shuffle=True,  num_workers=4, pin_memory=True, drop_last=True),
            DataLoader(va, batch_size=256, shuffle=False, num_workers=4, pin_memory=True))

def _mt_loaders(df_tr, df_va, lc, bs=64):
    def prep(df):
        z = np.stack(df[lc].values).astype(np.float32)
        Y = df[TARGETS].values.astype(np.float32)
        M = (~np.isnan(Y)).astype(np.float32)
        return z, np.nan_to_num(Y), M
    def ds(df):
        z,Y,M = prep(df)
        return TensorDataset(torch.from_numpy(z), torch.from_numpy(Y), torch.from_numpy(M))
    return (DataLoader(ds(df_tr), batch_size=bs, shuffle=True,  drop_last=True, num_workers=4, pin_memory=True),
            DataLoader(ds(df_va), batch_size=256, shuffle=False, num_workers=4, pin_memory=True))

def _pw_loaders(df_tr, df_va, latent_col, bs=64):
    def ds(df):
        p = generate_pairs(df, latent_col)
        return TensorDataset(
            torch.tensor(np.stack(p['z_A'].values), dtype=torch.float32),
            torch.tensor(np.stack(p['z_B'].values), dtype=torch.float32),
            torch.tensor(p['label'].values,          dtype=torch.float32))
    return (DataLoader(ds(df_tr), batch_size=bs,  shuffle=True,  num_workers=4, pin_memory=True, drop_last=True),
            DataLoader(ds(df_va), batch_size=256, shuffle=False, num_workers=4, pin_memory=True))

def cycle_loss(model, Z, n=1024):
    i = torch.randint(0, len(Z), (3,n), device=Z.device)
    A,B,C = Z[i[0]], Z[i[1]], Z[i[2]]
    ab,bc,ca = model(A,B), model(B,C), model(C,A)
    return (torch.relu(ab)*torch.relu(bc)*torch.relu(ca)).mean()

def trans_loss(model, Z, y, n=1024, m=0.1):
    i = torch.randint(0, len(Z), (3,n), device=Z.device)
    mask = (y[i[0]]>y[i[1]]) & (y[i[1]]>y[i[2]])
    if mask.sum() < 16: return torch.tensor(0., device=Z.device)
    A,B,C = Z[i[0]][mask], Z[i[1]][mask], Z[i[2]][mask]
    return torch.relu(m - model(A,C) + model(A,B)).mean()

def _pw(m, b, d, Z=None, y=None, β=0.3, γ=0.1):
    a, c, t = (x.to(d) for x in b)
    l = nn.functional.binary_cross_entropy_with_logits(m(a,c), t)
    if Z is not None:
        l = l + β*cycle_loss(m, Z) + γ*trans_loss(m, Z, y)
    return l, t, torch.sigmoid(m(a,c))

def _reg_eval(model, ckpt, z, y, task="nll", bs=256):
    dl = DataLoader(TensorDataset(torch.from_numpy(z), torch.from_numpy(y)),
                    batch_size=bs, shuffle=False, num_workers=4, pin_memory=True)
    p, t = predict(model, dl, device, task=task, ckpt=ckpt)
    return float(np.mean((p - t) ** 2))

def _mt(m, b, d):
    z, y, mk = (t.to(d) for t in b)
    p = m(z)
    mse = ((p - y)**2 * mk).sum(0) / mk.sum(0).clamp(min=1)
    l = (torch.exp(-m.log_sigma) * mse + m.log_sigma).mean()
    return l, y[:, :2], p[:, :2]   

def _mtm_eval(model, ckpt, df, lc, D):
    m = MTMLP(D).to(device)
    m.load_state_dict(torch.load(ckpt, map_location=device)["model"]); m.eval()
    Z = torch.from_numpy(np.stack(df[lc].values).astype(np.float32)).to(device)
    with torch.inference_mode(): P = m(Z).cpu().numpy()
    Y = df[TARGETS].values.astype(np.float32)
    j = TARGETS.index('Fa_mean')
    mk = ~np.isnan(Y[:, j])
    return float(np.mean((P[mk,j]-Y[mk,j])**2))

def _pw_eval(model, ckpt, df, latent_col, bs=256):
    p = generate_pairs(df, latent_col)
    dl = DataLoader(
        TensorDataset(
            torch.tensor(np.stack(p['z_A'].values), dtype=torch.float32),
            torch.tensor(np.stack(p['z_B'].values), dtype=torch.float32),
            torch.tensor(p['label'].values,          dtype=torch.float32)),
        batch_size=bs, shuffle=False, num_workers=4, pin_memory=True)
    preds, targets = predict(model, dl, device, task="bce", ckpt=ckpt)
    return float(roc_auc_score(targets, preds))


def _apw_eval(model, ckpt, df, latent_col, bs=256):
    p = generate_pairs(df, latent_col)
    dl = DataLoader(
        TensorDataset(
            torch.tensor(np.stack(p['z_A'].values), dtype=torch.float32),
            torch.tensor(np.stack(p['z_B'].values), dtype=torch.float32),
            torch.tensor(p['label'].values,          dtype=torch.float32)),
        batch_size=bs, shuffle=False, num_workers=4, pin_memory=True)
    preds, targets = predict(model, dl, device, task="pairwise", ckpt=ckpt)
    return float(roc_auc_score(targets, preds))
@dataclass
class Cfg:
    name: str; task: str="nll"; epochs: int=200; lr: float=1e-3; wd: float=1e-5; patience: int=25
    min_delta: float=1e-5; clip: float=1.0; accum: int=1; amp: bool=True; compile: bool=False
    ema: float=0.999; warmup: float=0.05; sched: str="plateau"; hist_every: int=5
    ckpt: Path=Path("./data/DL/models"); logs: Path=Path("runs"); seed: int=42
def _nll(m,b,d): z,y=(t.to(d) for t in b); mu,lv=m(z); return nn.functional.gaussian_nll_loss(mu,y,lv.exp()),y,mu
def _mse(m,b,d): z,y=(t.to(d) for t in b); p=m(z); return nn.functional.mse_loss(p,y),y,p
def _bce(m,b,d): a,c,y=(t.to(d) for t in b); l=m(a,c); return nn.functional.binary_cross_entropy_with_logits(l,y),y,torch.sigmoid(l)
R=lambda t,p:{"mse":float(np.mean((p-t)**2))}
C=lambda t,p:{"auc":roc_auc_score(t,p),"acc":accuracy_score(t,(p>.5).astype(np.float32))}
S   = {"nll":_nll, "mse":_mse, "pairwise":_pw, "mt":_mt, "bce":_bce}
M   = {"nll":R, "mse":R, "pairwise":C, "bce":C,
       "mt": lambda t, p: {
           "mse_Fa":  float(np.mean((p[:,0]-t[:,0])**2)),
           "mse_Q":   float(np.mean((p[:,1]-t[:,1])**2)),
           "mse_FaQ": float(np.mean((p[:,0]-t[:,0])**2) + np.mean((p[:,1]-t[:,1])**2)),
       }}
MON = {"nll":("val_mse",-1), "mse":("val_mse",-1),
       "pairwise":("val_auc",1), "mt":("val_mse_FaQ",-1), "bce":("val_auc",1)}
def train(model, tr, va, cfg: Cfg, device):
    random.seed(cfg.seed); np.random.seed(cfg.seed); torch.manual_seed(cfg.seed)
    torch.cuda.manual_seed_all(cfg.seed); torch.backends.cudnn.benchmark=True; cfg.ckpt.mkdir(parents=True,exist_ok=True)
    cb=cfg.ckpt/f"{cfg.name}_best.pt"
    if cfg.compile and hasattr(torch,"compile"): model.forward = torch.compile(model.forward, mode="max-autotune")
    opt=AdamW(model.parameters(),lr=cfg.lr,weight_decay=cfg.wd); spe=len(tr)
    mkey,sign=MON[cfg.task]; best=-sign*math.inf
    if cfg.sched=="cosine":
        T,W_=max(1,cfg.epochs*spe),max(1,int(cfg.warmup*cfg.epochs*spe))
        sch=LambdaLR(opt,lambda s:s/W_ if s<W_ else .5*(1+math.cos(math.pi*(s-W_)/max(1,T-W_)))); sm="step"
    else:
        sch=ReduceLROnPlateau(opt,mode=("min" if sign<0 else "max"), 
                              factor=.5,patience=15,min_lr=1e-6); sm="epoch"
    amp=cfg.amp and device.type=="cuda"; sc=GradScaler("cuda",enabled=amp); ac=lambda:autocast("cuda",enabled=amp)
    ema={k:v.detach().clone() for k,v in model.state_dict().items() if v.dtype.is_floating_point} if cfg.ema else None
    bad=0
    W=SummaryWriter(str(cfg.logs/f"{cfg.name}_{cfg.task}")); W.add_text("cfg",json.dumps({k:str(v) for k,v in cfg.__dict__.items()}))
    step,mfn=S[cfg.task],M[cfg.task]; itr,iva=1.0 / (len(tr) * tr.batch_size),1/len(va.dataset)
    pbar=tqdm(range(1,cfg.epochs+1),desc=f"[{cfg.name}]",unit="ep",dynamic_ncols=False,leave=True)
    for ep in range(1,cfg.epochs+1):
        model.train(); run=torch.zeros((),device=device); opt.zero_grad(set_to_none=True)
        for i,b in enumerate(tr):
            with ac(): l,y,_=step(model,b,device)
            loss=l/cfg.accum
            sc.scale(loss).backward()
            run+=l.detach()*y.size(0) 
            is_last = (i + 1 == len(tr))
            if (i+1)%cfg.accum==0 or is_last:
                sc.unscale_(opt)
                if torch.isfinite(torch.nn.utils.clip_grad_norm_(model.parameters(),cfg.clip)): sc.step(opt)
                sc.update(); opt.zero_grad(set_to_none=True)
                if sm=="step": sch.step()
                if ema is not None:
                    for k,v in model.state_dict().items():
                        if k in ema: ema[k].mul_(cfg.ema).add_(v.detach(),alpha=1-cfg.ema)
        trl=(run*itr).item()
        bak={k:v.detach().clone() for k,v in model.state_dict().items()} if ema else None
        if ema: model.load_state_dict({**model.state_dict(),**ema},strict=False)
        model.eval(); run=torch.zeros((),device=device); ps,ts=[],[]
        with torch.inference_mode(),ac():
            for b in va:
                l,y,p=step(model,b,device); run+=l.detach()*y.size(0)
                ps.append(p.float().cpu()); ts.append(y.cpu())
        val=(run*iva).item(); mt={f"val_{k}":v for k,v in mfn(torch.cat(ts).numpy(),torch.cat(ps).numpy()).items()}
        if ema: model.load_state_dict(bak)
        if sm=="epoch": sch.step(mt.get(mkey,val))
        lr=opt.param_groups[0]["lr"]
        post={"tr":f"{trl:.4f}","va":f"{val:.4f}","lr":f"{lr:.1e}","bad":f"{bad}/{cfg.patience}"}
        post.update({k[4:]:f"{v:.4f}" for k,v in mt.items()})
        pbar.set_postfix(post,refresh=False)
        W.add_scalars("Loss",{"train":trl,"val":val},ep); W.add_scalar("LR",lr,ep); W.add_scalar("Loss/gap",trl-val,ep)
        for k,v in mt.items(): W.add_scalar(f"Metrics/{k[4:]}",v,ep)
        if "val_mse" in mt: W.add_scalar("Metrics/rmse",math.sqrt(mt["val_mse"]),ep)
        if cfg.hist_every and (ep%cfg.hist_every==0 or ep==1):
            for n,p in model.named_parameters():
                W.add_histogram(f"W/{n}",p.data,ep)
                if p.grad is not None: W.add_histogram(f"G/{n}",p.grad.detach(),ep)
        mv=mt.get(mkey,val)
        if sign*(mv-best)>cfg.min_delta:
            best,bad=mv,0
            torch.save({"model":model.state_dict(),"ema":ema,"epoch":ep,"best":best,"monitor":mkey,"cfg":cfg.__dict__},cb)
        else: bad+=1
        W.add_scalars("EarlyStop",{"bad":bad,"best":best},ep)
        if bad>=cfg.patience: pbar.set_description(f"[{cfg.name}] stop@{ep} best {mkey}={best:.5f}");break
    pbar.close(); W.close(); return best

@torch.inference_mode()
def predict(model, loader, device, task="nll", ckpt=None):
    if ckpt and Path(ckpt).exists():
        c=torch.load(ckpt,map_location=device); model.load_state_dict(c["model"])
        if c.get("ema"): model.load_state_dict({**model.state_dict(),**c["ema"]},strict=False)
    model.eval().to(device); fn=S[task]; ps,ts=[],[]
    for b in loader:
        _,y,p=fn(model,b,device); ps.append(p.float().cpu()); ts.append(y.cpu())
    return torch.cat(ps).numpy(),torch.cat(ts).numpy()
```

```python
VAES   = ['CLS', 'SET', 'DS']
COMMON = dict(epochs=10000, lr=1e-4, wd=1e-5, patience=40)
N_AUG  = 30
rows   = []
```

```python
for vae in VAES:
    lc = f'latent_{vae}'
    print(f"\n{'='*60}\nVAE = {vae}  ({lc})\n{'='*60}")

    z_tr_aug, y_tr_aug = augment_and_create_tensors(df_train, lc, seed=42, n_aug=N_AUG)
    z_va_aug, y_va_aug = augment_and_create_tensors(df_val,   lc, seed=42, n_aug=N_AUG)

    z_tr = np.stack(df_train[lc].values).astype(np.float32); y_tr = df_train['Fa_mean'].values.astype(np.float32)
    z_va = np.stack(df_val[lc].values).astype(np.float32);   y_va = df_val['Fa_mean'].values.astype(np.float32)
    z_te = np.stack(df_test[lc].values).astype(np.float32);  y_te = df_test['Fa_mean'].values.astype(np.float32)
    D = z_tr.shape[1]

    tr_l, va_l = _reg_loaders(z_tr_aug, y_tr_aug, z_va_aug, y_va_aug)
    train(MLPNLL(D).to(device), tr_l, va_l,
                 Cfg(name=f"MLPNLL_{vae}", task="nll", **COMMON), device)
    ck = f"./data/DL/models/MLPNLL_{vae}_best.pt"; m = MLPNLL(D).to(device)
    rows.append(dict(VAE=vae, Model='MLPNLL',
                     Train=_reg_eval(m, ck, z_tr_aug, y_tr_aug, task="nll"),
                     Val  =_reg_eval(m, ck, z_va_aug, y_va_aug, task="nll"),
                     Test =_reg_eval(m, ck, z_te,     y_te,     task="nll")))

    tr_l, va_l = _reg_loaders(z_tr, y_tr, z_va, y_va)
    train(MLP(D).to(device), tr_l, va_l,
                 Cfg(name=f"MLP_{vae}", task="mse", **COMMON), device)
    ck = f"./data/DL/models/MLP_{vae}_best.pt"; m = MLP(D).to(device)
    rows.append(dict(VAE=vae, Model='MLP',
                     Train=_reg_eval(m, ck, z_tr, y_tr, task="mse"),
                     Val  =_reg_eval(m, ck, z_va, y_va, task="mse"),
                     Test =_reg_eval(m, ck, z_te, y_te, task="mse")))
    
    tr_l, va_l = _mt_loaders(df_train, df_val, lc)
    train(MTMLP(D).to(device), tr_l, va_l,
          Cfg(name=f"MTM_{vae}", task="mt", **COMMON), device)
    ck = f"./data/DL/models/MTM_{vae}_best.pt"
    m = MTMLP(D).to(device); m.load_state_dict(torch.load(ck, map_location=device)["model"]); m.eval()
    with torch.inference_mode():
        Zte = torch.from_numpy(np.stack(df_test[lc].values).astype(np.float32)).to(device)
        P = m(Zte).cpu().numpy()
    Yte = df_test[TARGETS].values.astype(np.float32)
    rows.append(dict(VAE=vae, Model='MTM',
         Train=_mtm_eval(m, ck, df_train, lc, D),
         Val  =_mtm_eval(m, ck, df_val,   lc, D),
         Test =_mtm_eval(m, ck, df_test,  lc, D)))
    for j, t in enumerate(TARGETS):
        mk = ~np.isnan(Yte[:, j])
        if mk.sum() < 5: continue
        rows.append(dict(VAE=vae, Model=f'MTM_{t}',
                         MSE=float(np.mean((P[mk,j]-Yte[mk,j])**2)),
                         Spearman=float(spearmanr(P[mk,j], Yte[mk,j]).statistic)))
    torch.cuda.empty_cache()

    tr_l, va_l = _pw_loaders(df_train, df_val, lc)
    train(PSMLP(D).to(device), tr_l, va_l,
          Cfg(name=f"PSMLP_{vae}", task="bce", sched="plateau", **COMMON), device)
    ck = f"./data/DL/models/PSMLP_{vae}_best.pt"
    m = PSMLP(D).to(device)
    rows.append(dict(VAE=vae, Model='PSMLP',
                     Train=_pw_eval(m, ck, df_train, lc),
                     Val  =_pw_eval(m, ck, df_val,   lc),
                     Test =_pw_eval(m, ck, df_test,  lc)))

    torch.cuda.empty_cache()

    
    tr_l, va_l = _pw_loaders(df_train, df_val, lc)
    Z_tr_t = torch.tensor(np.stack(df_train[lc].values), dtype=torch.float32, device=device)
    y_tr_t = torch.tensor(df_train['Fa_mean'].values,    dtype=torch.float32, device=device)
    S["pairwise"] = lambda m, b, d: _pw(m, b, d, Z=Z_tr_t, y=y_tr_t, β=0.3, γ=0.1)
    train(APSMLP().to(device), tr_l, va_l,
                 Cfg(name=f"APSMLP_{vae}", task="pairwise", sched="plateau", **COMMON), device)
    ck = f"./data/DL/models/APSMLP_{vae}_best.pt"; m = APSMLP().to(device)
    rows.append(dict(VAE=vae, Model='APSMLP',
                     Train=_apw_eval(m, ck, df_train, lc),
                     Val  =_apw_eval(m, ck, df_val,   lc),
                     Test =_apw_eval(m, ck, df_test,  lc)))

    torch.cuda.empty_cache()

    tr_l, va_l = _pw_loaders(df_train, df_val, lc)
    train(DSMLP(D).to(device), tr_l, va_l,
          Cfg(name=f"DSMLP_{vae}", task="bce", sched="plateau", **COMMON), device)
    ck = f"./data/DL/models/DSMLP_{vae}_best.pt"
    m = DSMLP(D).to(device)
    rows.append(dict(VAE=vae, Model='DSMLP',
                     Train=_pw_eval(m, ck, df_train, lc),
                     Val  =_pw_eval(m, ck, df_val,   lc),
                     Test =_pw_eval(m, ck, df_test,  lc)))

    torch.cuda.empty_cache()

df_res = pd.DataFrame(rows)
pd.set_option('display.float_format', '{:.4f}'.format)
d = df_res.drop_duplicates(['VAE','Model'], keep='first')
main = d[~d.Model.str.startswith('MTM_')]
mt   = d[d.Model.str.startswith('MTM_')]
print(main.pivot(index='VAE', columns='Model',
                 values=['Train','Val','Test']).to_string())
print(mt.pivot(index='VAE', columns='Model',
               values=['MSE','Spearman']).to_string())
```

## Sort descret 

```python
def load_model(path, cls, device=None, prefer_ema=True, **overrides):
    ckpt = torch.load(path, map_location="cpu", weights_only=False)
    if isinstance(ckpt, dict) and any(k in ckpt for k in ("model", "ema", "state_dict")):
        order = ("ema", "model", "state_dict") if prefer_ema else ("model", "ema", "state_dict")
        sd = next((ckpt[k] for k in order if ckpt.get(k) is not None), None)
        if sd is None:
            raise KeyError(f"no model/ema/state_dict in {path}")
        cfg = ckpt.get("cfg", {}) or {}
    else:
        sd, cfg = ckpt, {}
    for p in ("module.", "_orig_mod."):
        if all(k.startswith(p) for k in sd):
            sd = {k[len(p):]: v for k, v in sd.items()}
    params = set(inspect.signature(cls.__init__).parameters) - {"self"}
    kwargs = {k: v for k, v in cfg.items() if k in params}
    kwargs.update(overrides)
    model = cls(**kwargs)
    model.load_state_dict(sd, strict=True)
    if device is not None:
        model = model.to(device)
    return model.eval()
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
H5, MD = "./data/DL/latent_db_all.h5", "./data/DL/models"
VAES = ('CLS', 'SET', 'DS')
K, bs, KS = 1_000_000, 65536, (10, 100, 1000)
MODELS = [('PSMLP',  PSMLP,  'ps'), ('DSMLP',  DSMLP,  'ds'), ('MTM',    MTMLP,  'mt'), ('MLP',    MLP,    'reg'), ('MLPNLL', MLPNLL, 'reg'),]

def fwd(m, *xs, kind='reg'):
    n, out = len(xs[0]), np.empty(len(xs[0]), np.float32)
    for s in range(0, n, bs):
        e = min(s + bs, n)
        with torch.inference_mode():
            args = tuple(torch.from_numpy(x[s:e]).to(device) for x in xs)
            r = m(*args)
            if kind == 'mt':
                r = r[:, 0]
            if isinstance(r, tuple):
                r = r[0]
            out[s:e] = r.view(-1).cpu().numpy()
    return out


def gt(m, kind, X, Y):
    if kind in ('ps', 'ds'):
        return fwd(m, X, Y, kind=kind) > 0
    if kind == 'mt':
        return fwd(m, X, kind='mt') > fwd(m, Y, kind='mt')
    return fwd(m, X, kind='reg') > fwd(m, Y, kind='reg')


def score(m, kind, z):
    if kind == 'reg':  return fwd(m, z, kind='reg')
    if kind == 'mt':   return fwd(m, z, kind='mt')
    N = len(z)
    return gt(m, kind, np.repeat(z, N, 0), np.tile(z, (N, 1))).reshape(N, N).sum(1)

def r_metrics(y, s):
    N = len(y)
    order = np.argsort(-s) if spearmanr(s, y).statistic >= 0 else np.argsort(s)
    y_ord = y[order]
    tr = np.empty(N, int); tr[np.argsort(-y)] = np.arange(N)
    pr = np.empty(N, int); pr[order] = np.arange(N)
    sp, kt = spearmanr(pr, tr).statistic, kendalltau(pr, tr).statistic
    d = lambda k: 1.0 / np.log2(np.arange(2, k + 2))
    ym = y.min() - 1e-9
    ndcg = {k: ((y_ord[:k]-ym)*d(k)).sum()/((np.sort(y)[::-1][:k]-ym)*d(k)).sum() for k in KS if k <= N}
    tt = np.argsort(-y)
    topk = {k: len(set(tt[:k]) & set(order[:k]))/k for k in KS if k <= N}
    bm = np.array([y_ord[i*(N//10):(i+1)*(N//10)].mean() for i in range(10)])
    return dict(N=N, desc=spearmanr(s, y).statistic > 0, sp=sp, kt=kt, inv=(1-kt)/2,
                ndcg=ndcg, topk=topk, adj=int((np.diff(y_ord) < 0).sum()),
                ht=float(y_ord[:N//10].mean() - y_ord[-N//10:].mean()),
                bv=int((np.diff(bm) < 0).sum()),
                y_head=y_ord[:5].tolist(), y_tail=y_ord[-5:].tolist())

def triplet(m, kind, A, B, C):
    if kind in ('reg', 'mt'):
        fa = fwd(m, A, kind=kind); fb = fwd(m, B, kind=kind); fc = fwd(m, C, kind=kind)
        return fa > fb, fb > fa, fb > fc, fc > fa, fa > fc
    return (fwd(m, A, B, kind=kind) > 0, fwd(m, B, A, kind=kind) > 0,
            fwd(m, B, C, kind=kind) > 0, fwd(m, C, A, kind=kind) > 0,
            fwd(m, A, C, kind=kind) > 0)

def t_metrics(ab, ba, bc, ca, ac):
    cyc_fwd = (ab & bc & ca).mean()
    cyc_bwd = (~ab & ~bc & ~ca).mean()
    conc    = 1.0 - cyc_fwd - cyc_bwd
    chain   = (ab & bc & ac).mean()
    return dict(
        antisym = (ab ^ ba).mean(),
        bt      = (ab & ba).mean(),
        bf      = (~ab & ~ba).mean(),
        chain   = chain,
        cov     = (ab & bc).mean(),
        trans   = 1.0 - cyc_fwd - cyc_bwd,
        cyc_fwd = cyc_fwd,
        cyc_bwd = cyc_bwd,
        cyc     = cyc_fwd + cyc_bwd,
        tau     = (conc - cyc_fwd - cyc_bwd) / (conc + cyc_fwd + cyc_bwd + 1e-12),
    )


def run_triplet():
    with h5py.File(H5, "r") as f:
        for key in sorted(k for k in f if k.startswith("mu_")):
            v, mu = key[3:], f[key][:]
            rng = np.random.default_rng(0); n = len(mu)
            A, B, C = mu[rng.integers(0, n, K)], mu[rng.integers(0, n, K)], mu[rng.integers(0, n, K)]
            print(f"\n[{v}] triplet  N={n:,}  d={mu.shape[1]}")
            for name, cls, kind in MODELS:
                try:
                    m = load_model(f"{MD}/{name}_{v}_best.pt", cls, device)
                    r = t_metrics(*triplet(m, kind, A, B, C))
                    print(f"  {name:8s} antisym={r['antisym']:.4f} (bt={r['bt']:.4f} bf={r['bf']:.4f})  "
                          f"chain={r['chain']:.4f}  trans={r['trans']:.4f}  "
                          f"cyc={r['cyc']:.4f} ({r['cyc_fwd']:.4f}/{r['cyc_bwd']:.4f})  "
                          f"tau={r['tau']:+.4f}")
                    del m
                except Exception as e:
                    print(f"  {name:8s} skip: {e}")
            del mu


def run_ranking(df, target='Fa_mean'):
    y = df[target].values.astype(np.float32); out = {}
    for v in VAES:
        col = f'latent_{v}'
        if col not in df.columns: continue
        z = np.stack(df[col].values).astype(np.float32)
        print(f"\n[{v}] ranking  N={len(z)}")
        for name, cls, kind in MODELS:
            try:
                m = load_model(f"{MD}/{name}_{v}_best.pt", cls, device)
                r = r_metrics(y, score(m, kind, z))
                print(f"  {name:8s} sp={r['sp']:+.4f} kt={r['kt']:+.4f} inv%={r['inv']*100:5.2f} "
                      f"adj={r['adj']:3d} bv={r['bv']} ht={r['ht']:+7.2f} "
                      f"ndcg10={r['ndcg'].get(10,0):.3f} top10={r['topk'].get(10,0):.3f}")
                out[(v, name)] = {k: r[k] for k in ('N','desc','sp','kt','inv','adj','bv','ht')}
                out[(v, name)]['ndcg10'] = r['ndcg'].get(10, np.nan)
                out[(v, name)]['top10'] = r['topk'].get(10, np.nan)
                del m
            except Exception as e:
                print(f"  {name:8s} skip: {e}")
    return pd.DataFrame(out).T


run_triplet()
print("\n", run_ranking(df_test))
```

```python
def _cmp(model, Z, ia, ib, bs, device):
    out = torch.empty(len(ia), dtype=torch.bool)
    for s in range(0, len(ia), bs):
        e = min(s + bs, len(ia))
        x = Z[ia[s:e]].to(device, non_blocking=True)
        y = Z[ib[s:e]].to(device, non_blocking=True)
        with torch.inference_mode():
            out[s:e] = (model(x, y).view(-1) > 0).cpu()
    return out


def quicksort(Z, model, pivot="mid", bs=65536, verbose=False, device=None):
    device = device or torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    N = len(Z)
    if N < 2: return np.arange(N, dtype=np.int64)
    order = torch.arange(N); lo = torch.tensor([0]); hi = torch.tensor([N - 1])
    it = tqdm(desc=f"[qsort/{pivot}]") if verbose else None
    try:
        while lo.numel():
            sz = hi - lo + 1; m = sz >= 2
            if not m.any(): break
            a_lo, a_hi, a_len = lo[m], hi[m], sz[m]
            S, T = a_lo.numel(), int(a_len.sum())
            psid = torch.repeat_interleave(torch.arange(S), a_len)
            cum = torch.cumsum(a_len, 0) - a_len
            pos = a_lo[psid] + torch.arange(T) - cum[psid]
            piv = (((a_lo + a_hi) // 2) if pivot == "mid"
                   else (a_lo + (torch.rand(S) * a_len.float()).long().clamp(max=a_len - 1)))[psid]
            ip = pos == piv
            gt = _cmp(model, Z, order[pos], order[piv], bs, device)
            gt[ip] = False
            w = ~gt & ~ip
            z = torch.zeros(1, dtype=torch.long)
            cg = torch.cat([z, gt.long().cumsum(0)]); cw = torch.cat([z, w.long().cumsum(0)])
            nb = cg[cum + a_len] - cg[cum]; L, NB = a_lo[psid], nb[psid]
            npos = torch.empty(T, dtype=torch.long)
            npos[gt] = L[gt] + cg[1:][gt] - cg[cum][psid][gt] - 1
            npos[w]  = L[w]  + NB[w] + cw[1:][w] - cw[cum][psid][w]
            npos[ip] = L[ip] + NB[ip]
            new_order = order.clone(); new_order.index_copy_(0, npos, order[pos]); order = new_order
            sp = a_lo + nb; ml, mr = nb > 0, a_len - nb - 1 > 0
            lo = torch.cat([a_lo[ml], (sp + 1)[mr], lo[~m]])
            hi = torch.cat([(sp - 1)[ml], a_hi[mr], hi[~m]])
            if it is not None: it.update(1)
    finally:
        if it is not None: it.close()
    return order.numpy()


def score_sort(Z, model, bs=65536, kind="ds", verbose=False, device=None):
    device = device or torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    N = len(Z); pred = np.empty(N, dtype=np.float32)
    it = tqdm(total=N, desc=f"[{kind}]") if verbose else None
    try:
        for s in range(0, N, bs):
            e = min(s + bs, N)
            x = Z[s:e].to(device, non_blocking=True)
            with torch.inference_mode():
                out = model.score(x) if kind == "ds" else model(x)[:, 0]
            pred[s:e] = out.view(-1).cpu().numpy()
            if it is not None: it.update(e - s)
    finally:
        if it is not None: it.close()
    return np.argsort(-pred).astype(np.int64)


def reg_sort(Z, model, bs=65536, mode="mean", beta=1.0, verbose=False, device=None):
    device = device or torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    N = len(Z); pred = np.empty(N, dtype=np.float32)
    it = tqdm(total=N, desc=f"[reg/{mode}]") if verbose else None
    try:
        for s in range(0, N, bs):
            e = min(s + bs, N)
            x = Z[s:e].to(device, non_blocking=True)
            with torch.inference_mode():
                out = model(x)
            if isinstance(out, tuple):
                mean = out[0].view(-1)
                logvar = out[1].view(-1) if len(out) > 1 else None
                if logvar is not None and mode != "mean":
                    std = torch.exp(0.5 * logvar)
                    v = (mean - beta * std) if mode == "lcb" else (mean + beta * std)
                else:
                    v = mean
            else:
                v = out.view(-1)
            pred[s:e] = v.cpu().numpy()
            if it is not None: it.update(e - s)
    finally:
        if it is not None: it.close()
    return np.argsort(-pred).astype(np.int64)


def _ndcg(idx, y, k):
    k = min(k, len(idx))
    g = y[idx[:k]] - y.min() + 1e-6
    d = 1.0 / np.log2(np.arange(2, k + 2))
    ideal = np.sort(y)[::-1][:k] - y.min() + 1e-6
    return float((g * d).sum() / (ideal * d).sum())


def _kendall_sampled(a, b, n=500_000, rng=None):
    N = len(a)
    rng = rng or np.random.default_rng(0)
    i = rng.integers(0, N, n); j = rng.integers(0, N, n)
    k = i != j
    return float((np.sign(a[i[k]] - a[j[k]]) * np.sign(b[i[k]] - b[j[k]])).mean())


def eval_vs_target(idx, y):
    r = np.empty(len(idx), dtype=np.int64); r[idx] = np.arange(len(idx)) + 1
    t = rankdata(-y)
    return dict(kendall=float(_kendall_sampled(r, t)),
                spearman=float(spearmanr(r, t).statistic),
                ndcg10=_ndcg(idx, y, 10),
                ndcg100=_ndcg(idx, y, 100),
                ndcg1000=_ndcg(idx, y, 1000))


def cross_agreement(a, b):
    N = len(a)
    ra = np.empty(N, np.int32); ra[a] = np.arange(N)
    rb = np.empty(N, np.int32); rb[b] = np.arange(N)
    return dict(kendall=_kendall_sampled(ra, rb),
                spearman=float(spearmanr(ra, rb).statistic),
                top100=len(set(a[:100]) & set(b[:100])) / 100)


METHODS = [
    ("ps_qmid",     "ps",  lambda Z, m, bs, v, d: quicksort(Z, m, "mid",    bs, v, d)),
    ("ps_qrand",    "ps",  lambda Z, m, bs, v, d: quicksort(Z, m, "random", bs, v, d)),
    ("aps_qmid",    "aps", lambda Z, m, bs, v, d: quicksort(Z, m, "mid",    bs, v, d)),
    ("aps_qrand",   "aps", lambda Z, m, bs, v, d: quicksort(Z, m, "random", bs, v, d)),
    ("ds_qmid",     "ds",  lambda Z, m, bs, v, d: quicksort(Z, m, "mid",    bs, v, d)),
    ("ds_qrand",    "ds",  lambda Z, m, bs, v, d: quicksort(Z, m, "random", bs, v, d)),
    ("ds_fast",     "ds",  lambda Z, m, bs, v, d: score_sort(Z, m, bs, "ds", v, d)),
    ("reg_mlp",     "mlp", lambda Z, m, bs, v, d: reg_sort(Z, m, bs, "mean", 1.0, v, d)),
    ("reg_nll",     "nll", lambda Z, m, bs, v, d: reg_sort(Z, m, bs, "mean", 1.0, v, d)),
    ("reg_nll_lcb", "nll", lambda Z, m, bs, v, d: reg_sort(Z, m, bs, "lcb",  1.0, v, d)),
    ("mtm_fa",      "mt",  lambda Z, m, bs, v, d: score_sort(Z, m, bs, "mt", v, d)),
]


def run_all_and_save(h5_path, models, device, y_key="Fa_mean",
                     bs=262144, overwrite=True, verbose=True):
    idx_all = {}
    with h5py.File(h5_path, "r+") as f:
        for k in [k for k in f if k.startswith("mu_")]:
            vae = k[3:]
            if vae not in models: continue
            Z = torch.from_numpy(np.ascontiguousarray(f[k][:], np.float32)).pin_memory()
            grp = f.require_group(f"sorted_indices/{vae}")
            idx_all[vae] = {}
            for name, kind, fn in METHODS:
                if kind not in models[vae]: continue
                if name in grp and not overwrite:
                    idx = grp[name][:]
                else:
                    if verbose: print(f"[{vae}] {name}")
                    idx = fn(Z, models[vae][kind], bs, verbose, device)
                    if name in grp: del grp[name]
                    grp.create_dataset(name, data=idx.astype(np.int32), compression="lzf")
                idx_all[vae][name] = idx
            del Z; gc.collect()
            if device.type == "cuda": torch.cuda.empty_cache()

        if y_key and y_key in f.attrs:
            y = f.attrs[y_key][:]
            rows = [dict(VAE=v, Method=m, **eval_vs_target(idx_all[v][m], y))
                    for v in idx_all for m in idx_all[v]
                    if len(y) == len(idx_all[v][m])]
            if rows:
                print("\n=== vs target ===")
                print(pd.DataFrame(rows).to_string(index=False, float_format="%.4f"))

    print("\n=== cross-method (Kendall, 500k) ===")
    for v, per in idx_all.items():
        n = list(per)
        M = pd.DataFrame(np.eye(len(n)), index=n, columns=n)
        for i, a in enumerate(n):
            for j, b in enumerate(n):
                if i < j:
                    M.iloc[i, j] = M.iloc[j, i] = cross_agreement(per[a], per[b])["kendall"]
        print(f"\n[{v}]")
        print(M.to_string(float_format="%.3f"))

    print("\n=== cross-VAE (per method) ===")
    for name, _, _ in METHODS:
        vs = [v for v in idx_all if name in idx_all[v]]
        for i, a in enumerate(vs):
            for b in vs[i + 1:]:
                m = cross_agreement(idx_all[a][name], idx_all[b][name])
                print(f"[{name}] {a} vs {b}: τ={m['kendall']:+.3f} "
                      f"ρ={m['spearman']:+.3f} top100={m['top100']:.3f}")
    return idx_all
```

```python
def _cmp(model, a, b, bs=65536):
    out = torch.empty(len(a), dtype=torch.bool, device=a.device)
    for s in range(0, len(a), bs):
        e = min(s + bs, len(a))
        with torch.inference_mode():
            out[s:e] = model(a[s:e], b[s:e]).view(-1) > 0
    return out


def quicksort(Z, model, pivot="mid", bs=65536, verbose=False):
    N = len(Z)
    if N < 2: return np.arange(N, dtype=np.int64)
    order = torch.arange(N, device=Z.device)
    lo = torch.tensor([0], device=Z.device)
    hi = torch.tensor([N - 1], device=Z.device)
    it = tqdm(desc=f"[qsort/{pivot}]") if verbose else None
    try:
        while lo.numel():
            sz = hi - lo + 1
            m = sz >= 2
            if not m.any(): break
            a_lo, a_hi, a_len = lo[m], hi[m], sz[m]
            S, T = a_lo.numel(), int(a_len.sum())
            psid = torch.repeat_interleave(torch.arange(S, device=Z.device), a_len)
            cum = torch.cumsum(a_len, 0) - a_len
            pos = a_lo[psid] + torch.arange(T, device=Z.device) - cum[psid]
            if pivot == "mid":
                piv = ((a_lo + a_hi) // 2)[psid]
            else:
                r = (torch.rand(S, device=Z.device) * a_len.float()).long().clamp(max=a_len - 1)
                piv = (a_lo + r)[psid]
            ip = pos == piv
            gt = _cmp(model, Z[order[pos]], Z[order[piv]], bs)
            gt[ip] = False
            w = ~gt & ~ip
            z = torch.zeros(1, dtype=torch.long, device=Z.device)
            cg = torch.cat([z, gt.long().cumsum(0)])
            cw = torch.cat([z, w.long().cumsum(0)])
            nb = cg[cum + a_len] - cg[cum]
            L, NB = a_lo[psid], nb[psid]
            npos = torch.empty(T, dtype=torch.long, device=Z.device)
            npos[gt] = L[gt] + cg[1:][gt] - cg[cum][psid][gt] - 1
            npos[w]  = L[w]  + NB[w] + cw[1:][w] - cw[cum][psid][w]
            npos[ip] = L[ip] + NB[ip]
            new_order = order.clone()
            new_order.index_copy_(0, npos, order[pos])
            order = new_order
            sp = a_lo + nb
            ml, mr = nb > 0, a_len - nb - 1 > 0
            lo = torch.cat([a_lo[ml], (sp + 1)[mr], lo[~m]])
            hi = torch.cat([(sp - 1)[ml], a_hi[mr], hi[~m]])
            if it is not None: it.update(1)
    finally:
        if it is not None: it.close()
    return order.cpu().numpy()


def score_sort(Z, model, bs=65536, kind="ds", verbose=False):
    N = len(Z); pred = np.empty(N, dtype=np.float32)
    it = tqdm(total=N, desc=f"[{kind}]") if verbose else None
    try:
        for s in range(0, N, bs):
            e = min(s + bs, N)
            with torch.inference_mode():
                out = model.score(Z[s:e]) if kind == "ds" else model(Z[s:e])[:, 0]
            pred[s:e] = out.view(-1).cpu().numpy()
            if it is not None: it.update(e - s)
    finally:
        if it is not None: it.close()
    return np.argsort(-pred).astype(np.int64)


def reg_sort(Z, model, bs=65536, mode="mean", beta=1.0, verbose=False):
    N = len(Z); pred = np.empty(N, dtype=np.float32)
    it = tqdm(total=N, desc=f"[reg/{mode}]") if verbose else None
    try:
        for s in range(0, N, bs):
            e = min(s + bs, N)
            with torch.inference_mode():
                out = model(Z[s:e])
            if isinstance(out, tuple):
                mean = out[0].view(-1)
                logvar = out[1].view(-1) if len(out) > 1 else None
                if logvar is not None and mode != "mean":
                    std = torch.exp(0.5 * logvar)
                    v = (mean - beta * std) if mode == "lcb" else (mean + beta * std)
                else:
                    v = mean
            else:
                v = out.view(-1)
            pred[s:e] = v.cpu().numpy()
            if it is not None: it.update(e - s)
    finally:
        if it is not None: it.close()
    return np.argsort(-pred).astype(np.int64)


def _ndcg(idx, y, k):
    k = min(k, len(idx))
    g = y[idx[:k]] - y.min() + 1e-6
    d = 1.0 / np.log2(np.arange(2, k + 2))
    ideal = np.sort(y)[::-1][:k] - y.min() + 1e-6
    return float((g * d).sum() / (ideal * d).sum())


def _kendall_sampled(a, b, n=500_000, rng=None):
    N = len(a)
    rng = rng or np.random.default_rng(0)
    i = rng.integers(0, N, n); j = rng.integers(0, N, n)
    k = i != j
    return float((np.sign(a[i[k]] - a[j[k]]) * np.sign(b[i[k]] - b[j[k]])).mean())


def eval_vs_target(idx, y):
    r = np.empty(len(idx), dtype=np.int64); r[idx] = np.arange(len(idx)) + 1
    t = rankdata(-y)
    return dict(kendall=float(_kendall_sampled(r, t)),
                spearman=float(spearmanr(r, t).statistic),
                ndcg10=_ndcg(idx, y, 10),
                ndcg100=_ndcg(idx, y, 100),
                ndcg1000=_ndcg(idx, y, 1000))


def cross_agreement(a, b):
    N = len(a)
    ra = np.empty(N, np.int32); ra[a] = np.arange(N)
    rb = np.empty(N, np.int32); rb[b] = np.arange(N)
    return dict(kendall=_kendall_sampled(ra, rb),
                spearman=float(spearmanr(ra, rb).statistic),
                top100=len(set(a[:100]) & set(b[:100])) / 100)


METHODS = [
    ("ps_qmid",     "pairwise", lambda Z, m, bs, v: quicksort(Z, m, "mid",    bs, v)),
    ("ps_qrand",    "pairwise", lambda Z, m, bs, v: quicksort(Z, m, "random", bs, v)),
    ("ds_qmid",     "ds",       lambda Z, m, bs, v: quicksort(Z, m, "mid",    bs, v)),
    ("ds_qrand",    "ds",       lambda Z, m, bs, v: quicksort(Z, m, "random", bs, v)),
    ("ds_fast",     "ds",       lambda Z, m, bs, v: score_sort(Z, m, bs, "ds", v)),
    ("reg_mlp",     "mlp",      lambda Z, m, bs, v: reg_sort(Z, m, bs, "mean", 1.0, v)),
    ("reg_nll",     "nll",      lambda Z, m, bs, v: reg_sort(Z, m, bs, "mean", 1.0, v)),
    ("reg_nll_lcb", "nll",      lambda Z, m, bs, v: reg_sort(Z, m, bs, "lcb",  1.0, v)),
    ("mtm_fa",      "mt",       lambda Z, m, bs, v: score_sort(Z, m, bs, "mt", v)),
]


def run_all_and_save(h5_path, models, device, y_key="Fa_mean",
                     bs=65536, overwrite=True, verbose=True):
    idx_all = {}
    with h5py.File(h5_path, "r+") as f:
        for k in [k for k in f if k.startswith("mu_")]:
            vae = k[3:]
            if vae not in models: continue
            Z = torch.from_numpy(np.ascontiguousarray(f[k][:], np.float32)).to(device)
            grp = f.require_group(f"sorted_indices/{vae}")
            idx_all[vae] = {}
            for name, kind, fn in METHODS:
                if kind not in models[vae]: continue
                if name in grp and not overwrite:
                    idx = grp[name][:]
                else:
                    if verbose: print(f"[{vae}] {name}")
                    idx = fn(Z, models[vae][kind], bs, verbose)
                    if name in grp: del grp[name]
                    grp.create_dataset(name, data=idx.astype(np.int32), compression="lzf")
                idx_all[vae][name] = idx
            del Z
            if device.type == "cuda": torch.cuda.empty_cache()

        if y_key and y_key in f.attrs:
            y = f.attrs[y_key][:]
            rows = [dict(VAE=v, Method=m, **eval_vs_target(idx_all[v][m], y))
                    for v in idx_all for m in idx_all[v]
                    if len(y) == len(idx_all[v][m])]
            if rows:
                print("\n=== vs target ===")
                print(pd.DataFrame(rows).to_string(index=False, float_format="%.4f"))

    print("\n=== cross-method (Kendall, 500k) ===")
    for v, per in idx_all.items():
        n = list(per)
        M = pd.DataFrame(np.eye(len(n)), index=n, columns=n)
        for i, a in enumerate(n):
            for j, b in enumerate(n):
                if i < j:
                    M.iloc[i, j] = M.iloc[j, i] = cross_agreement(per[a], per[b])["kendall"]
        print(f"\n[{v}]")
        print(M.to_string(float_format="%.3f"))

    print("\n=== cross-VAE (per method) ===")
    for name, _, _ in METHODS:
        vs = [v for v in idx_all if name in idx_all[v]]
        for i, a in enumerate(vs):
            for b in vs[i + 1:]:
                m = cross_agreement(idx_all[a][name], idx_all[b][name])
                print(f"[{name}] {a} vs {b}: τ={m['kendall']:+.3f} "
                      f"ρ={m['spearman']:+.3f} top100={m['top100']:.3f}")
    return idx_all
```

```python
import numpy as np
import pandas as pd
import h5py
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import spearmanr


def _ranks(idx):
    r = np.empty(len(idx), dtype=np.int32)
    r[idx] = np.arange(len(idx), dtype=np.int32)
    return r


def _kendall_sampled(a, b, n=500_000, rng=None):
    N = len(a)
    rng = rng or np.random.default_rng(0)
    i, j = rng.integers(0, N, n), rng.integers(0, N, n)
    k = i != j
    return float((np.sign(a[i[k]] - a[j[k]]) * np.sign(b[i[k]] - b[j[k]])).mean())


def agree(a, b, N, top_ks=(100, 1000, 10_000)):
    ra, rb = _ranks(a), _ranks(b)
    out = dict(
        kendall=_kendall_sampled(ra, rb),
        spearman=float(spearmanr(ra, rb).statistic),
    )
    for k in top_ks:
        if k > len(a) or k > len(b):
            continue
        observed = len(set(a[:k].tolist()) & set(b[:k].tolist())) / k
        expected = k / N  # under independence, E[overlap] = k/N
        out[f"top{k}_enrichment"] = observed / expected if expected > 0 else float("nan")
    return out


def _matrix(names, get):
    M = pd.DataFrame(np.eye(len(names)), index=names, columns=names)
    for i, a in enumerate(names):
        for j, b in enumerate(names):
            if i < j:
                M.iloc[i, j] = M.iloc[j, i] = get(a, b)
    return M


def compute_matrices(h5_path, metric="kendall"):
    with h5py.File(h5_path, "r") as f:
        root = f["sorted_indices"]
        vaes = sorted(root.keys())
        methods = sorted(root[vaes[0]].keys())
        N = len(f[f"mu_{vaes[0]}"])

        cross_method = {
            v: _matrix(
                methods,
                lambda a, b, v=v: agree(
                    root[v][a][:].astype(np.int32),
                    root[v][b][:].astype(np.int32),
                    N,
                )[metric],
            )
            for v in vaes
        }

        pairs = [(vaes[i], vaes[j]) for i in range(len(vaes)) for j in range(i + 1, len(vaes))]
        cols = [f"{a}×{b}" for a, b in pairs]
        keys = ("kendall", "spearman", "top100_enrichment")
        cross_vae = {k: pd.DataFrame(index=methods, columns=cols, dtype=float) for k in keys}

        for m in methods:
            for a, b in pairs:
                r = agree(
                    root[a][m][:].astype(np.int32),
                    root[b][m][:].astype(np.int32),
                    N,
                    top_ks=(100,),
                )
                c = f"{a}×{b}"
                for k in keys:
                    cross_vae[k].loc[m, c] = r[k]

    return vaes, methods, cross_method, cross_vae


# --- plotting ---

def _heatmap(ax, M, title, fmt=".3f", center=None):
    Mv = M.astype(float)
    vmin, vmax = float(Mv.min().min()), float(Mv.max().max())
    pad = max((vmax - vmin) * 0.05, 1e-6)
    sns.heatmap(Mv, ax=ax, annot=True, fmt=fmt, cmap="RdYlGn", center=center,
                vmin=vmin - pad, vmax=vmax + pad, cbar_kws={"shrink": 0.7},
                square=True, annot_kws={"size": 7})
    ax.set_title(title, fontsize=10)
    ax.tick_params(axis="x", rotation=45, labelsize=8)
    ax.tick_params(axis="y", rotation=0, labelsize=8)


def plot_cross_method(cross_method, save=None):
    vaes = list(cross_method)
    fig, axes = plt.subplots(1, len(vaes), figsize=(5.2 * len(vaes), 4.6))
    for ax, v in zip(np.atleast_1d(axes), vaes):
        _heatmap(ax, cross_method[v], f"[{v}] cross-method (Kendall)")
    fig.suptitle("Cross-method Kendall τ per VAE", fontsize=12, y=1.02)
    fig.tight_layout()
    if save:
        fig.savefig(save, dpi=150, bbox_inches="tight")
    plt.show()


def plot_cross_vae(cross_vae, save=None):
    panels = [
        ("kendall", "Kendall τ", 0.0),
        ("spearman", "Spearman ρ", 0.0),
        ("top100_enrichment", "top-100 enrichment (1.0=случайно)", 1.0),
    ]
    fig, axes = plt.subplots(1, 3, figsize=(15, 4.6))
    for ax, (k, t, center) in zip(axes, panels):
        _heatmap(ax, cross_vae[k], f"Cross-VAE — {t}", center=center)
    fig.suptitle("Cross-VAE agreement per method", fontsize=12, y=1.02)
    fig.tight_layout()
    if save:
        fig.savefig(save, dpi=150, bbox_inches="tight")
    plt.show()


def plot_method_bars(cross_vae, metric="kendall", save=None):
    fig, ax = plt.subplots(figsize=(8, 4.5))
    cross_vae[metric].plot(kind="bar", ax=ax, width=0.8, edgecolor="black")
    ax.set_ylabel(metric)
    ax.set_xlabel("method")
    ax.axhline(1.0 if "enrichment" in metric else 0.0, color="gray", ls="--", lw=1)
    ax.set_title(f"Cross-VAE {metric} per method")
    ax.legend(title="VAE pair", fontsize=8, loc="lower right")
    ax.grid(axis="y", alpha=0.3)
    plt.xticks(rotation=30, ha="right")
    fig.tight_layout()
    if save:
        fig.savefig(save, dpi=150, bbox_inches="tight")
    plt.show()


def plot_summary(cross_vae, save=None):
    k = cross_vae["kendall"]
    fig, axes = plt.subplots(1, 2, figsize=(13, 4.2))
    for ax, s, c, t in [
        (axes[0], k.mean(0), "steelblue", "mean Kendall per VAE pair"),
        (axes[1], k.mean(1), "indianred", "mean Kendall per method"),
    ]:
        s.sort_values().plot.barh(ax=ax, color=c, edgecolor="black")
        ax.set_xlabel("mean Kendall τ")
        ax.set_title(t)
        ax.grid(axis="x", alpha=0.3)
    fig.tight_layout()
    if save:
        fig.savefig(save, dpi=150, bbox_inches="tight")
    plt.show()


def report_and_plot(h5_path, save_dir=None, metric="kendall"):
    save_dir = save_dir or "."
    vaes, methods, cross_method, cross_vae = compute_matrices(h5_path, metric=metric)
    print(f"VAEs   : {vaes}\nmethods: {methods}\n")
    for v, M in cross_method.items():
        print(f"[{v}] cross-method ({metric}):")
        print(M.to_string(float_format="%.3f"), "\n")
    for k in ("kendall", "spearman", "top100_enrichment"):
        print(f"cross-VAE ({k}):")
        print(cross_vae[k].to_string(float_format="%.3f"), "\n")
    plot_cross_method(cross_method, save=f"{save_dir}/cross_method.png")
    plot_cross_vae(cross_vae,       save=f"{save_dir}/cross_vae.png")
    plot_method_bars(cross_vae,     save=f"{save_dir}/cross_vae_bars.png")
    plot_summary(cross_vae,         save=f"{save_dir}/cross_vae_summary.png")
    return cross_method, cross_vae
```

```python
import h5py
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import torch

%matplotlib inline
sns.set_theme(style="whitegrid")
plt.rcParams['figure.dpi'] = 150
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 3000)

H5_PATH = "./data/DL/latent_db_all.h5"
MD = "./data/DL/models"
TOP_K = 5
VAES = ['CLS', 'DS', 'SET']
BETA = 1.0
DEV = torch.device('cuda' if torch.cuda.is_available() else 'cpu')


def build_pool(sorted_grp, vae):
    methods = list(sorted_grp[vae].keys())
    cand = {m: sorted_grp[vae][m][:TOP_K].tolist() for m in methods}
    return methods, cand, sorted(set().union(*cand.values()))


def infer(m_mlp, m_nll, Z):
    with torch.inference_mode():
        s_mlp = m_mlp(Z).view(-1).cpu().numpy()
        out = m_nll(Z)
        if isinstance(out, tuple):
            mu = out[0].view(-1).cpu().numpy()
            logvar = out[1].view(-1).cpu().numpy()
        else:
            mu = out.view(-1).cpu().numpy()
            logvar = np.zeros_like(mu)
    return s_mlp, mu, np.exp(0.5 * logvar)


def analyze(f, sorted_grp, vae):
    methods, cand, union = build_pool(sorted_grp, vae)
    m_mlp = load_model(f"{MD}/MLP_{vae}_best.pt", MLP, DEV).eval()
    m_nll = load_model(f"{MD}/MLPNLL_{vae}_best.pt", MLPNLL, DEV).eval()
    Z = torch.from_numpy(f[f"mu_{vae}"][:][union]).to(DEV)
    s_mlp, mu, std = infer(m_mlp, m_nll, Z)
    df = pd.DataFrame({'idx': union, 'mlp': s_mlp, 'mu': mu, 'std': std,
                       'ucb': mu + BETA * std})
    for m in methods:
        df[f'in_{m}'] = df['idx'].isin(cand[m])
    df = df.sort_values('ucb', ascending=False).reset_index(drop=True)
    del m_mlp, m_nll, Z
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    return df


def report(df, vae):
    top = df.iloc[0]
    risk = df.loc[df['std'].idxmax()]
    print(f"[{vae}] |pool|={len(df)}  max_mu={df['mu'].max():.4f}  "
          f"max_UCB={df['ucb'].max():.4f}  "
          f"argmax_idx={int(top['idx'])}  "
          f"max_sigma={risk['std']:.2f}@idx{int(risk['idx'])}")


def plot(df, vae):
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.errorbar(df['mu'], df['mlp'], xerr=df['std'], fmt='o',
                markersize=8, alpha=0.85, capsize=4, elinewidth=1.5,
                color='#3b6ea5', ecolor='#a8c1e0')
    ax.axvline(df['mu'].median(), color='gray', ls='--', lw=0.8, alpha=0.5)
    ax.axhline(df['mlp'].median(), color='gray', ls='--', lw=0.8, alpha=0.5)
    ax.annotate(f"argmax UCB={df['ucb'].max():.4f}",
                xy=(df['mu'].iloc[0], df['mlp'].iloc[0]),
                xytext=(8, 8), textcoords='offset points', fontsize=9)
    ax.set_title(f"Candidate landscape [{vae}]", fontsize=13, fontweight='bold', pad=12)
    ax.set_xlabel("Expected quality (MLPNLL mu) -> higher is better")
    ax.set_ylabel("Base regressor score (MLP) -> higher is better")
    plt.tight_layout()
    plt.show()


with h5py.File(H5_PATH, "r") as f:
    sorted_grp = f["sorted_indices"]
    for vae in VAES:
        if vae not in sorted_grp or f"mu_{vae}" not in f:
            continue
        df = analyze(f, sorted_grp, vae)
        print(f"\n=== {vae} ===")
        print(df.to_string(index=False))
        report(df, vae)
        plot(df, vae)
```

```python
import h5py
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import torch

%matplotlib inline
sns.set_theme(style="whitegrid")
plt.rcParams['figure.dpi'] = 150
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 3000)

H5_PATH = "./data/DL/latent_db_all.h5"
MD = "./data/DL/models"
TOP_K = 25          # было 5 -- слишком маленький пул для честного отбора
N_FINAL = 10        # сколько финальных кандидатов на VAE отдать в тест
MIN_DIST = 0.15      # порог diversity-фильтра по составу (L1 на долях)
VAES = ['CLS', 'DS', 'SET']
BETA = 1.0
DEV = torch.device('cuda' if torch.cuda.is_available() else 'cpu')


def dedup_mask(Z, decimals=5):
    _, inv = np.unique(np.round(Z, decimals), axis=0, return_inverse=True)
    return pd.Series(inv).duplicated(keep="first").to_numpy()


def decode_composition(row, names, smiles, numbers, eps=1e-6):
    nz = np.where(row > eps)[0]
    order = nz[np.argsort(-row[nz])]
    return [{"name": names[i], "smiles": smiles[i], "number": int(numbers[i]), "fraction": float(row[i])}
            for i in order]


def composition_key(row, decimals=3, eps=1e-6):
    nz = np.where(row > eps)[0]
    order = np.argsort(nz)
    return tuple(zip(nz[order].tolist(), np.round(row[nz], decimals)[order].tolist()))


def composition_distance(a, b):
    da = {c["number"]: c["fraction"] for c in a}
    db = {c["number"]: c["fraction"] for c in b}
    return sum(abs(da.get(k, 0.0) - db.get(k, 0.0)) for k in set(da) | set(db))


def build_pool(sorted_grp, vae, top_k=TOP_K):
    methods = list(sorted_grp[vae].keys())
    cand = {m: sorted_grp[vae][m][:top_k].tolist() for m in methods}
    union = sorted(set().union(*cand.values()))
    membership = pd.DataFrame({m: pd.Series(union).isin(cand[m]).to_numpy() for m in methods}, index=union)
    membership["n_methods_agree"] = membership[methods].sum(axis=1)
    return methods, union, membership


def infer(m_mlp, m_nll, Z):
    with torch.inference_mode():
        s_mlp = m_mlp(Z).view(-1).cpu().numpy()
        out = m_nll(Z)
        if isinstance(out, tuple):
            mu, logvar = out[0].view(-1).cpu().numpy(), out[1].view(-1).cpu().numpy()
        else:
            mu, logvar = out.view(-1).cpu().numpy(), np.zeros(len(Z))
    return s_mlp, mu, np.exp(0.5 * logvar)


def analyze(f, sorted_grp, vae):
    methods, union, membership = build_pool(sorted_grp, vae)
    m_mlp = load_model(f"{MD}/MLP_{vae}_best.pt", MLP, DEV).eval()
    m_nll = load_model(f"{MD}/MLPNLL_{vae}_best.pt", MLPNLL, DEV).eval()
    fractions = f["fractions"][:][union]
    names, smiles, numbers = f.attrs["monomer_names"], f.attrs["monomer_smiles"], f.attrs["monomer_numbers"]
    Z_np = f[f"mu_{vae}"][:][union]
    Z = torch.from_numpy(Z_np).to(DEV)
    s_mlp, mu, std = infer(m_mlp, m_nll, Z)

    df = pd.DataFrame({'idx': union, 'vae': vae, 'mlp': s_mlp, 'mu': mu, 'std': std,
                       'ucb': mu + BETA * std,
                       'dup_latent': dedup_mask(Z_np),
                       'dup_composition': pd.Series([composition_key(r) for r in fractions]).duplicated(keep="first").to_numpy()})
    df['composition'] = [decode_composition(r, names, smiles, numbers) for r in fractions]
    df['composition_summary'] = df['composition'].apply(
        lambda c: " / ".join(f"{m['name']}:{m['fraction']:.2f}" for m in c))
    df = df.merge(membership.rename_axis('idx').reset_index(), on='idx')
    df = df.sort_values('ucb', ascending=False).reset_index(drop=True)

    del m_mlp, m_nll, Z
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    return df


def select_diverse_topN(df, N, rank_col='ucb', min_dist=MIN_DIST):
    ranked = df[~df['dup_latent'] & ~df['dup_composition']].sort_values(rank_col, ascending=False)
    selected = []
    for _, row in ranked.iterrows():
        if all(composition_distance(row['composition'], s['composition']) >= min_dist for s in selected):
            selected.append(row)
        if len(selected) >= N:
            break
    return pd.DataFrame(selected)


def report(df, vae, final):
    top = df.iloc[0]
    risk = df.loc[df['std'].idxmax()]
    n_dup_lat, n_dup_comp = int(df['dup_latent'].sum()), int(df['dup_composition'].sum())
    print(f"[{vae}] |pool|={len(df)}  dup_latent={n_dup_lat}  dup_composition={n_dup_comp}  "
          f"max_mu={df['mu'].max():.4f}  max_UCB={df['ucb'].max():.4f}  "
          f"argmax_idx={int(top['idx'])}  max_sigma={risk['std']:.2f}@idx{int(risk['idx'])}  "
          f"selected_final={len(final)}")


def plot(df, vae):
    fig, ax = plt.subplots(figsize=(10, 5))
    clean = df[~df['dup_latent'] & ~df['dup_composition']]
    dup = df[df['dup_latent'] | df['dup_composition']]
    sc = ax.scatter(clean['mu'], clean['mlp'], c=clean['n_methods_agree'], cmap='viridis',
                     s=70, alpha=0.9, edgecolor='k', linewidth=0.4)
    ax.scatter(dup['mu'], dup['mlp'], marker='x', color='red', s=50, alpha=0.6, label='duplicate (excluded)')
    plt.colorbar(sc, ax=ax, label='n_methods_agree')
    ax.axvline(df['mu'].median(), color='gray', ls='--', lw=0.8, alpha=0.5)
    ax.axhline(df['mlp'].median(), color='gray', ls='--', lw=0.8, alpha=0.5)
    ax.set_title(f"Candidate landscape [{vae}]", fontsize=13, fontweight='bold', pad=12)
    ax.set_xlabel("Expected quality (MLPNLL mu) -> higher is better")
    ax.set_ylabel("Base regressor score (MLP) -> higher is better")
    ax.legend(fontsize=8)
    plt.tight_layout()
    plt.show()


final_candidates = {}
with h5py.File(H5_PATH, "r") as f:
    sorted_grp = f["sorted_indices"]
    for vae in VAES:
        if vae not in sorted_grp or f"mu_{vae}" not in f:
            continue
        df = analyze(f, sorted_grp, vae)
        final = select_diverse_topN(df, N_FINAL)
        final_candidates[vae] = final
        print(f"\n=== {vae} ===")
        report(df, vae, final)
        plot(df, vae)
        print(f"\n[{vae}] финальные кандидаты для теста ({len(final)}):")
        print(final[['idx', 'mu', 'std', 'ucb', 'n_methods_agree', 'composition_summary']].to_string(index=False))

final_table = pd.concat(final_candidates.values(), ignore_index=True)
print(f"\n=== ИТОГО по всем VAE: {len(final_table)} кандидатов на тест ===")
final_table[['vae', 'idx', 'mu', 'std', 'ucb', 'n_methods_agree', 'composition_summary']]
```

```python
def _cmp(model, Z, ia, ib, bs, device):
    out = torch.empty(len(ia), dtype=torch.bool)
    for s in range(0, len(ia), bs):
        e = min(s + bs, len(ia))
        x = Z[ia[s:e]].to(device, non_blocking=True)
        y = Z[ib[s:e]].to(device, non_blocking=True)
        with torch.inference_mode():
            out[s:e] = (model(x, y).view(-1) > 0).cpu()
    return out


def quicksort(Z, model, pivot="mid", bs=65536, verbose=False, device=None):
    device = device or torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    N = len(Z)
    if N < 2: return np.arange(N, dtype=np.int64)
    order = torch.arange(N)
    lo = torch.tensor([0]); hi = torch.tensor([N - 1])
    it = tqdm(desc=f"[qsort/{pivot}]") if verbose else None
    try:
        while lo.numel():
            sz = hi - lo + 1; m = sz >= 2
            if not m.any(): break
            a_lo, a_hi, a_len = lo[m], hi[m], sz[m]
            S, T = a_lo.numel(), int(a_len.sum())
            psid = torch.repeat_interleave(torch.arange(S), a_len)
            cum = torch.cumsum(a_len, 0) - a_len
            pos = a_lo[psid] + torch.arange(T) - cum[psid]
            if pivot == "mid":
                piv = ((a_lo + a_hi) // 2)[psid]
            else:
                r = (torch.rand(S) * a_len.float()).long().clamp(max=a_len - 1)
                piv = (a_lo + r)[psid]
            ip = pos == piv
            gt = _cmp(model, Z, order[pos], order[piv], bs, device)
            gt[ip] = False
            w = ~gt & ~ip
            z = torch.zeros(1, dtype=torch.long)
            cg = torch.cat([z, gt.long().cumsum(0)])
            cw = torch.cat([z, w.long().cumsum(0)])
            nb = cg[cum + a_len] - cg[cum]
            L, NB = a_lo[psid], nb[psid]
            npos = torch.empty(T, dtype=torch.long)
            npos[gt] = L[gt] + cg[1:][gt] - cg[cum][psid][gt] - 1
            npos[w]  = L[w]  + NB[w] + cw[1:][w] - cw[cum][psid][w]
            npos[ip] = L[ip] + NB[ip]
            new_order = order.clone()
            new_order.index_copy_(0, npos, order[pos])
            order = new_order
            sp = a_lo + nb; ml, mr = nb > 0, a_len - nb - 1 > 0
            lo = torch.cat([a_lo[ml], (sp + 1)[mr], lo[~m]])
            hi = torch.cat([(sp - 1)[ml], a_hi[mr], hi[~m]])
            if it is not None: it.update(1)
    finally:
        if it is not None: it.close()
    return order.numpy()


def score_sort(Z, model, bs=65536, kind="ds", verbose=False, device=None):
    device = device or torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    N = len(Z); pred = np.empty(N, dtype=np.float32)
    it = tqdm(total=N, desc=f"[{kind}]") if verbose else None
    try:
        for s in range(0, N, bs):
            e = min(s + bs, N)
            x = Z[s:e].to(device, non_blocking=True)
            with torch.inference_mode():
                out = model.score(x) if kind == "ds" else model(x)[:, 0]
            pred[s:e] = out.view(-1).cpu().numpy()
            if it is not None: it.update(e - s)
    finally:
        if it is not None: it.close()
    return np.argsort(-pred).astype(np.int64)


def reg_sort(Z, model, bs=65536, mode="mean", beta=1.0, verbose=False, device=None):
    device = device or torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    N = len(Z); pred = np.empty(N, dtype=np.float32)
    it = tqdm(total=N, desc=f"[reg/{mode}]") if verbose else None
    try:
        for s in range(0, N, bs):
            e = min(s + bs, N)
            x = Z[s:e].to(device, non_blocking=True)
            with torch.inference_mode():
                out = model(x)
            if isinstance(out, tuple):
                mean = out[0].view(-1)
                logvar = out[1].view(-1) if len(out) > 1 else None
                if logvar is not None and mode != "mean":
                    std = torch.exp(0.5 * logvar)
                    v = (mean - beta * std) if mode == "lcb" else (mean + beta * std)
                else:
                    v = mean
            else:
                v = out.view(-1)
            pred[s:e] = v.cpu().numpy()
            if it is not None: it.update(e - s)
    finally:
        if it is not None: it.close()
    return np.argsort(-pred).astype(np.int64)

def _ndcg(idx, y, k):
    k = min(k, len(idx))
    g = y[idx[:k]] - y.min() + 1e-6
    d = 1.0 / np.log2(np.arange(2, k + 2))
    ideal = np.sort(y)[::-1][:k] - y.min() + 1e-6
    return float((g * d).sum() / (ideal * d).sum())


def _kendall_sampled(a, b, n=500_000, rng=None):
    N = len(a)
    rng = rng or np.random.default_rng(0)
    i = rng.integers(0, N, n); j = rng.integers(0, N, n)
    k = i != j
    return float((np.sign(a[i[k]] - a[j[k]]) * np.sign(b[i[k]] - b[j[k]])).mean())


def eval_vs_target(idx, y):
    r = np.empty(len(idx), dtype=np.int64); r[idx] = np.arange(len(idx)) + 1
    t = rankdata(-y)
    return dict(kendall=float(_kendall_sampled(r, t)),
                spearman=float(spearmanr(r, t).statistic),
                ndcg10=_ndcg(idx, y, 10),
                ndcg100=_ndcg(idx, y, 100),
                ndcg1000=_ndcg(idx, y, 1000))


def cross_agreement(a, b):
    N = len(a)
    ra = np.empty(N, np.int32); ra[a] = np.arange(N)
    rb = np.empty(N, np.int32); rb[b] = np.arange(N)
    return dict(kendall=_kendall_sampled(ra, rb),
                spearman=float(spearmanr(ra, rb).statistic),
                top100=len(set(a[:100]) & set(b[:100])) / 100)


METHODS = [
    ("ps_qmid",     "pairwise", lambda Z, m, bs, v, d: quicksort(Z, m, "mid",    bs, v, d)),
    ("ps_qrand",    "pairwise", lambda Z, m, bs, v, d: quicksort(Z, m, "random", bs, v, d)),
    ("ds_fast",     "ds",       lambda Z, m, bs, v, d: score_sort(Z, m, bs, "ds", v, d)),
    ("reg_mlp",     "mlp",      lambda Z, m, bs, v, d: reg_sort(Z, m, bs, "mean", 1.0, v, d)),
    ("reg_nll",     "nll",      lambda Z, m, bs, v, d: reg_sort(Z, m, bs, "mean", 1.0, v, d)),
    ("reg_nll_lcb", "nll",      lambda Z, m, bs, v, d: reg_sort(Z, m, bs, "lcb",  1.0, v, d)),
    ("mtm_fa",      "mt",       lambda Z, m, bs, v, d: score_sort(Z, m, bs, "mt", v, d)),
]


def run_all_and_save(h5_path, models, device, y_key="Fa_mean",
                     bs=262144, overwrite=True, verbose=True):
    idx_all = {}
    with h5py.File(h5_path, "r+") as f:
        for k in [k for k in f if k.startswith("mu_")]:
            vae = k[3:]
            if vae not in models: continue
            # CPU pinned — обязательно для 25M
            Z = torch.from_numpy(np.ascontiguousarray(f[k][:], np.float32)).pin_memory()
            grp = f.require_group(f"sorted_indices/{vae}")
            idx_all[vae] = {}
            for name, kind, fn in METHODS:
                if kind not in models[vae]: continue
                if name in grp and not overwrite:
                    idx = grp[name][:]
                else:
                    if verbose: print(f"[{vae}] {name}")
                    idx = fn(Z, models[vae][kind], bs, verbose, device)
                    if name in grp: del grp[name]
                    grp.create_dataset(name, data=idx.astype(np.int32), compression="lzf")
                idx_all[vae][name] = idx
            del Z; gc.collect()
            if device.type == "cuda": torch.cuda.empty_cache()

        if y_key and y_key in f.attrs:
            y = f.attrs[y_key][:]
            rows = [dict(VAE=v, Method=m, **eval_vs_target(idx_all[v][m], y))
                    for v in idx_all for m in idx_all[v]
                    if len(y) == len(idx_all[v][m])]
            if rows:
                print("\n=== vs target ===")
                print(pd.DataFrame(rows).to_string(index=False, float_format="%.4f"))

    print("\n=== cross-method (Kendall, 500k) ===")
    for v, per in idx_all.items():
        n = list(per)
        M = pd.DataFrame(np.eye(len(n)), index=n, columns=n)
        for i, a in enumerate(n):
            for j, b in enumerate(n):
                if i < j:
                    M.iloc[i, j] = M.iloc[j, i] = cross_agreement(per[a], per[b])["kendall"]
        print(f"\n[{v}]")
        print(M.to_string(float_format="%.3f"))

    print("\n=== cross-VAE (per method) ===")
    for name, _, _ in METHODS:
        vs = [v for v in idx_all if name in idx_all[v]]
        for i, a in enumerate(vs):
            for b in vs[i + 1:]:
                m = cross_agreement(idx_all[a][name], idx_all[b][name])
                print(f"[{name}] {a} vs {b}: τ={m['kendall']:+.3f} "
                      f"ρ={m['spearman']:+.3f} top100={m['top100']:.3f}")
    return idx_all
```

```python
import numpy as np, pandas as pd, h5py, torch, gc
from scipy.stats import spearmanr, norm
from sklearn.cluster import KMeans
from sklearn.covariance import EmpiricalCovariance

def _ranks(a):
    r = np.empty(len(a), np.int32); r[a] = np.arange(len(a), dtype=np.int32); return r

def _kendall(a, b, n=500_000, rng=None):
    rng = rng or np.random.default_rng(0); N = len(a)
    i, j = rng.integers(0, N, n), rng.integers(0, N, n); k = i != j
    return float((np.sign(a[i[k]] - a[j[k]]) * np.sign(b[i[k]] - b[j[k]])).mean())

def _rbo(a, b, p=0.9, k=10000):
    k = min(k, len(a), len(b))
    sa, sb, ov, ws, w = set(), set(), 0.0, 0.0, 1.0
    for i in range(k):
        sa.add(a[i]); sb.add(b[i]); ov += len(sa & sb) / (i + 1); ws += w; w *= p
    return (1 - p) * ov / ws if ws > 0 else 0.0

def _enrich(a, b, N, ks=(10, 100, 1000, 10000)):
    return {f"top{k}": (len(set(a[:k]) & set(b[:k])) / k) / (k / N)
            for k in ks if k <= len(a) and k <= len(b)}

def _pairwise(scores, N, n=500_000):
    names = list(scores); idx = {n_: np.argsort(-scores[n_]) for n_ in names}
    rng = np.random.default_rng(0); rows = []
    for i, a in enumerate(names):
        for b in names[i+1:]:
            ra, rb = _ranks(idx[a]), _ranks(idx[b])
            r = dict(a=a, b=b, kendall=_kendall(ra, rb, n, rng),
                     spearman=float(spearmanr(ra, rb).statistic),
                     rbo90=_rbo(idx[a], idx[b], p=0.90, k=10000),
                     rbo95=_rbo(idx[a], idx[b], p=0.95, k=10000))
            r.update(_enrich(idx[a], idx[b], N)); rows.append(r)
    return pd.DataFrame(rows)

@torch.inference_mode()
def _infer(models_v, Z, device, bs=262144):
    N = len(Z); out = {n: np.empty(N, np.float32) for n in models_v if n != 'pairwise'}
    for name, m in models_v.items():
        if name == 'pairwise': continue
        for s in range(0, N, bs):
            e = min(s + bs, N); x = Z[s:e].to(device, non_blocking=True)
            if name == 'nll':
                mu, lv = m(x); out[name][s:e] = mu.view(-1).cpu().numpy()
                out.setdefault('nll_sigma', np.empty(N, np.float32))[s:e] = torch.exp(0.5 * lv).view(-1).cpu().numpy()
            elif name == 'mt':
                y = m(x)
                for j, t in enumerate(['fa','Q','G','TanD','k']):
                    out.setdefault(f'mt_{t}', np.empty(N, np.float32))[s:e] = y[:, j].cpu().numpy()
                out[name][s:e] = y[:, 0].cpu().numpy()
            elif name == 'ds': out[name][s:e] = m.score(x).cpu().numpy()
            else: out[name][s:e] = m(x).view(-1).cpu().numpy()
    if 'nll' in out:
        mu, sig = out['nll'], out['nll_sigma']
        out['ucb'], out['lcb'] = mu + sig, mu - sig
        z = (mu - mu.max()) / (sig + 1e-8)
        out['ei'] = sig * (z * norm.cdf(z) + norm.pdf(z))
        out['ts'] = mu + sig * np.random.default_rng(0).standard_normal(N).astype(np.float32)
    return out

def _ood(Z_tr, Z_te):
    return EmpiricalCovariance().fit(Z_tr).mahalanobis(Z_te).astype(np.float32)

def _select(scores, Z, pool, N=10):
    df = pd.DataFrame({'idx': pool})
    for k, v in scores.items():
        if k.endswith('_sigma'): continue
        df[k] = v[pool]
    df['n_agree'] = sum((scores[k] >= np.quantile(scores[k], 0.99))[pool]
                        for k in scores if not k.endswith('_sigma'))
    df['cluster'] = KMeans(min(N * 2, len(pool)), n_init=10, random_state=0).fit_predict(Z[pool])
    df = df.sort_values('ucb', ascending=False)
    keep, seen = [], set()
    for _, r in df.iterrows():
        if r['cluster'] in seen: continue
        keep.append(r); seen.add(r['cluster'])
        if len(keep) >= N: break
    return pd.DataFrame(keep), df

def run(H5, MD, Ztr, device, TOP_K=25, N=10):
    res = {}
    CKPT = [('mlp', MLP, 'MLP'), ('nll', MLPNLL, 'MLPNLL'),
            ('ds', DSMLP, 'DSMLP'), ('mt', MTMLP, 'MTM')]
    with h5py.File(H5, "r") as f:
        for vae in f["sorted_indices"].keys():
            Zc = torch.from_numpy(f[f"mu_{vae}"][:].astype(np.float32)).pin_memory()
            M = {n: load_model(f"{MD}/{cls_name}_{vae}_best.pt", cls, device=device)
                 for n, cls, cls_name in CKPT}
            sc = _infer(M, Zc, device)
            if vae in Ztr: sc['ood'] = _ood(Ztr[vae], f[f"mu_{vae}"][:])
            pw = _pairwise(sc, len(Zc))
            pool = sorted(set().union(*[np.argsort(-sc[k])[:TOP_K]
                                        for k in sc if not k.endswith('_sigma')]))
            final, df = _select(sc, f[f"mu_{vae}"][:], pool, N)
            res[vae] = dict(pairwise=pw, pool=df, final=final)
            print(f"\n[{vae}] {len(final)}/{len(pool)}")
            print(final[['idx','ucb','n_agree','cluster']].to_string(index=False))
            del M, Zc, sc; gc.collect()
            if device.type == 'cuda': torch.cuda.empty_cache()
    return res

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
VAES = ["CLS", "SET", "DS"]
Ztr = {v: np.stack(df_train[f'latent_{v}'].values).astype(np.float32) for v in VAES}
res = run("./data/DL/latent_db.h5", "./data/DL/models", Ztr, device, TOP_K=25, N=10)
```

### Model parametrization

```python
import torch, torch.nn as nn, numpy as np, pandas as pd, random, os
from scipy.stats import kendalltau, spearmanr

SEED = 42
def set_seed(s=SEED):
    random.seed(s); np.random.seed(s); os.environ['PYTHONHASHSEED'] = str(s)
    torch.manual_seed(s); torch.cuda.manual_seed_all(s)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
set_seed()


class Fusion(nn.Module):
    def __init__(s, D=None): super().__init__()
    def forward(s, a, b): raise NotImplementedError

class F_diff(Fusion):     forward = lambda s,a,b: a-b
class F_prod(Fusion):     forward = lambda s,a,b: torch.cat([a-b, a*b], -1)
class F_cat(Fusion):      forward = lambda s,a,b: torch.cat([a, b], -1)
class F_cos(Fusion):      forward = lambda s,a,b: torch.cat([a-b, nn.functional.cosine_similarity(a,b,dim=-1,eps=1e-8).unsqueeze(-1)], -1)
class F_absdiff(Fusion):  forward = lambda s,a,b: torch.cat([a-b, torch.abs(a-b)], -1)

class F_smooth(Fusion):
    def forward(s,a,b):
        d=a-b; return torch.cat([d,(a*b)*torch.tanh(10*d)],-1)
class F_relu(Fusion):
    def forward(s,a,b):
        wA,wB=torch.relu(a-b),torch.relu(b-a); return torch.cat([a*wA,b*wB,wA-wB],-1)
class F_multi(Fusion):
    def forward(s,a,b):
        d=torch.clamp(a-b,-3,3); return torch.cat([d,d**3/6,torch.tanh(d),torch.sign(d)*d**2/2],-1)
class F_gate(Fusion):
    def __init__(s,D): super().__init__(D); s.g=nn.Linear(2*D,2*D)
    def forward(s,a,b):
        d=a-b; anti=torch.cat([d,(a*b)*torch.tanh(10*d)],-1)
        return anti*torch.sigmoid(s.g(torch.cat([torch.abs(d),a+b],-1)))
class F_bilinear(Fusion):
    def __init__(s,D): super().__init__(D); s.p=nn.Linear(D*D,2*D)
    def forward(s,a,b):
        o=(a-b).unsqueeze(-1)*(a+b).unsqueeze(-2); return s.p(o.flatten(1))
class F_direct(Fusion):
    def __init__(s,D): super().__init__(D); s.proj=nn.Linear(D,1)
    def forward(s,a,b): return s.proj(a)-s.proj(b)
class F_gated(Fusion):
    def __init__(s,D): super().__init__(D); s.g=nn.Linear(4*D,4*D)
    def forward(s,a,b):
        z=torch.cat([a,b,a-b,a*b],-1); return z*torch.sigmoid(s.g(z))

FUSIONS = {'P1_diff':F_diff,'P2_prod':F_prod,'P3_cat':F_cat,'P4_cos':F_cos,'P5_absdiff':F_absdiff,
           'V1_smooth':F_smooth,'V2_relu':F_relu,'V3_multi':F_multi,'V4_gate':F_gate,
           'S1_direct':F_direct,'S2_bilinear':F_bilinear,'S3_gated':F_gated}


class SymPSMLP(nn.Module):
    def __init__(s, D=64, fusion='V1_smooth', hidden=128):
        super().__init__()
        s.direct = (fusion=='S1_direct'); s.fuse = FUSIONS[fusion](D)
        if not s.direct:
            s.tower = nn.Sequential(
                nn.Linear(D,hidden), nn.LayerNorm(hidden), nn.Mish(), nn.Dropout(0.2),
                nn.Linear(hidden,D), nn.LayerNorm(D), nn.Mish(), nn.Dropout(0.2))
            s.head = nn.Sequential(nn.LazyLinear(hidden), nn.Mish(), nn.Dropout(0.2),
                                   nn.Linear(hidden,32), nn.Mish(), nn.Linear(32,1))
    def _phi(s,hA,hB): return s.head(s.fuse(hA,hB)).squeeze(-1)
    def forward(s,xA,xB):
        if s.direct: return s.fuse(xA,xB).squeeze(-1)
        hA,hB = s.tower(xA), s.tower(xB)
        return 0.5*(s._phi(hA,hB) - s._phi(hB,hA))


class Baseline_MLP(nn.Module):
    def __init__(s,D=64,hidden=256,dropout=0.15):
        super().__init__()
        s.net = nn.Sequential(
            nn.Linear(D,hidden), nn.LayerNorm(hidden), nn.Mish(), nn.Dropout(dropout),
            nn.Linear(hidden,hidden), nn.LayerNorm(hidden), nn.Mish(),
            nn.Linear(hidden,hidden//4), nn.LayerNorm(hidden//4), nn.Mish(),
            nn.Linear(hidden//4,1))
    def forward(s,a,b): return (s.net(a)-s.net(b)).squeeze(-1)


class Baseline_PSMLP_orig(nn.Module):
    def __init__(s,D=64,hidden=128):
        super().__init__()
        s.tower = nn.Sequential(
            nn.Linear(D,hidden), nn.LayerNorm(hidden), nn.Mish(), nn.Dropout(0.2),
            nn.Linear(hidden,D), nn.LayerNorm(D), nn.Mish(), nn.Dropout(0.2))
        s.head = nn.Sequential(nn.Linear(2*D,hidden), nn.Mish(), nn.Dropout(0.2),
                                nn.Linear(hidden,32), nn.Mish(), nn.Linear(32,1))
    def forward(s,a,b):
        hA,hB = s.tower(a), s.tower(b); d = hA - hB
        return s.head(torch.cat([d,(hA*hB)*torch.sign(d)],-1)).squeeze(-1)


def cycle_loss(model, Z, n=1024):
    i = torch.randint(0, len(Z), (3,n), device=Z.device)
    A,B,C = Z[i[0]], Z[i[1]], Z[i[2]]
    ab,bc,ca = model(A,B), model(B,C), model(C,A)
    return (torch.relu(ab)*torch.relu(bc)*torch.relu(ca)).mean()


def trans_loss(model, Z, y, n=1024, m=0.1):
    i = torch.randint(0, len(Z), (3,n), device=Z.device)
    mask = (y[i[0]]>y[i[1]]) & (y[i[1]]>y[i[2]])
    if mask.sum() < 16: return torch.tensor(0., device=Z.device)
    A,B,C = Z[i[0]][mask], Z[i[1]][mask], Z[i[2]][mask]
    return torch.relu(m - model(A,C) + model(A,B)).mean()


def pw_step(model, batch, device, Z, y, β=0.3, γ=0.1):
    a,b,t = (x.to(device) for x in batch)
    logit = model(a,b)
    l = nn.functional.binary_cross_entropy_with_logits(logit, t)
    l = l + β*cycle_loss(model,Z) + γ*trans_loss(model,Z,y)
    return l, t, torch.sigmoid(logit)


@torch.inference_mode()
def copeland(Z, model, bs=256):
    model.eval()
    N = len(Z); S = torch.zeros(N, device=Z.device)
    for i in range(0, N, bs):
        e = min(i+bs, N); A = Z[i:e]
        for j in range(0, N, bs):
            f = min(j+bs, N); B = Z[j:f]
            Ar = A.unsqueeze(1).expand(-1,B.size(0),-1).reshape(-1,A.size(1))
            Br = B.unsqueeze(0).expand(A.size(0),-1,-1).reshape(-1,B.size(1))
            out = model(Ar,Br).view(A.size(0), B.size(0))
            S[i:e] += (out>0).float().sum(1); S[j:f] -= (out>0).float().sum(0)
    return torch.argsort(-S).cpu().numpy()


@torch.inference_mode()
def diagnostics(model, Z, n=50_000):
    model.eval()
    i = torch.randint(0, len(Z), (n,), device=Z.device)
    j = torch.randint(0, len(Z), (n,), device=Z.device)
    ab, ba = model(Z[i],Z[j]), model(Z[j],Z[i])
    k = torch.randint(0, len(Z), (3,n), device=Z.device)
    A,B,C = Z[k[0]], Z[k[1]], Z[k[2]]
    ab2,bc,ca = model(A,B), model(B,C), model(C,A)
    base = ((ab2>0)&(bc>0)).float()
    return dict(
        antisym = (torch.sign(ab)==-torch.sign(ba)).float().mean().item(),
        bt      = ((ab>0)&(ba>0)).float().mean().item(),
        trans   = base.mean().item(),
        cyc     = (((ab2>0)&(bc>0)&(ca>0)).float().sum() / base.sum().clamp(min=1)).item())


def rank_metrics(idx, y):
    r = np.empty(len(idx), np.int64); r[idx] = np.arange(len(idx))
    return dict(sp=spearmanr(-r,y).statistic, kt=kendalltau(-r,y).statistic)


MODELS = list(FUSIONS) + ['BASE_MLP','BASE_PSMLP_orig','BASE_SymPSMLP']
AGG = ['qsort','copeland']; rows = []
```

```python
import torch, torch.nn as nn, numpy as np, pandas as pd, random, os
from scipy.stats import kendalltau, spearmanr

SEED = 42
def set_seed(s=SEED):
    random.seed(s); np.random.seed(s); os.environ['PYTHONHASHSEED'] = str(s)
    torch.manual_seed(s); torch.cuda.manual_seed_all(s)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
set_seed()


class Fusion(nn.Module):
    def __init__(s, D=None): super().__init__()
    def forward(s, a, b): raise NotImplementedError

class F_diff(Fusion):     forward = lambda s,a,b: a-b
class F_prod(Fusion):     forward = lambda s,a,b: torch.cat([a-b, a*b], -1)
class F_cat(Fusion):      forward = lambda s,a,b: torch.cat([a, b], -1)
class F_cos(Fusion):      forward = lambda s,a,b: torch.cat([a-b, nn.functional.cosine_similarity(a,b,dim=-1,eps=1e-8).unsqueeze(-1)], -1)
class F_absdiff(Fusion):  forward = lambda s,a,b: torch.cat([a-b, torch.abs(a-b)], -1)

class F_smooth(Fusion):
    def forward(s,a,b):
        d=a-b; return torch.cat([d,(a*b)*torch.tanh(10*d)],-1)
class F_relu(Fusion):
    def forward(s,a,b):
        wA,wB=torch.relu(a-b),torch.relu(b-a); return torch.cat([a*wA,b*wB,wA-wB],-1)
class F_multi(Fusion):
    def forward(s,a,b):
        d=torch.clamp(a-b,-3,3); return torch.cat([d,d**3/6,torch.tanh(d),torch.sign(d)*d**2/2],-1)
class F_gate(Fusion):
    def __init__(s,D): super().__init__(D); s.g=nn.Linear(2*D,2*D)
    def forward(s,a,b):
        d=a-b; anti=torch.cat([d,(a*b)*torch.tanh(10*d)],-1)
        return anti*torch.sigmoid(s.g(torch.cat([torch.abs(d),a+b],-1)))
class F_bilinear(Fusion):
    def __init__(s,D): super().__init__(D); s.p=nn.Linear(D*D,2*D)
    def forward(s,a,b):
        o=(a-b).unsqueeze(-1)*(a+b).unsqueeze(-2); return s.p(o.flatten(1))
class F_direct(Fusion):
    def __init__(s,D): super().__init__(D); s.proj=nn.Linear(D,1)
    def forward(s,a,b): return s.proj(a)-s.proj(b)
class F_gated(Fusion):
    def __init__(s,D): super().__init__(D); s.g=nn.Linear(4*D,4*D)
    def forward(s,a,b):
        z=torch.cat([a,b,a-b,a*b],-1); return z*torch.sigmoid(s.g(z))

FUSIONS = {'P1_diff':F_diff,'P2_prod':F_prod,'P3_cat':F_cat,'P4_cos':F_cos,'P5_absdiff':F_absdiff,
           'V1_smooth':F_smooth,'V2_relu':F_relu,'V3_multi':F_multi,'V4_gate':F_gate,
           'S1_direct':F_direct,'S2_bilinear':F_bilinear,'S3_gated':F_gated}


class SymPSMLP(nn.Module):
    def __init__(s, D=64, fusion='V1_smooth', hidden=128):
        super().__init__()
        s.direct = (fusion=='S1_direct'); s.fuse = FUSIONS[fusion](D)
        if not s.direct:
            s.tower = nn.Sequential(
                nn.Linear(D,hidden), nn.LayerNorm(hidden), nn.Mish(), nn.Dropout(0.2),
                nn.Linear(hidden,D), nn.LayerNorm(D), nn.Mish(), nn.Dropout(0.2))
            s.head = nn.Sequential(nn.LazyLinear(hidden), nn.Mish(), nn.Dropout(0.2),
                                   nn.Linear(hidden,32), nn.Mish(), nn.Linear(32,1))
    def _phi(s,hA,hB): return s.head(s.fuse(hA,hB)).squeeze(-1)
    def forward(s,xA,xB):
        if s.direct: return s.fuse(xA,xB).squeeze(-1)
        hA,hB = s.tower(xA), s.tower(xB)
        return 0.5*(s._phi(hA,hB) - s._phi(hB,hA))


class Baseline_MLP(nn.Module):
    def __init__(s,D=64,hidden=256,dropout=0.15):
        super().__init__()
        s.net = nn.Sequential(
            nn.Linear(D,hidden), nn.LayerNorm(hidden), nn.Mish(), nn.Dropout(dropout),
            nn.Linear(hidden,hidden), nn.LayerNorm(hidden), nn.Mish(),
            nn.Linear(hidden,hidden//4), nn.LayerNorm(hidden//4), nn.Mish(),
            nn.Linear(hidden//4,1))
    def forward(s,a,b): return (s.net(a)-s.net(b)).squeeze(-1)


class Baseline_PSMLP_orig(nn.Module):
    def __init__(s,D=64,hidden=128):
        super().__init__()
        s.tower = nn.Sequential(
            nn.Linear(D,hidden), nn.LayerNorm(hidden), nn.Mish(), nn.Dropout(0.2),
            nn.Linear(hidden,D), nn.LayerNorm(D), nn.Mish(), nn.Dropout(0.2))
        s.head = nn.Sequential(nn.Linear(2*D,hidden), nn.Mish(), nn.Dropout(0.2),
                                nn.Linear(hidden,32), nn.Mish(), nn.Linear(32,1))
    def forward(s,a,b):
        hA,hB = s.tower(a), s.tower(b); d = hA - hB
        return s.head(torch.cat([d,(hA*hB)*torch.sign(d)],-1)).squeeze(-1)


def cycle_loss(model, Z, n=1024):
    i = torch.randint(0, len(Z), (3,n), device=Z.device)
    A,B,C = Z[i[0]], Z[i[1]], Z[i[2]]
    ab,bc,ca = model(A,B), model(B,C), model(C,A)
    return (torch.relu(ab)*torch.relu(bc)*torch.relu(ca)).mean()


def trans_loss(model, Z, y, n=1024, m=0.1):
    i = torch.randint(0, len(Z), (3,n), device=Z.device)
    mask = (y[i[0]]>y[i[1]]) & (y[i[1]]>y[i[2]])
    if mask.sum() < 16: return torch.tensor(0., device=Z.device)
    A,B,C = Z[i[0]][mask], Z[i[1]][mask], Z[i[2]][mask]
    return torch.relu(m - model(A,C) + model(A,B)).mean()


def pw_step(model, batch, device, Z, y, β=0.3, γ=0.1):
    a,b,t = (x.to(device) for x in batch)
    logit = model(a,b)
    l = nn.functional.binary_cross_entropy_with_logits(logit, t)
    l = l + β*cycle_loss(model,Z) + γ*trans_loss(model,Z,y)
    return l, t, torch.sigmoid(logit)


@torch.inference_mode()
def copeland(Z, model, bs=256):
    model.eval()
    N = len(Z); S = torch.zeros(N, device=Z.device)
    for i in range(0, N, bs):
        e = min(i+bs, N); A = Z[i:e]
        for j in range(0, N, bs):
            f = min(j+bs, N); B = Z[j:f]
            Ar = A.unsqueeze(1).expand(-1,B.size(0),-1).reshape(-1,A.size(1))
            Br = B.unsqueeze(0).expand(A.size(0),-1,-1).reshape(-1,B.size(1))
            out = model(Ar,Br).view(A.size(0), B.size(0))
            S[i:e] += (out>0).float().sum(1); S[j:f] -= (out>0).float().sum(0)
    return torch.argsort(-S).cpu().numpy()


@torch.inference_mode()
def diagnostics(model, Z, n=50_000):
    model.eval()
    i = torch.randint(0, len(Z), (n,), device=Z.device)
    j = torch.randint(0, len(Z), (n,), device=Z.device)
    ab, ba = model(Z[i],Z[j]), model(Z[j],Z[i])
    k = torch.randint(0, len(Z), (3,n), device=Z.device)
    A,B,C = Z[k[0]], Z[k[1]], Z[k[2]]
    ab2,bc,ca = model(A,B), model(B,C), model(C,A)
    base = ((ab2>0)&(bc>0)).float()
    return dict(
        antisym = (torch.sign(ab)==-torch.sign(ba)).float().mean().item(),
        bt      = ((ab>0)&(ba>0)).float().mean().item(),
        trans   = base.mean().item(),
        cyc     = (((ab2>0)&(bc>0)&(ca>0)).float().sum() / base.sum().clamp(min=1)).item())


def rank_metrics(idx, y):
    r = np.empty(len(idx), np.int64); r[idx] = np.arange(len(idx))
    return dict(sp=spearmanr(-r,y).statistic, kt=kendalltau(-r,y).statistic)


MODELS = list(FUSIONS) + ['BASE_MLP','BASE_PSMLP_orig','BASE_SymPSMLP']
AGG = ['qsort','copeland']; rows = []

for vae in ['CLS','SET','DS']:
    lc = f'latent_{vae}'
    z_tr = torch.tensor(np.stack(df_train[lc].values), dtype=torch.float32, device=device)
    y_tr = torch.tensor(df_train['Fa_mean'].values, dtype=torch.float32, device=device)
    z_te = torch.tensor(np.stack(df_test[lc].values),  dtype=torch.float32, device=device)
    y_te = df_test['Fa_mean'].values
    D = z_tr.shape[1]
    tr_l, va_l = _pw_loaders(df_train, df_val, lc)

    for name in MODELS:
        set_seed()
        if name == 'BASE_MLP':           m = Baseline_MLP(D)
        elif name == 'BASE_PSMLP_orig':  m = Baseline_PSMLP_orig(D)
        elif name == 'BASE_SymPSMLP':    m = SymPSMLP(D, 'V1_smooth')
        else:                             m = SymPSMLP(D, name)
        m = m.to(device)

        with torch.no_grad():
            dummy = torch.zeros(2, D, device=device)
            m(dummy, dummy)

        β, γ = (0.0, 0.0) if name.startswith('BASE_') or name=='S1_direct' else (0.3, 0.1)
        def step(mm, bb, dd, β=β, γ=γ):
            return pw_step(mm, bb, dd, Z=z_tr, y=y_tr, β=β, γ=γ)
        globals()['S']['pairwise'] = step

        train(m, tr_l, va_l,
              Cfg(name=f"{name}_{vae}", task="pairwise", sched="plateau", seed=SEED, **COMMON), device)
        ck = f"./data/DL/models/{name}_{vae}_best.pt"
        m.load_state_dict(torch.load(ck, map_location=device)["model"])
        m.eval()

        diag = diagnostics(m, z_tr)
        for agg in AGG:
            try:
                idx = quicksort(z_te, m) if agg=='qsort' else copeland(z_te, m)
                rm = rank_metrics(idx, y_te)
            except Exception as e:
                print(f"[{name}/{vae}/{agg}] FAILED: {type(e).__name__}: {e}")
                rm = dict(sp=np.nan, kt=np.nan)
            rows.append(dict(VAE=vae, Model=name, Agg=agg, **diag, **rm))

        del m; torch.cuda.empty_cache()

df_res = pd.DataFrame(rows)
pd.set_option('display.float_format', '{:.4f}'.format)
print(df_res.pivot_table(index=['VAE','Model'], columns='Agg',
                          values=['antisym','bt','trans','cyc','sp','kt']).to_string())
```

```python
def recover(VAES, MODELS, AGG, ckpt_dir='./data/DL/models'):
    rows = []
    for vae in VAES:
        lc = f'latent_{vae}'
        z_tr = torch.tensor(np.stack(df_train[lc].values), dtype=torch.float32, device=device)
        z_te = torch.tensor(np.stack(df_test[lc].values),  dtype=torch.float32, device=device)
        y_te = df_test['Fa_mean'].values
        D = z_tr.shape[1]

        for name in MODELS:
            ck = f"{ckpt_dir}/{name}_{vae}_best.pt"
            if not Path(ck).exists():
                print(f"[{name}/{vae}] no ckpt, skip"); continue
            try:
                set_seed()
                if name == 'BASE_MLP':          m = Baseline_MLP(D)
                elif name == 'BASE_PSMLP_orig': m = Baseline_PSMLP_orig(D)
                elif name == 'BASE_SymPSMLP':   m = SymPSMLP(D, 'V1_smooth')
                else:                            m = SymPSMLP(D, name)
                m = m.to(device)
                with torch.no_grad():
                    dummy = torch.zeros(2, D, device=device)
                    m(dummy, dummy)
                m.load_state_dict(torch.load(ck, map_location=device)["model"]); m.eval()

                diag = diagnostics(m, z_tr)
                for agg in AGG:
                    try:
                        idx = quicksort(z_te, m) if agg=='qsort' else copeland(z_te, m)
                        rm = rank_metrics(idx, y_te)
                    except Exception as e:
                        print(f"[{name}/{vae}/{agg}] FAILED: {e}"); rm = dict(sp=np.nan, kt=np.nan)
                    rows.append(dict(VAE=vae, Model=name, Agg=agg, **diag, **rm))
                del m; torch.cuda.empty_cache()
                print(f"[{name}/{vae}] ok")
            except Exception as e:
                print(f"[{name}/{vae}] LOAD FAILED: {e}")

    df_res = pd.DataFrame(rows)
    if df_res.empty: print("No results."); return df_res
    pd.set_option('display.float_format','{:.4f}'.format)
    print(df_res.pivot_table(index=['VAE','Model'], columns='Agg',
                              values=['antisym','bt','trans','cyc','sp','kt']).to_string())
    return df_res

MODELS = list(FUSIONS) + ['BASE_MLP','BASE_PSMLP_orig','BASE_SymPSMLP']
df_res = recover(['CLS','SET','DS'], MODELS, ['qsort','copeland'])
```

```python
from scipy.stats import spearmanr
for vae in ['CLS','SET','DS']:
    lc = f'latent_{vae}'
    Z_tr = np.stack(df_train[lc].values)
    Z_te = np.stack(df_test[lc].values)
    y_tr, y_te = df_train['Fa_mean'].values, df_test['Fa_mean'].values
    r_tr = spearmanr(Z_tr[:, 0], y_tr).statistic
    r_te = spearmanr(Z_te[:, 0], y_te).statistic
    print(f"{vae}: PC1 corr train={r_tr:+.3f}  test={r_te:+.3f}")

from scipy.stats import spearmanr
for vae in ['CLS','SET','DS']:
    lc = f'latent_{vae}'
    Z_tr = np.stack(df_train[lc].values)
    y_tr = df_train['Fa_mean'].values
    corrs = np.array([spearmanr(Z_tr[:,k], y_tr).statistic for k in range(Z_tr.shape[1])])
    top = np.argsort(-np.abs(corrs))[:5]
    print(f"{vae}: top-5 |corr| = {[(int(k), round(corrs[k],3)) for k in top]}")

from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score
for vae in ['CLS','SET','DS']:
    lc = f'latent_{vae}'
    Z_tr = np.stack(df_train[lc].values); y_tr = df_train['Fa_mean'].values
    Z_te = np.stack(df_test[lc].values);  y_te = df_test['Fa_mean'].values
    r = Ridge(alpha=1.0).fit(Z_tr, y_tr)
    print(f"{vae}: R² train={r.score(Z_tr,y_tr):.3f}  test={r.score(Z_te,y_te):.3f}")

import numpy as np, torch
from scipy.stats import spearmanr

def auc_by_bins(model, Z, y, n_pairs=50_000, seed=0):
    """AUC по бинам |Δy|."""
    rng = np.random.default_rng(seed)
    N = len(Z)
    i = rng.integers(0, N, n_pairs)
    j = rng.integers(0, N, n_pairs)
    mask = i != j
    i, j = i[mask], j[mask]

    A = Z[torch.from_numpy(i).to(Z.device)]
    B = Z[torch.from_numpy(j).to(Z.device)]
    with torch.inference_mode():
        pred = (model(A, B) - model(B, A)).cpu().numpy()    # антисимметричная версия
    dy   = y[i] - y[j]
    d    = np.abs(dy)
    sign = np.sign(dy)

    bins = np.quantile(d, [0, 0.2, 0.4, 0.6, 0.8, 0.9, 0.95, 1.0])
    print(f"  n_pairs={len(i)}   |Δy| median={np.median(d):.3f}")
    for lo, hi in zip(bins[:-1], bins[1:]):
        m = (d >= lo) & (d < hi)
        if m.sum() < 50: continue
        acc = (np.sign(pred[m]) == sign[m]).mean()
        print(f"  |Δy| in [{lo:6.2f}, {hi:6.2f}):  n={m.sum():>6}  acc={acc:.3f}")


# применение ко всем VAE и всем моделям, что есть
for vae in ['CLS','SET','DS']:
    lc = f'latent_{vae}'
    Z = torch.tensor(np.stack(df_test[lc].values), dtype=torch.float32, device=device)
    y = df_test['Fa_mean'].values

    print(f"\n═══ {vae}  (n={len(y)}) ═══")
    for name in ['P4_cos','P5_absdiff','V1_smooth','BASE_MLP']:
        ck = f"./data/DL/models/{name}_{vae}_best.pt"
        from pathlib import Path
        if not Path(ck).exists():
            continue
        D = Z.shape[1]
        if   name == 'BASE_MLP':   m = Baseline_MLP(D)
        elif name == 'BASE_PSMLP_orig': m = Baseline_PSMLP_orig(D)
        else:                       m = SymPSMLP(D, name)
        m = m.to(device)
        with torch.no_grad(): m(torch.zeros(2,D,device=device), torch.zeros(2,D,device=device))
        m.load_state_dict(torch.load(ck, map_location=device)["model"]); m.eval()
        print(f"\n[{name}]")
        auc_by_bins(m, Z, y, n_pairs=30_000)
        del m; torch.cuda.empty_cache()

for vae in ['CLS','SET','DS']:
    Z = np.stack(df_unique[f'latent_{vae}'].values)
    print(f"{vae}: std={Z.std(0).mean():.4f}  range=[{Z.min():.4f}, {Z.max():.4f}]")

from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
Z = np.stack(df_unique['latent_SET'].values)
pca = PCA(2).fit_transform(Z)
import matplotlib.pyplot as plt
plt.scatter(pca[:,0], pca[:,1], c=df_unique['Fa_mean'], cmap='viridis')
plt.colorbar(); plt.title('SET latent PCA, colored by Fa_mean'); plt.show()
```

```python
def decode2fract(z, vae, mono_feat, seq_len=1000, temperature=1.0,
                               device='cuda', use_min_median=True, eps=1e-7):
    vae.eval()
    if z.dim() == 1:
        z = z.unsqueeze(0)
    z = z.to(device)

    with torch.no_grad():
        recon_ffp, recon_auto = vae.decode(z)
        recon_feat = torch.cat([recon_ffp, recon_auto], dim=-1).squeeze(0)

    mono_feat = mono_feat.to(device)
    mono_norm = mono_feat / (mono_feat.norm(dim=1, keepdim=True) + 1e-8)
    recon_norm = recon_feat / (recon_feat.norm(dim=1, keepdim=True) + 1e-8)

    diff = recon_norm.unsqueeze(1) - mono_norm.unsqueeze(0)
    dist_sq = (diff ** 2).sum(dim=-1)

    with torch.no_grad():
        if use_min_median:
            min_dist_per_pos, _ = dist_sq.min(dim=1)
            scale = torch.median(min_dist_per_pos)
        else:
            scale = torch.median(dist_sq)
        adaptive_temp = temperature * (scale + eps)
        adaptive_temp = torch.clamp(adaptive_temp, min=1e-4)  
        logits = -dist_sq / adaptive_temp
        logits_max = logits.max(dim=1, keepdim=True)[0]
        probs = torch.softmax(logits - logits_max, dim=1)
    
        expected_counts = probs.sum(dim=0)
        fractions = (expected_counts / seq_len).cpu().numpy()

    return fractions, probs.cpu().numpy()
```

```python
class HighScoreOptimizer:
    def __init__(self, predictor, emp_mean, emp_latents, lr=0.03, lambda_reg=1e-4, sigma_mult=1.64, pred_type='mlp'):
        self.p = predictor.eval()
        self.m = torch.tensor(emp_mean, device=next(predictor.parameters()).device)
        self.l = torch.tensor(emp_latents, device=self.m.device)
        self.lr, self.lam, self.sm, self.ptype = lr, lambda_reg, sigma_mult, pred_type
        self.n = len(emp_latents)

    def _score(self, z):
        if self.ptype == 'mlp':
            return self.p(z).squeeze(), None
        else:
            mu, lv = self.p(z)
            sigma = torch.exp(0.5 * lv)
            return (mu + self.sm * sigma).squeeze(), lv

    def optimize(self, restarts=700, steps=1000, log_every=20, tb_writer=None, tag='opt', save_traj=True):
        best_z, best_score, best_hist, all_restarts = None, -float('inf'), None, []
        for _ in range(restarts):
            z = (self.l[np.random.randint(self.n)] + torch.randn(self.l.shape[1], device=self.m.device) * 0.05 * self.m.std()).clone().requires_grad_(True)
            opt = optim.Adam([z], lr=self.lr)
            traj_z, traj_y, traj_reg = [], [], []
            for step in range(steps):
                opt.zero_grad()
                y, lv = self._score(z.unsqueeze(0))
                if self.ptype == 'mlp':
                    lam = self.lam * max(0, 1 - step/steps)
                    reg = lam * (z - self.m).pow(2).mean()
                else:
                    reg = self.lam * ((z - self.m).pow(2).mean() + 0.5 * torch.exp(lv).mean())
                loss = -y + reg
                loss.backward(); opt.step()
                if step % log_every == 0 or step == steps-1:
                    traj_z.append(z.detach().cpu().numpy().copy())
                    traj_y.append(y.item())
                    traj_reg.append(reg.item())
            final_score = y.item()
            info = {'z_final': z.detach().cpu().numpy().copy(), 'y_final': final_score,
                    'trajectory': np.array(traj_z), 'y_history': np.array(traj_y), 'reg_history': np.array(traj_reg)}
            all_restarts.append(info)
            if final_score > best_score:
                best_z, best_score, best_hist = z.detach().clone(), final_score, info
        if tb_writer and best_hist:
            tb_writer.add_scalar(f'{tag}/best_score', best_score, 0)
            for i, (y, r) in enumerate(zip(best_hist['y_history'], best_hist['reg_history'])):
                tb_writer.add_scalar(f'{tag}/y_best', y, i*log_every)
                tb_writer.add_scalar(f'{tag}/reg_best', r, i*log_every)
            tb_writer.add_histogram(f'{tag}/latent_best', best_z.cpu(), 0)
        return best_z, best_score, best_hist['trajectory'], best_hist['y_history'], best_hist['reg_history'], all_restarts

def run_pipeline(model_type, ptype, vae_ckpt, pred_ckpt, db_path, valid_df, device, n_restarts=700, steps=1000, lr=0.03, lam=1e-4, sm=1.64, n_neighbors=50, tol=0.05, log_every=20, save_restarts=True, make_anim=False):
    writer = SummaryWriter(log_dir=f"runs/{model}_opt_{ptype.upper()}")
    vae = load_pretrained_vae(model, vae_ckpt, device)
    pred_class = MLP if ptype == 'mlp' else MLPNLL
    pred = pred_class().to(device)
    pred.load_state_dict(torch.load(pred_ckpt, map_location=device)); pred.eval()
    emp_mean, real_lats = get_real_latents(vae)
    del vae; torch.cuda.empty_cache()

    opt = HighScoreOptimizer(pred, emp_mean, real_lats, lr, lam, sm, ptype)
    best_z, best_score, best_traj, best_y_hist, best_reg_hist, all_restarts = opt.optimize(
        restarts=n_restarts, steps=steps, log_every=log_every, tb_writer=writer, tag=model, save_traj=save_restarts)
    print(f"Optimal Y ({ptype}): {best_score:.4f}")
    vae = load_pretrained_vae(model, vae_ckpt, device)
    fractions_dec, _ = decode2fract(
        best_z, vae, mono_feat_tensor,
        seq_len=1000, temperature=1.0,
        device=device, use_min_median=True
    )
    fractions_dec = fractions_dec / fractions_dec.sum()  
    z_from_dec = get_latent_for_composition(
        fractions_dec, vae, mon_indices_64, features_source_64, device
    )
    with torch.no_grad():
        if ptype == 'mlp':
            y_dec = pred(torch.tensor(z_from_dec, device=device).unsqueeze(0)).item()
        else:
            mu_dec, lv_dec = pred(torch.tensor(z_from_dec, device=device).unsqueeze(0))
            y_dec = (mu_dec + sm * torch.exp(0.5 * lv_dec)).item()
    
    deviation_dec = abs(y_dec - best_score) / (abs(best_score) + 1e-8)
    print(f"Decoder composition Y: {y_dec:.4f} (opt: {best_score:.4f}, dev: {deviation_dec*100:.2f}%)")
    del vae
    torch.cuda.empty_cache()    

    with h5py.File(db_path, 'r') as f:
        lats_db = f['latent'][:] if 'latent' in f else f['mu'][:]
        fracs = f['fractions'][:]
    z_opt_np = best_z.cpu().numpy().reshape(1, -1)
    idx = np.argsort(cdist(z_opt_np, lats_db).flatten())[:n_neighbors]
    with torch.no_grad():
        if ptype == 'mlp':
            y_n = pred(torch.tensor(lats_db[idx], device=device)).cpu().numpy().flatten()
        else:
            mu_n, lv_n = pred(torch.tensor(lats_db[idx], device=device))
            y_n = (mu_n + sm * torch.exp(0.5*lv_n)).cpu().numpy().flatten()
    best_i = np.argmax(y_n); best_frac = fracs[idx[best_i]]; best_y = y_n[best_i]
    rel_diff = abs(best_y - best_score) / (abs(best_score) + 1e-8)
    writer.add_scalar('final/best_y', best_y, 0); writer.add_scalar('final/deviation', rel_diff, 0)
    writer.add_histogram('final/latent_best', best_z.cpu(), 0)

    mon_names = valid_df['name'].tolist(); mask = best_frac > 0
    result = {'z_opt': z_opt_np.flatten(), 'y_opt': best_score, 'best_y': best_y, 'deviation': rel_diff,
              'monomer_indices': np.where(mask)[0].tolist(), 'fractions': best_frac[mask].tolist(),
              'monomer_names': [mon_names[i] for i in np.where(mask)[0]], 'within_tolerance': rel_diff <= tol}
    result.update({'y_decoder': y_dec, 'deviation_decoder': deviation_dec, 'fractions_decoder': fractions_dec,
                    'monomer_names_decoder': [valid_df['name'].iloc[i] for i in np.where(fractions_dec > 1e-6)[0]]})    
    if save_restarts:
        endpoints = np.array([r['z_final'] for r in all_restarts]); y_vals = np.array([r['y_final'] for r in all_restarts])
        if endpoints.shape[1] > 2:
            pca = PCA(n_components=2); endpoints_2d = pca.fit_transform(endpoints)
        else: endpoints_2d = endpoints
        plt.figure(figsize=(10,8)); sc = plt.scatter(endpoints_2d[:,0], endpoints_2d[:,1], c=y_vals, cmap='viridis', s=60, alpha=0.7)
        plt.colorbar(sc, label='Final Y'); plt.title(f'{model} {ptype} restarts'); plt.savefig(f'./data/img/{model}_{ptype}_restarts.png', dpi=150); plt.close()

        if best_traj is not None and len(best_traj) > 1:
            if best_traj.shape[1] > 2:
                pca2 = PCA(n_components=2); traj_2d = pca2.fit_transform(best_traj)
            else: traj_2d = best_traj
            fig, ax = plt.subplots(figsize=(10,8)); scatter = ax.scatter(traj_2d[:,0], traj_2d[:,1], c=best_y_hist, cmap='viridis', s=40, alpha=0.8)
            ax.plot(traj_2d[:,0], traj_2d[:,1], 'gray', alpha=0.5, linestyle='--'); ax.scatter(traj_2d[0,0], traj_2d[0,1], color='red', marker='o', s=120, label='start')
            ax.scatter(traj_2d[-1,0], traj_2d[-1,1], color='blue', marker='*', s=200, label='end'); ax.set_xlabel('PC1'); ax.set_ylabel('PC2'); ax.legend()
            plt.colorbar(scatter, label='Y'); plt.title(f'{model} {ptype} trajectory'); plt.savefig(f'./data/img/{model}_{ptype}_traj.png', dpi=150); plt.close()

            if make_anim and len(traj_2d) > 10:
                fig, ax = plt.subplots(figsize=(8,6)); ax.set_xlim(traj_2d[:,0].min()-0.5, traj_2d[:,0].max()+0.5); ax.set_ylim(traj_2d[:,1].min()-0.5, traj_2d[:,1].max()+0.5)
                line, = ax.plot([], [], 'b-', alpha=0.3); point, = ax.plot([], [], 'ro', markersize=8)
                def init(): line.set_data([], []); point.set_data([], []); return line, point
                def animate(i): line.set_data(traj_2d[:i+1,0], traj_2d[:i+1,1]); point.set_data([traj_2d[i,0]], [traj_2d[i,1]]); return line, point
                ani = animation.FuncAnimation(fig, animate, frames=len(traj_2d), init_func=init, interval=100, blit=True, repeat=False)
                ani.save(f'./data/img/{model}_{ptype}_traj.gif', writer='pillow', fps=10); plt.close()
    writer.close()
    return result

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
models = ['CLS', 'SET', 'DS']
all_results = {}

for model in models:
    for ptype in ['mlp', 'nll']:
        print(f"\n{'='*60}\nProcessing {model} ({ptype.upper()})\n{'='*60}")
        result = run_pipeline(
            model_type=model, ptype=ptype,
            vae_ckpt=f'./data/DL/models/{model}_VAE.pt',
            pred_ckpt=f'./data/DL/models/{model}_{"MLP" if ptype=="mlp" else "MLPNLL"}.pt',
            db_path=f'./data/DL/latent_db_{model}.h5',
            valid_df=valid_df, device=device,
            n_restarts=700, steps=1000, lr=0.03, lam=1e-4,
            sm=1.64 if ptype=="nll" else None,
            n_neighbors=50, tol=0.05, log_every=20,
            save_restarts=True, make_anim=True
        )
        all_results[f"{model}_{ptype}"] = result
        print(f"Best composition for {model} {ptype}:")
        print(f"  Y = {result['best_y']:.4f} (opt: {result['y_opt']:.4f}, deviation: {result['deviation']*100:.2f}%)")
        print(f"  Monomers: {', '.join(result['monomer_names'])}")
        print(f"  Fractions: {[round(f,4) for f in result['fractions']]}")
        if torch.cuda.is_available(): torch.cuda.empty_cache()
        gc.collect()

for ptype in ['mlp', 'nll']:
    df = pd.DataFrame([{
        'model': m.split('_')[0],
        'best_y_nn': r['best_y'],
        'y_opt': r['y_opt'],
        'deviation_nn_%': r['deviation'] * 100,
        'y_decoder': r.get('y_decoder', np.nan),
        'deviation_decoder_%': r.get('deviation_decoder', np.nan) * 100,
        'within_tolerance': r['within_tolerance'],
        'monomers_nn': ' + '.join(r['monomer_names']),
        'fractions_nn': str([round(f,4) for f in r['fractions']]),
        'monomers_dec': ' + '.join(r.get('monomer_names_decoder', [])),
        'fractions_dec': str([round(f,4) for f in r.get('fractions_decoder', [])])
    } for m, r in all_results.items() if m.endswith(ptype)])
    df.to_csv(f'./data/DL/model_comparison_{ptype.upper()}.csv', index=False)
    print(f"\n{'='*60}\nMODEL COMPARISON ({ptype.upper()}) – including decoder\n{'='*60}")
    print(df.to_string(index=False))
```

## Optimizer on MLL fast

```python
class HighScoreOptimizer:
    def __init__(self, predictor, emp_mean, emp_latents, lr=0.03, lambda_reg=0.0001):
        self.predictor = predictor.eval()
        self.emp_mean = torch.tensor(emp_mean, device=next(predictor.parameters()).device)
        self.emp_latents = torch.tensor(emp_latents, device=next(predictor.parameters()).device)
        self.lr = lr
        self.lambda_reg = lambda_reg
        self.n_init = len(emp_latents)

    def optimize(self, n_restarts=None, steps=1000, log_every=20, save_all_trajectories=False, tb_writer=None, tag='opt'):
        if n_restarts is None:
            n_restarts = self.n_init
        device = next(self.predictor.parameters()).device

        best_score = -float('inf')
        best_z = None
        best_traj = None
        best_y_hist = None
        best_reg_hist = None
        all_restarts = []

        for restart in range(n_restarts):
            base = self.emp_latents[np.random.randint(0, self.n_init)]
            noise = torch.randn(self.emp_latents.shape[1], device=device) * 0.05 * self.emp_mean.std()
            z = (base + noise).clone().requires_grad_(True)
            opt = optim.Adam([z], lr=self.lr)

            traj_z = []
            traj_y = []
            traj_reg = []

            for step in range(steps):
                opt.zero_grad()
                y = self.predictor(z.unsqueeze(0)).squeeze()
                lam = self.lambda_reg * max(0, 1 - step / steps)
                reg = lam * (z - self.emp_mean).pow(2).mean()
                loss = -y + reg
                loss.backward()
                opt.step()

                if step % log_every == 0 or step == steps - 1:
                    traj_z.append(z.detach().cpu().numpy().copy())
                    traj_y.append(y.item())
                    traj_reg.append(reg.item())

            with torch.no_grad():
                final_score = self.predictor(z.unsqueeze(0)).item()

            restart_info = {
                'z_final': z.detach().cpu().numpy().copy(),
                'y_final': final_score,
                'trajectory': np.array(traj_z),
                'y_history': np.array(traj_y),
                'reg_history': np.array(traj_reg)
            }
            all_restarts.append(restart_info)

            if final_score > best_score:
                best_score = final_score
                best_z = z.detach().clone()
                best_traj = np.array(traj_z)
                best_y_hist = np.array(traj_y)
                best_reg_hist = np.array(traj_reg)

        if tb_writer is not None and best_traj is not None:
            tb_writer.add_scalar(f'{tag}/best_score', best_score, 0)
            for step, (y_val, reg_val) in enumerate(zip(best_y_hist, best_reg_hist)):
                tb_writer.add_scalar(f'{tag}/y_best', y_val, step * log_every)
                tb_writer.add_scalar(f'{tag}/reg_best', reg_val, step * log_every)
            tb_writer.add_histogram(f'{tag}/latent_best', best_z.cpu(), 0)

        return best_z, best_score, best_traj, best_y_hist, best_reg_hist, all_restarts
```

```python
def run_optimization_pipeline(
    model_type,
    vae_checkpoint,
    predictor_checkpoint,
    db_path,
    valid_df,
    device=None,
    n_restarts=700,
    steps=1000,
    lr=0.03,
    lambda_reg=0.0001,
    n_neighbors=50,
    tolerance=0.05,
    log_every=20,
    save_restarts=True,
    make_animation=False
):
    if device is None:
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    writer = SummaryWriter(log_dir=f"runs/{model_type}_optimizer_MLP")

    vae = load_pretrained_vae(model_type, vae_checkpoint, device)
    predictor = MLP(latent_dim=64).to(device)
    predictor.load_state_dict(torch.load(predictor_checkpoint, map_location=device))
    predictor.eval()

    emp_mean, real_latents = get_real_latents(vae)

    del vae
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    optimizer = HighScoreOptimizer(predictor, emp_mean, real_latents, lr, lambda_reg)
    best_z, best_score, best_traj, best_y_hist, best_reg_hist, all_restarts = optimizer.optimize(
        n_restarts=n_restarts, steps=steps, log_every=log_every,
        save_all_trajectories=save_restarts, tb_writer=writer, tag=model_type
    )

    print(f"Optimal Y: {best_score:.4f}")

    with h5py.File(db_path, 'r') as f:
        if 'latent' in f:
            latents = f['latent'][:]
        elif 'mu' in f:
            latents = f['mu'][:]
        else:
            raise KeyError("No latent dataset found in HDF5 file (expected 'latent' or 'mu')")
        
        fracs = f['fractions'][:]

    z_opt_np = best_z.cpu().numpy().reshape(1, -1)
    dists = cdist(z_opt_np, latents).flatten()
    neighbor_indices = np.argsort(dists)[:n_neighbors]

    neighbor_mu = torch.tensor(latents[neighbor_indices], device=device)
    with torch.no_grad():
        y_neighbors = predictor(neighbor_mu).cpu().numpy().flatten()

    best_local_idx = np.argmax(y_neighbors)
    best_frac = fracs[neighbor_indices[best_local_idx]]   # вектор долей
    best_y = y_neighbors[best_local_idx]

    mask = best_frac > 0
    best_mono_indices = np.where(mask)[0]
    best_frac_values = best_frac[mask]

    rel_diff = abs(best_y - best_score) / (abs(best_score) + 1e-8)
    writer.add_scalar('final/best_y', best_y, 0)
    writer.add_scalar('final/deviation', rel_diff, 0)
    writer.add_histogram('final/latent_best', best_z.cpu(), 0)

    mon_names = valid_df['name'].tolist()
    result = {
        'z_opt': z_opt_np.flatten(),
        'y_opt': best_score,
        'best_y': best_y,
        'deviation': rel_diff,
        'monomer_indices': best_mono_indices.tolist(),
        'fractions': best_frac_values.tolist(),
        'monomer_names': [mon_names[i] for i in best_mono_indices],
        'within_tolerance': rel_diff <= tolerance
    }

    if save_restarts:
        endpoints = np.array([r['z_final'] for r in all_restarts])
        y_vals = np.array([r['y_final'] for r in all_restarts])
        if endpoints.shape[1] > 2:
            pca = PCA(n_components=2)
            endpoints_2d = pca.fit_transform(endpoints)
        else:
            endpoints_2d = endpoints
        plt.figure(figsize=(10,8))
        sc = plt.scatter(endpoints_2d[:,0], endpoints_2d[:,1], c=y_vals, cmap='viridis', s=60, alpha=0.7)
        plt.colorbar(sc, label='Final Y')
        plt.title(f'{model_type} restarts endpoints')
        plt.savefig(f'./data/img/{model_type}_restarts.png', dpi=150)
        plt.close()

        if best_traj is not None and len(best_traj) > 1:
            if best_traj.shape[1] > 2:
                pca2 = PCA(n_components=2)
                traj_2d = pca2.fit_transform(best_traj)
            else:
                traj_2d = best_traj
            fig, ax = plt.subplots(figsize=(10,8))
            scatter = ax.scatter(traj_2d[:,0], traj_2d[:,1], c=best_y_hist, cmap='viridis', s=40, alpha=0.8)
            ax.plot(traj_2d[:,0], traj_2d[:,1], 'gray', alpha=0.5, linestyle='--')
            ax.scatter(traj_2d[0,0], traj_2d[0,1], color='red', marker='o', s=120, label='start')
            ax.scatter(traj_2d[-1,0], traj_2d[-1,1], color='blue', marker='*', s=200, label='end')
            ax.set_xlabel('PC1'); ax.set_ylabel('PC2')
            ax.legend()
            plt.colorbar(scatter, label='Y')
            plt.title(f'{model_type} best trajectory')
            plt.savefig(f'./data/img/{model_type}_trajectory_2d.png', dpi=150)
            plt.close()

            if make_animation and len(traj_2d) > 10:
                fig, ax = plt.subplots(figsize=(8,6))
                ax.set_xlim(traj_2d[:,0].min()-0.5, traj_2d[:,0].max()+0.5)
                ax.set_ylim(traj_2d[:,1].min()-0.5, traj_2d[:,1].max()+0.5)
                line, = ax.plot([], [], 'b-', alpha=0.3)
                point, = ax.plot([], [], 'ro', markersize=8)
                def init(): line.set_data([], []); point.set_data([], []); return line, point
                def animate(i):
                    line.set_data(traj_2d[:i+1,0], traj_2d[:i+1,1])
                    point.set_data([traj_2d[i,0]], [traj_2d[i,1]])
                    return line, point
                ani = animation.FuncAnimation(fig, animate, frames=len(traj_2d), init_func=init, interval=100, blit=True, repeat=False)
                ani.save(f'./data/img/{model_type}_trajectory.gif', writer='pillow', fps=10)
                plt.close()

    writer.close()
    return result
```

```python
def get_real_latents(vae, h5_path='./data/DL/small_poly_datasets.h5', device=None, batch_size=64):
    if device is None:
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    vae.eval()
    
    all_mu = []
    with h5py.File(h5_path, 'r') as f:
        total = f['cffp_256'].shape[0]
        for start in range(0, total, batch_size):
            end = min(start + batch_size, total)
            ffp = torch.from_numpy(f['cffp_256'][start:end]).float().to(device)
            auto = torch.from_numpy(f['autocorr'][start:end]).float().to(device)
            with torch.no_grad():
                mu, _ = vae.encode(ffp, auto)
            all_mu.append(mu.cpu().numpy())
            del ffp, auto, mu
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
    mu_all = np.concatenate(all_mu, axis=0)
    return mu_all.mean(axis=0), mu_all
```

## Optimizer on MLL NLL

```python
class HighScoreOptimizer:
    def __init__(self, predictor, emp_mean, emp_latents, lr=0.03, lambda_reg=0.0001, sigma_multiplier=1.64):
        self.predictor = predictor.eval()
        self.emp_mean = torch.tensor(emp_mean, device=next(predictor.parameters()).device)
        self.emp_latents = torch.tensor(emp_latents, device=next(predictor.parameters()).device)
        self.lr = lr
        self.lambda_reg = lambda_reg
        self.sigma_multiplier = sigma_multiplier
        self.n_init = len(emp_latents)

    def optimize(self, n_restarts=None, steps=1000, log_every=20, save_all_trajectories=False, tb_writer=None, tag='opt'):
        if n_restarts is None:
            n_restarts = self.n_init
        device = next(self.predictor.parameters()).device

        best_score = -float('inf')
        best_z = None
        best_traj = None
        best_y_hist = None
        best_reg_hist = None
        all_restarts = []

        for restart in range(n_restarts):
            base = self.emp_latents[np.random.randint(0, self.n_init)]
            noise = torch.randn(self.emp_latents.shape[1], device=device) * 0.05 * self.emp_mean.std()
            z = (base + noise).clone().requires_grad_(True)
            opt = optim.Adam([z], lr=self.lr)

            traj_z = []
            traj_score = []
            traj_reg = []

            for step in range(steps):
                opt.zero_grad()
                mu, log_var = self.predictor(z.unsqueeze(0))
                sigma = torch.exp(0.5 * log_var)  
                y = mu + self.sigma_multiplier * sigma
                y = y.squeeze()
                reg = self.lambda_reg * (
                    (z - self.emp_mean).pow(2).mean() + 
                    0.5 * torch.exp(log_var).mean()
                )
                loss = -y + reg
                loss.backward()
                opt.step()

                if step % log_every == 0 or step == steps - 1:
                    traj_z.append(z.detach().cpu().numpy().copy())
                    traj_score.append(y.item())
                    traj_reg.append(reg.item())

            with torch.no_grad():
                mu_final, log_var_final = self.predictor(z.unsqueeze(0))
                sigma_final = torch.exp(0.5 * log_var_final)
                final_score = (mu_final + self.sigma_multiplier * sigma_final).item()

            restart_info = {
                'z_final': z.detach().cpu().numpy().copy(),
                'y_final': final_score,
                'trajectory': np.array(traj_z),
                'score_history': np.array(traj_score),
                'reg_history': np.array(traj_reg)
            }
            all_restarts.append(restart_info)

            if final_score > best_score:
                best_score = final_score
                best_z = z.detach().clone()
                best_traj = np.array(traj_z)
                best_y_hist = np.array(traj_score)
                best_reg_hist = np.array(traj_reg)

        if tb_writer is not None and best_traj is not None:
            tb_writer.add_scalar(f'{tag}/best_score', best_score, 0)
            for step, (y_val, reg_val) in enumerate(zip(best_y_hist, best_reg_hist)):
                tb_writer.add_scalar(f'{tag}/score_best', y_val, step * log_every)
                tb_writer.add_scalar(f'{tag}/reg_best', reg_val, step * log_every)
            tb_writer.add_histogram(f'{tag}/latent_best', best_z.cpu(), 0)

        return best_z, best_score, best_traj, best_y_hist, best_reg_hist, all_restarts
```

```python
def run_optimization_pipeline(
    model_type,
    vae_checkpoint,
    predictor_checkpoint,
    db_path,
    valid_df,
    device=None,
    n_restarts=700,
    steps=1000,
    lr=0.03,
    lambda_reg=0.0001,
    sigma_multiplier=1.64,
    n_neighbors=50,
    tolerance=0.05,
    log_every=20,
    save_restarts=True,
    make_animation=False,
    input_dim=64,
    hidden_dims=(512, 256, 128, 64),
    dropout=0.25
):
    if device is None:
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    writer = SummaryWriter(log_dir=f"runs/{model_type}_optimizer_MLPNLL")

    vae = load_pretrained_vae(model_type, vae_checkpoint, device)
    
    predictor = MLPNLL().to(device)
    predictor.load_state_dict(torch.load(predictor_checkpoint, map_location=device))
    predictor.eval()

    emp_mean, real_latents = get_real_latents(vae)

    del vae
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    optimizer = HighScoreOptimizer(
        predictor, emp_mean, real_latents, lr, lambda_reg, sigma_multiplier
    )
    best_z, best_score, best_traj, best_score_hist, best_reg_hist, all_restarts = optimizer.optimize(
        n_restarts=n_restarts, steps=steps, log_every=log_every,
        save_all_trajectories=save_restarts, tb_writer=writer, tag=model_type
    )

    print(f"Optimal (μ + {sigma_multiplier}σ): {best_score:.4f}")

    with h5py.File(db_path, 'r') as f:
        if 'latent' in f:
            latents = f['latent'][:]
        elif 'mu' in f:
            latents = f['mu'][:]
        else:
            raise KeyError("No latent dataset found in HDF5 file (expected 'latent' or 'mu')")
        fracs = f['fractions'][:]

    z_opt_np = best_z.cpu().numpy().reshape(1, -1)
    dists = cdist(z_opt_np, latents).flatten()
    neighbor_indices = np.argsort(dists)[:n_neighbors]

    neighbor_mu = torch.tensor(latents[neighbor_indices], device=device)
    with torch.no_grad():
        mu_neighbors, log_var_neighbors = predictor(neighbor_mu)
        sigma_neighbors = torch.exp(0.5 * log_var_neighbors)
        y_neighbors = (mu_neighbors + sigma_multiplier * sigma_neighbors).cpu().numpy().flatten()

    best_local_idx = np.argmax(y_neighbors)
    best_frac = fracs[neighbor_indices[best_local_idx]]
    best_y = y_neighbors[best_local_idx]

    rel_diff = abs(best_y - best_score) / (abs(best_score) + 1e-8)
    writer.add_scalar('final/best_y', best_y, 0)
    writer.add_scalar('final/deviation', rel_diff, 0)
    writer.add_histogram('final/latent_best', best_z.cpu(), 0)

    mon_names = valid_df['name'].tolist()
    mask = best_frac > 0
    best_mono_indices = np.where(mask)[0]
    best_frac_values = best_frac[mask]

    result = {
        'z_opt': z_opt_np.flatten(),
        'y_opt': best_score,
        'best_y': best_y,
        'deviation': rel_diff,
        'monomer_indices': best_mono_indices.tolist(),
        'fractions': best_frac_values.tolist(),
        'monomer_names': [mon_names[i] for i in best_mono_indices],
        'within_tolerance': rel_diff <= tolerance
    }

    if save_restarts:
        endpoints = np.array([r['z_final'] for r in all_restarts])
        y_vals = np.array([r['y_final'] for r in all_restarts])
        if endpoints.shape[1] > 2:
            pca = PCA(n_components=2)
            endpoints_2d = pca.fit_transform(endpoints)
        else:
            endpoints_2d = endpoints
        plt.figure(figsize=(10,8))
        sc = plt.scatter(endpoints_2d[:,0], endpoints_2d[:,1], c=y_vals, cmap='viridis', s=60, alpha=0.7)
        plt.colorbar(sc, label='Final Y')
        plt.title(f'{model_type} restarts endpoints')
        plt.savefig(f'./data/img/{model_type}_restarts.png', dpi=150)
        plt.close()

        if best_traj is not None and len(best_traj) > 1:
            if best_traj.shape[1] > 2:
                pca2 = PCA(n_components=2)
                traj_2d = pca2.fit_transform(best_traj)
            else:
                traj_2d = best_traj
            fig, ax = plt.subplots(figsize=(10,8))
            scatter = ax.scatter(traj_2d[:,0], traj_2d[:,1], c=best_score_hist, cmap='viridis', s=40, alpha=0.8)
            ax.plot(traj_2d[:,0], traj_2d[:,1], 'gray', alpha=0.5, linestyle='--')
            ax.scatter(traj_2d[0,0], traj_2d[0,1], color='red', marker='o', s=120, label='start')
            ax.scatter(traj_2d[-1,0], traj_2d[-1,1], color='blue', marker='*', s=200, label='end')
            ax.set_xlabel('PC1'); ax.set_ylabel('PC2')
            ax.legend()
            plt.colorbar(scatter, label='Y')
            plt.title(f'{model_type} best trajectory')
            plt.savefig(f'./data/img/{model_type}_trajectory_2d.png', dpi=150)
            plt.close()

            if make_animation and len(traj_2d) > 10:
                fig, ax = plt.subplots(figsize=(8,6))
                ax.set_xlim(traj_2d[:,0].min()-0.5, traj_2d[:,0].max()+0.5)
                ax.set_ylim(traj_2d[:,1].min()-0.5, traj_2d[:,1].max()+0.5)
                line, = ax.plot([], [], 'b-', alpha=0.3)
                point, = ax.plot([], [], 'ro', markersize=8)
                def init(): line.set_data([], []); point.set_data([], []); return line, point
                def animate(i):
                    line.set_data(traj_2d[:i+1,0], traj_2d[:i+1,1])
                    point.set_data([traj_2d[i,0]], [traj_2d[i,1]])
                    return line, point
                ani = animation.FuncAnimation(fig, animate, frames=len(traj_2d), init_func=init, interval=100, blit=True, repeat=False)
                ani.save(f'./data/img/{model_type}_trajectory.gif', writer='pillow', fps=10)
                plt.close()

    writer.close()
    return result
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

models = ['CLS', 'SET', 'DS']
all_results = {}

for model in models:
    print(f"\n{'='*60}\nProcessing {model}\n{'='*60}")
    
    result = run_optimization_pipeline(
        model_type=model,
        vae_checkpoint=f'./data/DL/models/{model}_VAE.pt',
        predictor_checkpoint=f'./data/DL/models/{model}_MLPNLL.pt',
        db_path=f'./data/DL/latent_db_{model}.h5',
        valid_df=valid_df,
        device=device,
        n_restarts=700,
        steps=1000,
        lr=0.03,
        lambda_reg=0.0001,
        sigma_multiplier=1.64,
        n_neighbors=50,
        tolerance=0.05,
        log_every=20,
        save_restarts=True,
        make_animation=True,
        input_dim=64,                
        hidden_dims=(512, 256, 128, 64),
        dropout=0.25
    )
    
    all_results[model] = result
    
    print(f"Best composition for {model}:")
    print(f"  Y = {result['best_y']:.4f} (opt: {result['y_opt']:.4f}, deviation: {result['deviation']*100:.2f}%)")
    print(f"  Monomers: {', '.join(result['monomer_names'])}")
    print(f"  Fractions: {[round(f,4) for f in result['fractions']]}")
    
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    gc.collect()

df_results = pd.DataFrame([{
    'model': m,
    'best_y': r['best_y'],
    'y_opt': r['y_opt'],
    'deviation_%': r['deviation'] * 100,
    'within_tolerance': r['within_tolerance'],
    'monomers': ' + '.join(r['monomer_names']),
    'fractions': str([round(f,4) for f in r['fractions']])
} for m, r in all_results.items()])

df_results.to_csv('./data/DL/model_comparison_NLL.csv', index=False)

print("\n" + "="*60)
print("MODEL COMPARISON (with variance penalization)")
print("="*60)
print(df_results.to_string(index=False))

best_model = df_results.loc[df_results['best_y'].idxmax()]
print(f"\nBest model: {best_model['model']} with Y = {best_model['best_y']:.4f}")
```

## MLP fast_test

```python
class MLP(nn.Module):
    def __init__(self, latent_dim=64, hidden_dims=(128,64,32), dropout=0.2):
        super().__init__()
        layers = []
        in_dim = latent_dim
        for h in hidden_dims:
            layers += [nn.Linear(in_dim,h), nn.LayerNorm(h), nn.Mish(), nn.Dropout(dropout)]
            in_dim = h
        layers.append(nn.Linear(in_dim,1))
        self.network = nn.Sequential(*layers)

    def forward(self, z):
        if z.ndim == 3: z = z.mean(dim=1)
        return self.network(z).squeeze(-1)

class FullPredictivePipeline(nn.Module):
    def __init__(self, pretrained_vae, latent_dim=64):
        super().__init__()
        self.vae = pretrained_vae
        self.vae.eval()
        for p in self.vae.parameters(): p.requires_grad = False
        self.predictor = MLP(latent_dim=latent_dim)

    def forward(self, ffp, auto):
        with torch.no_grad():
            mu, _ = self.vae.encode(ffp, auto)
        return self.predictor(mu)
```

```python
class Dataset(Dataset):
    def __init__(self, h5_file, csv_file):
        with h5py.File(h5_file, "r") as f:
            self.ffp = f["cffp_256"][:].astype(np.float32)
            self.auto = f["autocorr"][:].astype(np.float32)

        df = pd.read_csv(csv_file)
        self.y = df["Glass (kPa)_max"].values.astype(np.float32)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return (
            torch.from_numpy(self.ffp[idx]),
            torch.from_numpy(self.auto[idx]),
            torch.tensor(self.y[idx], dtype=torch.float32),
        )
```

```python
dataset = Dataset(
    "./data/DL/small_poly_datasets.h5",
    "./data/df_341.csv",
)

train_size = int(0.8 * len(dataset))
val_size = len(dataset) - train_size

train_dataset, val_dataset = random_split(
    dataset,
    [train_size, val_size],
    generator=torch.Generator().manual_seed(42),
)

train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    shuffle=True,
    num_workers=2,
)

val_loader = DataLoader(
    val_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=2,
)
```

```python
predictor_train_config = {
    "batch_size": 32,
    "epochs": 500,
    "lr": 1e-3,
    "weight_decay": 1e-4,
    "patience": 20,
}

vae_checkpoints = {
    "CLS": "./data/DL/models/CLS_VAE.pt",
    "SET": "./data/DL/models/SET_VAE.pt",
    "DS":  "./data/DL/models/DS_VAE.pt",
}
```

```python
def train_predictor(model, train_loader, val_loader, epochs, lr, weight_decay, patience, device, model_name):
    writer = SummaryWriter(log_dir=f"runs/{model_name}_MLP")
    criterion = nn.MSELoss()
    optimizer = torch.optim.AdamW(model.predictor.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=10)
    best_loss = float('inf'); no_improve = 0; history = {'train_loss':[], 'val_loss':[]}

    for epoch in range(1, epochs+1):
        model.train(); train_loss = 0.0
        for ffp, auto, y in train_loader:
            ffp, auto, y = ffp.to(device), auto.to(device), y.to(device)
            optimizer.zero_grad()
            loss = criterion(model(ffp, auto), y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.predictor.parameters(), 1.0)
            optimizer.step()
            train_loss += loss.item() * y.size(0)
        train_loss /= len(train_loader.dataset)
        model.eval(); val_loss = 0.0
        with torch.no_grad():
            for ffp, auto, y in val_loader:
                ffp, auto, y = ffp.to(device), auto.to(device), y.to(device)
                val_loss += criterion(model(ffp, auto), y).item() * y.size(0)
        val_loss /= len(val_loader.dataset)

        scheduler.step(val_loss)
        history['train_loss'].append(train_loss); history['val_loss'].append(val_loss)
        current_lr = optimizer.param_groups[0]['lr']
        writer.add_scalar('Loss/Train', train_loss, epoch)
        writer.add_scalar('Loss/Validation', val_loss, epoch)
        writer.add_scalar('Learning_Rate', current_lr, epoch)

        if epoch % 5 == 0 or epoch == 1:
            for name, param in model.predictor.named_parameters():
                writer.add_histogram(f'Weights/{name}', param.data, epoch)
                if param.grad is not None:
                    writer.add_histogram(f'Gradients/{name}', param.grad.data, epoch)

        print(f"{model_name} | Epoch {epoch:03d} | Train {train_loss:.5f} | Val {val_loss:.5f} | LR {current_lr:.2e}")
        if val_loss < best_loss:
            best_loss = val_loss; no_improve = 0
            torch.save(model.predictor.state_dict(), f"./data/DL/models/{model_name}_MLP.pt")
        else:
            no_improve += 1
            if no_improve >= patience:
                print(f"Early stopping ({model_name})"); break

    writer.close()
    return history
```

```python
histories = {}

for name, checkpoint in vae_checkpoints.items():

    print(f"\n{'='*60}")
    print(f"Training adhesion predictor using {name} encoder")
    print(f"{'='*60}")

    vae = load_pretrained_vae(
        model_type=name,
        checkpoint_path=checkpoint,
        device=device,
    )

    model = FullPredictivePipeline(
        pretrained_vae=vae,
        latent_dim=64,
    ).to(device)

    histories[name] = train_predictor(
        model=model,
        train_loader=train_loader,
        val_loader=val_loader,
        epochs=predictor_train_config["epochs"],
        lr=predictor_train_config["lr"],
        weight_decay=predictor_train_config["weight_decay"],
        patience=predictor_train_config["patience"],
        device=device,
        
        model_name=name,
    )

    del model
    del vae

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    gc.collect()
```

## MLP NLL

```python
def read_sheet(path, cols):
    return pd.read_excel(path, sheet_name=0, header=None, skiprows=1) \
             .iloc[:, :len(cols)].set_axis(cols, axis=1)

def parse_std(s):
    s = str(s).replace(',', '.').strip()
    m = re.search(r'([\d.]+)\s*±\s*([\d.]+)', s)
    if m: return float(m.group(1)), float(m.group(2))
    try: return float(s), float(s) * 0.07
    except: return np.nan, np.nan

def load_and_prepare(data_dir='data'):
    c6 = ['Round','ML','Rank','No.','ΦHEA','ΦBA','ΦCBEA','ΦATAC','ΦPEA','ΦAAm','Fa_max','Fa_ave','d','Q']
    c5 = ['No.','ΦHEA','ΦBA','ΦCBEA','ΦATAC','ΦPEA','ΦAAm','Fa_max','Fa_ave','d','Q','G','Tanδ','k']
    cc = ['ΦHEA','ΦBA','ΦCBEA','ΦATAC','ΦPEA','ΦAAm']

    df = pd.concat([
        read_sheet(f'{data_dir}/41586_2025_9269_MOESM5_ESM.xlsx', c5),
        read_sheet(f'{data_dir}/41586_2025_9269_MOESM6_ESM.xlsx', c6)
    ], ignore_index=True)[cc + ['Fa_max','Fa_ave']]

    for c in cc + ['Fa_max']:
        df[c] = pd.to_numeric(df[c].astype(str).str.replace(',', '.'), errors='coerce')

    df = df.dropna(subset=cc)
    df = df[~df[cc].round(4).duplicated()].reset_index(drop=True)

    df[cc] = df[cc].div(df[cc].sum(axis=1), axis=0)

    parsed = df['Fa_ave'].apply(parse_std)
    df['Fa_mean'] = parsed.apply(lambda x: x[0])
    df['Fa_std']  = parsed.apply(lambda x: x[1])

    mask = df['Fa_mean'].isna() & df['Fa_max'].notna()
    df.loc[mask, 'Fa_mean'] = df.loc[mask, 'Fa_max']
    df.loc[mask, 'Fa_std']  = df.loc[mask, 'Fa_mean'] * 0.05

    df = df[df['Fa_mean'] > 0].copy()
    return df.reset_index(drop=True), cc
```

```python
def split_data(df, test_size=0.2, seed=42, n_bins=10):
    bins = pd.qcut(df['Fa_mean'], q=n_bins, labels=False, duplicates='drop')
    tr, va = next(StratifiedShuffleSplit(n_splits=1, test_size=test_size, random_state=seed).split(df, bins))
    return df.iloc[tr].reset_index(drop=True), df.iloc[va].reset_index(drop=True)

def get_latent_for_composition(fracs, vae, mon_indices, features, device):
    exact = fracs * 1000
    counts = np.floor(exact).astype(np.int32)
    counts[np.argsort(exact % 1)[::-1][:1000 - counts.sum()]] += 1
    chain = np.repeat(mon_indices, counts)
    np.random.shuffle(chain)
    with torch.no_grad():
        auto = torch.from_numpy(features['autocorr'][0][chain]).float().unsqueeze(0).to(device)
        cffp = torch.from_numpy(features['cffp_256'][0][chain]).float().unsqueeze(0).to(device)
        mu, _ = vae.encode(cffp, auto)
    return mu.squeeze(0).cpu().numpy()

def augment_and_create_tensors(df, latent_col, seed=42, n_aug=30):
    rng = np.random.default_rng(seed)
    z = np.repeat(np.stack(df[latent_col].values), n_aug, axis=0)
    m = np.repeat(df['Fa_mean'].values, n_aug)
    s = np.repeat(np.maximum(df['Fa_std'].values, 0.01), n_aug)
    y = np.clip(rng.normal(m, s), 0, None).astype(np.float32)
    return z.astype(np.float32), y
```

```python
class MLPNLL(nn.Module):
    def __init__(self, input_dim=64, hidden_dims=(512, 256, 128, 64, 32), dropout=0.25):
        super().__init__()
        layers = []
        in_dim = input_dim
        for h in hidden_dims[:-1]:
            layers += [nn.Linear(in_dim, h), nn.LayerNorm(h), nn.Mish(), nn.Dropout(dropout)]
            in_dim = h
        self.backbone = nn.Sequential(*layers)
        self.shared_last = nn.Sequential(
            nn.Linear(in_dim, hidden_dims[-1]),
            nn.LayerNorm(hidden_dims[-1]),
            nn.Mish(),
            nn.Dropout(dropout/2) 
        )
        self.mu_head = nn.Linear(hidden_dims[-1], 1)
        self.var_head = nn.Linear(hidden_dims[-1], 1)
        
        self._init_weights()
    
    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Linear):
                nn.init.kaiming_normal_(m.weight, nonlinearity='relu')
                if m.bias is not None:
                    nn.init.zeros_(m.bias)
        with torch.no_grad():
            self.mu_head.bias.data.fill_(100.0)          #  Fa ~100
            self.var_head.bias.data.fill_(np.log(100.0)) # σ ≈ 10, log_var ≈ 4.6
    
    def forward(self, z):
        if z.ndim == 3:
            z = z.mean(dim=1)
        h = self.backbone(z)
        h = self.shared_last(h)
        mu = self.mu_head(h).squeeze(-1)
        log_var = self.var_head(h).squeeze(-1)
        log_var = torch.clamp(log_var, min=-5, max=10)
        return mu, log_var

def gaussian_nll_loss(mu, log_var, target):
    var = torch.exp(log_var) + 1e-6
    loss = 0.5 * ((target - mu) ** 2 / var + log_var)
    return loss.mean()
```

```python
def train_predictor(model, train_loader, val_loader, epochs, lr, weight_decay, patience, device, model_name):
    writer = SummaryWriter(log_dir=f"runs/{model_name}_MLPNLL")

    optimizer = optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=15, min_lr=1e-6)

    best_loss = float('inf')
    best_mse = float('inf')
    no_improve = 0
    history = {'train_loss': [], 'val_loss': [], 'val_mse': []}

    for epoch in range(1, epochs + 1):
        model.train()
        train_loss = 0.0
        for z, y in train_loader:
            z, y = z.to(device), y.to(device)
            optimizer.zero_grad()
            mu, log_var = model(z)
            loss = gaussian_nll_loss(mu, log_var, y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_loss += loss.item() * y.size(0)
        train_loss /= len(train_loader.dataset)

        model.eval()
        val_loss = 0.0
        val_mse = 0.0
        with torch.no_grad():
            for z, y in val_loader:
                z, y = z.to(device), y.to(device)
                mu, log_var = model(z)
                val_loss += gaussian_nll_loss(mu, log_var, y).item() * y.size(0)
                val_mse += torch.mean((mu - y) ** 2).item() * y.size(0)
        val_loss /= len(val_loader.dataset)
        val_mse /= len(val_loader.dataset)

        scheduler.step(val_loss)
        current_lr = optimizer.param_groups[0]['lr']

        history['train_loss'].append(train_loss)
        history['val_loss'].append(val_loss)
        history['val_mse'].append(val_mse)

        writer.add_scalar('Loss/Train', train_loss, epoch)
        writer.add_scalar('Loss/Validation', val_loss, epoch)
        writer.add_scalar('MSE/Validation', val_mse, epoch)
        writer.add_scalar('Learning_Rate', current_lr, epoch)

        if epoch % 5 == 0 or epoch == 1:
            for name, param in model.named_parameters():
                writer.add_histogram(f'Weights/{name}', param.data, epoch)
                if param.grad is not None:
                    writer.add_histogram(f'Gradients/{name}', param.grad.data, epoch)

        print(f"{model_name} | Epoch {epoch:03d} | Train {train_loss:.5f} | Val NLL {val_loss:.5f} | Val MSE {val_mse:.1f} | LR {current_lr:.2e}")

        if val_loss < best_loss:
            best_loss = val_loss
            best_mse = val_mse
            no_improve = 0
            torch.save(model.state_dict(), f"./data/DL/models/{model_name}_MLPNLL.pt")
        else:
            no_improve += 1
            if no_improve >= patience:
                print(f"Early stopping ({model_name})")
                break

    writer.close()
    print(f"\nBest Val NLL: {best_loss:.5f}, Corresponding Val MSE: {best_mse:.1f}")
    return history, best_loss, best_mse
```

```python
def generate_pairs(df, latent_col='latent', target_col='Fa_mean'):
    pairs = []
    n = len(df)
    for i in range(n):
        for j in range(i+1, n):
            fa_i = df.iloc[i][target_col]
            fa_j = df.iloc[j][target_col]
            pairs.append({
                'z_A': df.iloc[i][latent_col],
                'z_B': df.iloc[j][latent_col],
                'label': 1.0 if fa_i > fa_j else 0.0
            })
    return pd.DataFrame(pairs)
```

```python
class PairwiseDataset(Dataset):
    def __init__(self, pairs_df):
        self.z_A = torch.tensor(np.stack(pairs_df['z_A'].values), dtype=torch.float32)
        self.z_B = torch.tensor(np.stack(pairs_df['z_B'].values), dtype=torch.float32)
        self.label = torch.tensor(pairs_df['label'].values, dtype=torch.float32)

    def __len__(self):
        return len(self.label)

    def __getitem__(self, idx):
        return self.z_A[idx], self.z_B[idx], self.label[idx]
```

```python
class PairwiseSiameseMLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.tower = nn.Sequential(
            nn.Linear(64, 32), nn.Mish(), nn.Dropout(0.2),
            nn.Linear(32, 16), nn.Mish(),nn.Dropout(0.2),
            nn.Linear(16, 16), nn.Mish(), nn.Dropout(0.2)
        )
        self.head = nn.Sequential(
            nn.Linear(32, 16), nn.Mish(), nn.Dropout(0.2),
            nn.Linear(16, 8), nn.Mish(), nn.Dropout(0.2),
            nn.Linear(8, 1)
        )

    def forward(self, x_A, x_B):
        h_A = self.tower(x_A)
        h_B = self.tower(x_B)
        diff = h_A - h_B
        merged = torch.cat([diff, torch.abs(diff)], dim=-1)
        return self.head(merged).squeeze(-1)
```

```python
def train_pairwise(model, train_loader, val_loader, epochs, lr, weight_decay, patience, device, model_name):
    writer = SummaryWriter(log_dir=f"runs/{model_name}_Pairwise")
    criterion = nn.BCEWithLogitsLoss()
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=10)

    best_loss = float('inf')
    best_auc = 0.0
    no_improve = 0
    history = {'train_loss': [], 'val_loss': [], 'val_auc': [], 'val_acc': []}

    for epoch in range(1, epochs + 1):
        model.train()
        train_loss = 0.0
        for z_A, z_B, y in train_loader:
            z_A, z_B, y = z_A.to(device), z_B.to(device), y.to(device)
            optimizer.zero_grad()
            logits = model(z_A, z_B)
            loss = criterion(logits, y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_loss += loss.item() * y.size(0)
        train_loss /= len(train_loader.dataset)
        model.eval()
        val_loss = 0.0
        val_preds, val_y = [], []
        with torch.no_grad():
            for z_A, z_B, y in val_loader:
                z_A, z_B, y = z_A.to(device), z_B.to(device), y.to(device)
                logits = model(z_A, z_B)
                val_loss += criterion(logits, y).item() * y.size(0)
                val_preds.append(torch.sigmoid(logits).cpu().numpy())
                val_y.append(y.cpu().numpy())
        val_loss /= len(val_loader.dataset)
        val_preds = np.concatenate(val_preds)
        val_y = np.concatenate(val_y)
        val_auc = roc_auc_score(val_y, val_preds)
        val_acc = accuracy_score(val_y, (val_preds > 0.5).astype(float))

        scheduler.step(val_loss)
        current_lr = optimizer.param_groups[0]['lr']
        history['train_loss'].append(train_loss)
        history['val_loss'].append(val_loss)
        history['val_auc'].append(val_auc)
        history['val_acc'].append(val_acc)

        writer.add_scalar('Loss/Train', train_loss, epoch)
        writer.add_scalar('Loss/Validation', val_loss, epoch)
        writer.add_scalar('Metrics/Val_AUC', val_auc, epoch)
        writer.add_scalar('Metrics/Val_Acc', val_acc, epoch)
        writer.add_scalar('Learning_Rate', current_lr, epoch)

        if epoch % 5 == 0 or epoch == 1:
            for name, param in model.named_parameters():
                writer.add_histogram(f'Weights/{name}', param.data, epoch)
                if param.grad is not None:
                    writer.add_histogram(f'Gradients/{name}', param.grad.data, epoch)

        print(f"{model_name} | Epoch {epoch:03d} | Train Loss {train_loss:.5f} | Val Loss {val_loss:.5f} | AUC {val_auc:.4f} | Acc {val_acc:.4f} | LR {current_lr:.2e}")
        if val_loss < best_loss:
            best_loss = val_loss
            best_auc = val_auc
            no_improve = 0
            torch.save(model.state_dict(), f"./data/DL/models/{model_name}_Pairwise.pt")
        else:
            no_improve += 1
            if no_improve >= patience:
                print(f"Early stopping ({model_name}) at epoch {epoch}")
                break

    writer.close()
    print(f"\nBest Val Loss: {best_loss:.5f} | Best AUC: {best_auc:.4f}")
    return history
```

## Sort Discrete (quick)

```python
def load_pairwise_model(model_path, device):
    model = PSMLP()
    state_dict = torch.load(model_path, map_location=device)
    model.load_state_dict(state_dict)
    model.to(device)
    model.eval()
    return model
```

```python
def pairwise_sort(arr, model, device, batch_size=2048, threshold=32, cache_limit=10_000_000):
    model.eval()
    indexed_arr = [(i, obj) for i, obj in enumerate(arr)]
    random.shuffle(indexed_arr)
    n = len(indexed_arr)
    cache = OrderedDict()
    pbar = tqdm(desc="Sorting (Stack size)", unit=" steps")

    def set_cache(id_a, id_b, val):
        if len(cache) >= cache_limit:
            cache.popitem(last=False)
        cache[(id_a, id_b)] = val

    def get_cache(id_a, id_b):
        key = (id_a, id_b)
        if key in cache:
            return cache[key], True
        rev = (id_b, id_a)
        if rev in cache:
            return not cache[rev], True
        return None, False

    def compare_idx(i, j):
        if i == j:
            return False
        id_a, a = indexed_arr[i]
        id_b, b = indexed_arr[j]
        val, found = get_cache(id_a, id_b)
        if found:
            return val
        a_t = torch.from_numpy(a).float().unsqueeze(0).to(device)
        b_t = torch.from_numpy(b).float().unsqueeze(0).to(device)
        with torch.no_grad():
            logit = model(a_t, b_t)
            logit_val = logit.item() if logit.dim() == 0 else logit.view(-1)[0].item() 
        res = logit_val > 0
        set_cache(id_a, id_b, res)
        return res

    def compare_batch(current_elements, pivot_obj):
        if not current_elements:
            return []
        id_p, pivot = pivot_obj
        results = [None] * len(current_elements)
        uncached_indices = []
        
        for idx, (id_o, _) in enumerate(current_elements):
            val, found = get_cache(id_o, id_p)
            if found:
                results[idx] = val
            else:
                uncached_indices.append(idx)
                
        if uncached_indices:
            for start in range(0, len(uncached_indices), batch_size):
                batch_idxs = uncached_indices[start : start + batch_size]
                left_objs = [current_elements[i][1] for i in batch_idxs]
                left = torch.stack([torch.from_numpy(o).float() for o in left_objs]).to(device)
                pivot_t = torch.from_numpy(pivot).float().to(device).unsqueeze(0)
                right = pivot_t.repeat(len(batch_idxs), 1)
                with torch.no_grad():
                    logits = model(left, right)
                if logits.dim() > 1:
                    logits = logits.squeeze(-1)
                res_list = (logits > 0).cpu().numpy().tolist()
                if not isinstance(res_list, list):
                    res_list = [res_list]
                for idx, res in zip(batch_idxs, res_list):
                    id_o, _ = current_elements[idx]
                    set_cache(id_o, id_p, res)
                    results[idx] = res
            
        return results
    stack = [(0, n - 1)]

    while stack:
        pbar.set_description(f"Sorting (Stack size: {len(stack)})")
        pbar.update(1)
        lo, hi = stack.pop()
        if hi - lo < threshold:
            for i in range(lo + 1, hi + 1):
                j = i
                while j > lo and compare_idx(j, j - 1):
                    indexed_arr[j], indexed_arr[j - 1] = indexed_arr[j - 1], indexed_arr[j]
                    j -= 1
            continue
        mid = (lo + hi) // 2
        if compare_idx(mid, lo): indexed_arr[lo], indexed_arr[mid] = indexed_arr[mid], indexed_arr[lo]
        if compare_idx(hi, lo):  indexed_arr[lo], indexed_arr[hi]  = indexed_arr[hi],  indexed_arr[lo]
        if compare_idx(hi, mid): indexed_arr[mid], indexed_arr[hi] = indexed_arr[hi], indexed_arr[mid]
        indexed_arr[mid], indexed_arr[hi] = indexed_arr[hi], indexed_arr[mid]
        pivot_obj = indexed_arr[hi]
        elements_to_compare = indexed_arr[lo:hi] 
        better = compare_batch(elements_to_compare, pivot_obj)
        left_vals = [elem for elem, ok in zip(elements_to_compare, better) if ok]
        right_vals = [elem for elem, ok in zip(elements_to_compare, better) if not ok]
        indexed_arr[lo : hi + 1] = left_vals + [pivot_obj] + right_vals
        new_pivot = lo + len(left_vals)
        if lo < new_pivot - 1:
            stack.append((lo, new_pivot - 1))
        if new_pivot + 1 < hi:
            stack.append((new_pivot + 1, hi))

    pbar.close()
    return [obj for _, obj in indexed_arr]
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = MLPNLL().to(device)
model.load_state_dict(torch.load('./data/DL/models/CLS_MLPNLL.pt', map_location=device))
model.eval()
h5_path = './data/DL/latent_db_CLS.h5'
batch_size, BETA = 10000, 1.0
with h5py.File(h5_path, 'r') as f:
    mu_features = f['mu']
    n_total = mu_features.shape[0]
    scores = np.zeros(n_total, dtype=np.float32)
    sigmas = np.zeros(n_total, dtype=np.float32)
    SC = np.zeros(n_total, dtype=np.float32)
    for start in tqdm(range(0, n_total, batch_size), desc="Predicting MLPNLL"):
        end = min(start + batch_size, n_total)
        z_batch = torch.tensor(mu_features[start:end], dtype=torch.float32, device=device)
        with torch.no_grad():
            pred_mu, pred_logvar = model(z_batch)
            pred_std = torch.exp(0.5 * pred_logvar)
            scores[start:end] = (pred_mu - BETA * pred_std).cpu().numpy()
            SC[start:end] = pred_mu.cpu().numpy()
            sigmas[start:end] = pred_std.cpu().numpy()
sorted_indices = np.argsort(scores)[::-1]
combined = np.column_stack((sorted_indices, SC[sorted_indices], sigmas[sorted_indices]))
np.save('./data/DL/sorted_indices_with_scores_MLPNLL_CLS.npy', combined)
```
